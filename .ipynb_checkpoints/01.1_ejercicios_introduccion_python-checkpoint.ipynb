### Ejercicio 4.4 — Sets para deduplicar
Dado el texto `texto = 'hoy hoy es un buen buen día'`, obtén las palabras únicas como un set y muéstralas ordenadas alfabéticamente.

Un `set` es una colección de elementos únicos y desordenados. Es perfecto para eliminar duplicados de una lista de forma eficiente. Como los sets no garantizan un orden, para mostrarlos de forma ordenada, puedes usar la función `sorted()`, que devuelve una nueva lista ordenada con los elementos del set.

````python
# Tu código aquí
texto = 'hoy hoy es un buen buen día'
palabras = texto.split()
print(palabras)
unicas = set(palabras)
print(unicas)
print(sorted(unicas))
````

### Ejercicio 4.5 — Combinado (lista + dict + set)
Dada `compras = ['pan', 'leche', 'pan', 'huevos']` y `precios = {'pan':1.0,'leche':1.2,'huevos':2.0}`, calcula el total gastado considerando cada artículo una sola vez (usa un set).

````python
# Tu código aquí
compras = ['pan', 'leche', 'pan', 'huevos']
precios = {'pan':1.0,'leche':1.2,'huevos':2.0}
articulos = set(compras)  # elementos únicos
print(articulos)
total = 0.0
for art in articulos:
    print(art)
    total += precios[art]
print(f"El total de la compra es {total:.2f}€")  # debería ser 4.2
````

<details>
<summary>💡 Solución</summary>


### Ejercicio 6.3 — Retornos múltiples
Implementa `estadisticas(nums)` que devuelva una tupla `(minimo, maximo, media)` de una lista de números.

````python
# Tu código aquí
def estadisticas(nums):
    minimo = min(nums)
    maximo = max(nums)
    media = sum(nums) / len(nums)
    return minimo, maximo, media

print(estadisticas([1,2,3,4, 23, 4, 3, 0]))  # (1,4,2.5)
````

### Ejercicio 7.2 — Métodos que modifican estado
Añade a `Persona` un método `cumpleanios()` que incremente `edad` en 1 y un método `es_mayor_de_edad()` que devuelva True si `edad` >= 18.

En este ejercicio, vamos a ampliar la clase `Persona` que creamos antes. La idea es añadirle dos comportamientos (métodos) nuevos que interactúen con los datos del objeto:

1.  **`cumpleanios()`**: Este método simulará que la persona cumple años. Cada vez que se llame a este método sobre un objeto `Persona`, su atributo `edad` deberá aumentar en 1. Este es un ejemplo de un método que **modifica el estado** del objeto, es decir, cambia el valor de uno de sus atributos internos (`self.edad`).

2.  **`es_mayor_de_edad()`**: Este método comprobará si la persona es mayor de edad. Devolverá `True` si la `edad` es 18 o superior, y `False` en caso contrario. A diferencia del anterior, este método **lee el estado** del objeto (`self.edad`) para realizar una comprobación, pero no lo modifica.

El objetivo es que puedas crear un objeto `Persona` con una edad inicial, llamar al método `cumpleanios()` para envejecerlo, y luego usar `es_mayor_de_edad()` para verificar si ha alcanzado la mayoría de edad.

````python
# Tu código aquí (puedes reutilizar la clase anterior)
class Persona:
    def __init__(self, nombre, edad): # Aquí va el constructor, que establece los atributos
        ...
    def presentarse(self): # Aquí va el método que presenta a la persona
        ...
    def cumpleanios(self): # Aquí va el método que incrementa la edad
        ...
    def es_mayor_de_edad(self): # Aquí va el método que comprueba si es mayor de edad
        ...

p = Persona('Luis', 17)
p.cumpleanios()
print(p.edad, p.es_mayor_de_edad())  # 18 True
````

<details>
<summary>💡 Solución</summary>

```python
class Persona:
    def __init__(self, nombre, edad):
        self.nombre = nombre
        self.edad = edad
    def presentarse(self):
        return f'Soy {self.nombre} y tengo {self.edad} años'
    def cumpleanios(self):
        self.edad += 1
    def es_mayor_de_edad(self):
        return self.edad >= 18

p = Persona('Luis', 17)
p.cumpleanios()
print(p.edad, p.es_mayor_de_edad())
```

</details>

### Ejercicio 7.3 — Clase CuentaBancaria
Crea `CuentaBancaria` con `titular` y `saldo` (inicia en 0), métodos `depositar(cantidad)` y `retirar(cantidad)` (no permitir saldo negativo). Muestra el saldo tras algunas operaciones.

````python
# Tu código aquí
class CuentaBancaria:
    def __init__(self, titular):
        ...
    def depositar(self, cantidad):
        ...
    def retirar(self, cantidad):
        ...

c = CuentaBancaria('Ana')
c.depositar(100)
c.retirar(30)
print(c.saldo)  # 70
````

<details>
<summary>💡 Solución</summary>

```python
class CuentaBancaria:
    def __init__(self, titular):
        self.titular = titular
        self.saldo = 0.0
    def depositar(self, cantidad):
        self.saldo += cantidad
    def retirar(self, cantidad):
        if cantidad > self.saldo:
            raise ValueError('Fondos insuficientes')
        self.saldo -= cantidad

c = CuentaBancaria('Ana')
c.depositar(100)
c.retirar(30)
print(c.saldo)
```

</details>

````python
# Tu código aquí (puedes reutilizar la clase)
class CuentaBancaria:
    def __init__(self, titular):
        ...
    def __str__(self):
        ...

c = CuentaBancaria('Ana')
print(c)
````

<details>
<summary>💡 Solución</summary>

```python
class CuentaBancaria:
    def __init__(self, titular):
        self.titular = titular
        self.saldo = 0.0
```

<summary>💡 Solución</summary>

```python
class Persona:
    def __init__(self, nombre, edad):
        self.nombre = nombre
        self.edad = edad
    def presentarse(self):
        return f'Soy {self.nombre} y tengo {self.edad} años'

class Estudiante(Persona):
    def __init__(self, nombre, edad, curso):
        super().__init__(nombre, edad)
        self.curso = curso
    def presentarse(self):
        base = super().presentarse()
        return base + f' y estudio {self.curso}'

e = Estudiante('Sofía', 22, 'ML')
print(e.presentarse())
```

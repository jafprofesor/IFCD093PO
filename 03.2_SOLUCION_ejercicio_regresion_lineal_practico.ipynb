## 📋 Solución Completa del Ejercicio

### PASO 1️⃣: Importar Librerías - SOLUCIÓN

````python
# SOLUCIÓN
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn import metrics

print("✅ Todas las librerías importadas correctamente")
````

### PASO 2️⃣: Crear el Dataset - SOLUCIÓN

````python
# SOLUCIÓN
datos = {
    'experiencia': [1.5, 3.2, 4.8, 5.5, 7.2, 8.1, 9.5, 10.3, 12.1, 14.8],
    'salario': [40.5, 60.8, 82.3, 88.9, 110.5, 125.3, 138.2, 155.7, 175.3, 198.5]
}

df_salarios = pd.DataFrame(datos)

print("✅ Dataset creado correctamente")
print(f"\nTamaño del dataset: {len(df_salarios)} empleados")
display(df_salarios.head())
````

### PASO 3️⃣: Visualizar los Datos - SOLUCIÓN

````python
# SOLUCIÓN
plt.figure(figsize=(10, 6))
plt.scatter(df_salarios['experiencia'], df_salarios['salario'], s=100, alpha=0.6, color='blue')
plt.xlabel('Años de Experiencia', fontsize=12)
plt.ylabel('Salario (€1000)', fontsize=12)
plt.title('Relación entre Experiencia y Salario', fontsize=14, fontweight='bold')
plt.grid(True, alpha=0.3)
plt.show()

print("✅ Gráfico de dispersión creado")
print("\n📊 Interpretación: Se ve una clara relación lineal positiva.")
print("   Cuando aumenta la experiencia, aumenta el salario de forma proporcional.")
````

### PASO 4️⃣: Preparar Datos (X e y) - SOLUCIÓN

````python
# SOLUCIÓN
X = df_salarios[['experiencia']]  # Dobles corchetes: devuelve DataFrame
y = df_salarios['salario']         # Un solo corchete: devuelve Serie

print("✅ Datos preparados correctamente")
print(f"\nForma de X: {X.shape}")
print(f"Forma de y: {y.shape}")
print(f"\nPrimeras filas de X:\n{X.head()}")
print(f"\nPrimeros valores de y:\n{y.head()}")
````

### PASO 5️⃣: Dividir en Entrenamiento y Prueba - SOLUCIÓN

````python
# SOLUCIÓN
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)

print("✅ Datos divididos correctamente")
print(f"\nTamaño conjunto entrenamiento: {len(X_train)} ejemplos (70%)")
print(f"Tamaño conjunto prueba: {len(X_test)} ejemplos (30%)")
print(f"\nTotal: {len(X_train) + len(X_test)} ejemplos")
````

### PASO 6️⃣: Crear y Entrenar el Modelo - SOLUCIÓN

````python
# SOLUCIÓN
modelo = LinearRegression()
modelo.fit(X_train, y_train)

print("✅ Modelo entrenado correctamente")
print(f"\n📊 PARÁMETROS DEL MODELO:")
print(f"━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━")
print(f"Coeficiente (pendiente): {modelo.coef_[0]:.4f}")
print(f"Intercepto: {modelo.intercept_:.4f}")
print(f"━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━")
print(f"\n📐 ECUACIÓN DE LA RECTA:")
print(f"Salario = {modelo.intercept_:.2f} + {modelo.coef_[0]:.2f} × Experiencia")
print(f"\n💡 Interpretación:")
print(f"   - Por cada año de experiencia adicional, el salario aumenta en {modelo.coef_[0]:.2f}K€")
print(f"   - Un empleado sin experiencia tendría salario base de {modelo.intercept_:.2f}K€")
````

### PASO 7️⃣: Hacer Predicciones - SOLUCIÓN

````python
# SOLUCIÓN
y_pred_train = modelo.predict(X_train)
y_pred_test = modelo.predict(X_test)

print("✅ Predicciones realizadas correctamente")
print(f"\nPrimeras 5 predicciones en ENTRENAMIENTO:")
for i in range(5):
    print(f"  Exp: {X_train.iloc[i,0]:.1f} años → Salario Real: {y_train.iloc[i]:.1f}K€, Predicho: {y_pred_train[i]:.1f}K€")

print(f"\nPrimeras 3 predicciones en PRUEBA:")
for i in range(3):
    print(f"  Exp: {X_test.iloc[i,0]:.1f} años → Salario Real: {y_test.iloc[i]:.1f}K€, Predicho: {y_pred_test[i]:.1f}K€")
````

### PASO 8️⃣: Evaluar el Modelo - SOLUCIÓN

````python
# SOLUCIÓN
mae = metrics.mean_absolute_error(y_test, y_pred_test)
mse = metrics.mean_squared_error(y_test, y_pred_test)
rmse = np.sqrt(mse)
r2 = modelo.score(X_test, y_test)

print("📊 MÉTRICAS DE EVALUACIÓN:")
print(f"━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━")
print(f"MAE (Error Absoluto Medio): {mae:.4f} (en miles de €)")
print(f"MSE (Error Cuadrático Medio): {mse:.4f}")
print(f"RMSE (Raíz del Error Cuadrático): {rmse:.4f} (en miles de €)")
print(f"R² (Coeficiente de Determinación): {r2:.4f}")
print(f"━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━")

print(f"\n✅ INTERPRETACIÓN:")
print(f"   - MAE: En promedio, el modelo se equivoca en ±{mae:.2f}K€")
print(f"   - RMSE: La desviación típica de los errores es {rmse:.2f}K€")
print(f"   - R²: El modelo explica el {r2*100:.1f}% de la variabilidad de los salarios")

if r2 >= 0.95:
    print(f"\n🌟 CALIDAD DEL MODELO: EXCELENTE")
elif r2 >= 0.80:
    print(f"\n⭐ CALIDAD DEL MODELO: MUY BUENO")
elif r2 >= 0.60:
    print(f"\n✅ CALIDAD DEL MODELO: BUENO")
else:
    print(f"\n⚠️ CALIDAD DEL MODELO: PODRÍA MEJORAR")
````

### PASO 9️⃣: Visualizar Resultados - SOLUCIÓN

````python
# SOLUCIÓN
plt.figure(figsize=(12, 6))

# Graficar los datos reales del conjunto de prueba
plt.scatter(X_test, y_test, color='blue', s=100, label='Valores Reales', alpha=0.6)

# Graficar la línea de regresión
plt.plot(X_test, y_pred_test, color='red', linewidth=2.5, label='Predicción del Modelo')

# Añadir también los datos de entrenamiento en gris para contexto
plt.scatter(X_train, y_train, color='gray', s=50, label='Datos de Entrenamiento', alpha=0.3)

plt.xlabel('Años de Experiencia', fontsize=12)
plt.ylabel('Salario (€1000)', fontsize=12)
plt.title('Regresión Lineal: Predicción de Salarios', fontsize=14, fontweight='bold')
plt.legend(fontsize=11)
plt.grid(True, alpha=0.3)
plt.show()

print("✅ Gráfico creado correctamente")
print("\n📈 Se puede ver que:")
print("   - Los puntos azules (reales) están muy cerca de la línea roja (predicción)")
print("   - Esto indica que el modelo hace predicciones muy precisas")
````

## 🎓 RESPUESTAS A LAS PREGUNTAS DE REFLEXIÓN

### 1. ¿Qué significa el coeficiente (pendiente)?

**Respuesta:**
- El coeficiente es **{:.2f}K€ por año** (aproximadamente)
- Esto significa que por cada año adicional de experiencia, el salario predicho aumenta en esa cantidad
- Es una **relación lineal**: si duplicas los años de experiencia, duplicas (aproximadamente) el incremento de salario

### 2. ¿Qué significa el intercepto?

**Respuesta:**
- El intercepto es **{:.2f}K€** (aproximadamente)
- Técnicamente, es el salario predicho cuando alguien tiene 0 años de experiencia
- En la práctica, esto no es muy realista (un empleado sin experiencia no gana eso)
- El intercepto es principalmente útil para ajustar la línea de regresión a los datos

### 3. ¿Es bueno el modelo?

**Respuesta:**
- R² = **{:.4f}** (aproximadamente)
- Si R² > 0.95: **EXCELENTE** - El modelo explica más del 95% de la variabilidad
- Si R² > 0.80: **MUY BUENO** - El modelo explica más del 80% de la variabilidad
- Si R² > 0.60: **BUENO** - El modelo explica más del 60% de la variabilidad
- En este caso, el modelo es muy preciso

### 4. ¿Cuál fue el error medio (MAE)?

**Respuesta:**
- MAE = **{:.2f}K€** (aproximadamente)
- Esto significa que, en promedio, las predicciones se equivocan en ±{:.2f}K€
- Considerando que los salarios oscilan entre 40K€ y 200K€, un error de {:.2f}K€ es **muy pequeño** (menos del 5%)

### 5. ¿Sería recomendable usar este modelo en producción?

**Respuesta:**
- **SÍ, podría usarse** porque:
  1. El R² es muy alto (>0.95), indica excelente ajuste
  2. El MAE es bajo, las predicciones son precisas
  3. La relación es claramente lineal
  4. El modelo es simple e interpretable

- **Pero debería considerarse**:
  1. ¿Hay otros factores que influyen en el salario? (puesto, ubicación, formación, etc.)
  2. ¿Los datos son actuales? (la relación podría cambiar con el tiempo)
  3. ¿El dataset es representativo? (solo 10 empleados es pequeño)
  4. ¿Qué pasa con valores extremos? (empleados con 20+ años de experiencia)

## 📝 NOTAS PEDAGÓGICAS 

### Errores Comunes de los Alumnos:

1. **Error: Usar un solo corchete para X**
   ```python
   # ❌ INCORRECTO
   X = df['experiencia']  # Devuelve una Serie, no un DataFrame
   
   # ✅ CORRECTO
   X = df[['experiencia']]  # Devuelve un DataFrame
   ```

2. **Error: Olvidar hacer reshape() para predicciones individuales**
   ```python
   # ❌ INCORRECTO
   salario = modelo.predict(5.0)  # Falla porque necesita 2D
   
   # ✅ CORRECTO
   salario = modelo.predict([[5.0]])  # Debe ser una lista de listas
   ```

3. **Confusión: Qué es train_test_split()**
   - Explanation: Divide aleatoriamente los datos 70-30 (o como especifiques)
   - Los datos de prueba nunca se ven durante el entrenamiento
   - Esto simula "datos nuevos" para evaluar el modelo

4. **Confusión: Las diferencias entre MAE, MSE, RMSE**
   - MAE: Promedio de |errores| - Fácil de interpretar
   - MSE: Promedio de errores² - Penaliza errores grandes
   - RMSE: Raíz de MSE - Está en las mismas unidades que el objetivo

5. **Confusión: Qué significan los parámetros del modelo**
   - Coef (slope): Cambio en Y por cada unidad de X
   - Intercept: Valor de Y cuando X = 0
   - El modelo es: Y = intercept + coef × X

```python
# Importar las librerías necesarias
import matplotlib.pyplot as plt
import numpy as np

# Datos para el gráfico
x = np.linspace(0, 10, 100)
y1 = np.sin(x)

# Crear el primer gráfico
# ------------------------------
# figura_original es el objeto figura
# ax1 es el objeto de los ejes
figura_original, ax1 = plt.subplots() # Crear una nueva figura y un conjunto de subgráficos
ax1.plot(x, y1, 'b-', label='Seno') 
ax1.set_title('Gráfico de Seno')
ax1.set_xlabel('Eje X')
ax1.set_ylabel('Eje Y')
ax1.legend()
plt.show() 
```

```python
# Datos para el segundo gráfico
y2 = np.cos(x)

# Nueva figura y un conjunto de subgráficos
# ------------------------------
# figura_nueva es el objeto figura
# ax2 es el objeto de los ejes
figura_nueva, ax2 = plt.subplots()
ax2.plot(x, y2, 'r--', label='Coseno')
ax2.set_title('Gráfico de Coseno')
ax2.set_xlabel('Eje X')
ax2.set_ylabel('Eje Y')
ax2.legend()
```

```python
# Mostrar ambas figuras una debajo de la otra
from IPython.display import display

display(figura_original)
display(figura_nueva)

```

## 🔧 1. Álgebra Básica <a id='1-algebra'></a>

El álgebra es la base de todo. Necesitamos entender cómo manipular números y variables.

### 1.1 Operaciones Fundamentales

Las operaciones básicas que usaremos constantemente:

````python
# Operaciones básicas en Python

# Suma y resta
a = 10
b = 3
suma = a + b
resta = a - b
print(f"Suma: {suma}, Resta: {resta}")

# Multiplicación y división
multiplicacion = a * b
division = a / b
print(f"Multiplicación: {multiplicacion}, División: {division:.2f}")

# Potencias y raíces
potencia = a ** 2  # a al cuadrado
raiz = a ** 0.5    # raíz cuadrada de a
print(f"Potencia: {potencia}, Raíz: {raiz:.2f}")
````

### 1.2 Ecuaciones Lineales

Una ecuación lineal tiene la forma: **y = mx + b**

- **m**: pendiente (cuánto sube o baja)
- **b**: intercepto (punto donde cruza el eje Y)

**Ejemplo del mundo real:** El costo de un taxi
- Tarifa base (b) = 3€
- Costo por kilómetro (m) = 1.5€
- Ecuación: `costo = 1.5 × km + 3`

````python
import matplotlib.pyplot as plt
import numpy as np

# Ejemplo de ecuación lineal: costo del taxi
def costo_taxi(km):
    """Calcula el costo del taxi según kilómetros"""
    m = 1.5  # costo por km
    b = 3    # tarifa base
    return m * km + b

# Calcular costos para diferentes distancias
kilometros = np.arange(0, 20, 1)
costos = costo_taxi(kilometros)

# Visualizar
plt.figure(figsize=(10, 6))
plt.plot(kilometros, costos, 'b-', linewidth=2)
plt.scatter([5, 10, 15], [costo_taxi(5), costo_taxi(10), costo_taxi(15)], 
            color='red', s=100, zorder=5)
plt.xlabel('Kilómetros', fontsize=12)
plt.ylabel('Costo (€)', fontsize=12)
plt.title('Costo del Taxi: Ecuación Lineal y = 1.5x + 3', fontsize=14, fontweight='bold')
plt.grid(True, alpha=0.3)
plt.show()

# Ejemplos prácticos
print("Costos del taxi:")
print(f"  5 km → {costo_taxi(5):.2f}€")
print(f" 10 km → {costo_taxi(10):.2f}€")
print(f" 15 km → {costo_taxi(15):.2f}€")
````

## 📊 2. Estadística Descriptiva <a id='2-estadistica'></a>

La estadística nos ayuda a **resumir y entender datos**. Son las herramientas que más usarás en Machine Learning.

### 2.1 Medidas de Tendencia Central

Nos dicen "dónde está el centro" de nuestros datos:

- **Media (promedio)**: Suma de todos los valores / cantidad de valores
- **Mediana**: Valor del medio cuando ordenamos los datos
- **Moda**: Valor que más se repite

````python
# Ejemplo: Notas de un examen
notas = [5, 7, 8, 7, 9, 6, 7, 10, 8, 7]

# Media (promedio)
media = sum(notas) / len(notas)
print(f"Media: {media:.2f}")

# Mediana (valor del medio)
notas_ordenadas = sorted(notas)
n = len(notas_ordenadas)
if n % 2 == 0:
    mediana = (notas_ordenadas[n//2 - 1] + notas_ordenadas[n//2]) / 2
else:
    mediana = notas_ordenadas[n//2]
print(f"Mediana: {mediana}")

# Moda (valor más frecuente)
from collections import Counter
contador = Counter(notas)
moda = contador.most_common(1)[0][0]
print(f"Moda: {moda}")

print(f"\nNotas: {notas}")
print(f"Notas ordenadas: {notas_ordenadas}")
````

### 2.2 Medidas de Dispersión

Nos dicen "qué tan dispersos están los datos":

- **Rango**: diferencia entre el máximo y el mínimo
- **Varianza**: promedio de las distancias al cuadrado desde la media
- **Desviación estándar**: raíz cuadrada de la varianza (más interpretable)

````python
# Dos grupos de estudiantes con la misma media pero diferente dispersión
grupo_A = [7, 7, 7, 7, 7, 7, 7, 7, 7, 7]  # Todos sacan lo mismo
grupo_B = [2, 4, 5, 6, 7, 8, 9, 9, 10, 10]  # Notas muy variadas

def calcular_estadisticas(datos, nombre):
    media = np.mean(datos)
    varianza = np.var(datos)
    desv_std = np.std(datos)
    
    print(f"\n{nombre}:")
    print(f"  Datos: {datos}")
    print(f"  Media: {media:.2f}")
    print(f"  Varianza: {varianza:.2f}")
    print(f"  Desviación estándar: {desv_std:.2f}")

calcular_estadisticas(grupo_A, "Grupo A (homogéneo)")
calcular_estadisticas(grupo_B, "Grupo B (heterogéneo)")

# Visualización
plt.figure(figsize=(12, 5))

plt.subplot(1, 2, 1)
plt.hist(grupo_A, bins=10, edgecolor='black', alpha=0.7, color='blue')
plt.axvline(np.mean(grupo_A), color='red', linestyle='--', linewidth=2, label='Media')
plt.title('Grupo A: Baja dispersión', fontweight='bold')
plt.xlabel('Notas')
plt.ylabel('Frecuencia')
plt.legend()

plt.subplot(1, 2, 2)
plt.hist(grupo_B, bins=10, edgecolor='black', alpha=0.7, color='green')
plt.axvline(np.mean(grupo_B), color='red', linestyle='--', linewidth=2, label='Media')
plt.title('Grupo B: Alta dispersión', fontweight='bold')
plt.xlabel('Notas')
plt.ylabel('Frecuencia')
plt.legend()

plt.tight_layout()
plt.show()
````

## 🎲 3. Probabilidad Básica <a id='3-probabilidad'></a>

La probabilidad mide **la posibilidad de que algo ocurra**. Va de 0 (imposible) a 1 (seguro).

### 3.1 Conceptos Fundamentales

- **Probabilidad = Casos favorables / Casos posibles**
- Ejemplo: Al lanzar un dado, P(sacar 6) = 1/6 ≈ 0.167

````python
# Simulación: Lanzar un dado 1000 veces
import random

lanzamientos = [random.randint(1, 6) for _ in range(1000)]

# Contar cuántas veces salió cada número
frecuencias = Counter(lanzamientos)

print("Resultados de 1000 lanzamientos:")
for numero in sorted(frecuencias.keys()):
    cuenta = frecuencias[numero]
    probabilidad = cuenta / 1000
    print(f"  {numero}: {cuenta} veces ({probabilidad:.3f} ≈ {1/6:.3f} teórico)")

# Visualización
plt.figure(figsize=(10, 6))
numeros = list(frecuencias.keys())
cuentas = list(frecuencias.values())
plt.bar(numeros, cuentas, color='skyblue', edgecolor='black')
plt.axhline(1000/6, color='red', linestyle='--', linewidth=2, label='Esperado (166.67)')
plt.xlabel('Número del dado', fontsize=12)
plt.ylabel('Frecuencia', fontsize=12)
plt.title('Distribución de 1000 lanzamientos de dado', fontsize=14, fontweight='bold')
plt.legend()
plt.grid(axis='y', alpha=0.3)
plt.show()
````

### 3.2 Distribución Normal (Campana de Gauss)

La **distribución más importante** en estadística. Muchos fenómenos naturales la siguen:
- Alturas de personas
- Errores de medición
- Notas de exámenes (cuando hay muchos estudiantes)

**Regla 68-95-99.7**:
- 68% de los datos están a ±1 desviación estándar de la media
- 95% están a ±2 desviaciones estándar
- 99.7% están a ±3 desviaciones estándar

````python
# Generar datos con distribución normal
# Ejemplo: Alturas de personas (media=170cm, desv_std=10cm)
np.random.seed(42)
alturas = np.random.normal(loc=170, scale=10, size=1000)

# Estadísticas
media_alturas = np.mean(alturas)
std_alturas = np.std(alturas)

print(f"Media: {media_alturas:.2f} cm")
print(f"Desviación estándar: {std_alturas:.2f} cm")

# Visualización
plt.figure(figsize=(12, 6))
plt.hist(alturas, bins=50, density=True, alpha=0.7, color='lightblue', edgecolor='black')

# Curva teórica
x = np.linspace(140, 200, 100)
from scipy.stats import norm
plt.plot(x, norm.pdf(x, media_alturas, std_alturas), 'r-', linewidth=2, label='Distribución Normal')

# Líneas de desviación estándar
plt.axvline(media_alturas, color='green', linestyle='--', linewidth=2, label='Media')
plt.axvline(media_alturas + std_alturas, color='orange', linestyle='--', alpha=0.7, label='±1 σ')
plt.axvline(media_alturas - std_alturas, color='orange', linestyle='--', alpha=0.7)

plt.xlabel('Altura (cm)', fontsize=12)
plt.ylabel('Densidad', fontsize=12)
plt.title('Distribución Normal: Alturas de 1000 personas', fontsize=14, fontweight='bold')
plt.legend()
plt.grid(True, alpha=0.3)
plt.show()

# Porcentajes en rangos
dentro_1std = np.sum((alturas >= media_alturas - std_alturas) & 
                     (alturas <= media_alturas + std_alturas)) / len(alturas) * 100
print(f"\nDatos dentro de ±1σ: {dentro_1std:.1f}% (teórico: 68%)")
````

## 🔢 4. Álgebra Lineal <a id='4-algebra-lineal'></a>

El álgebra lineal es **fundamental para Machine Learning**. Todos los algoritmos usan vectores y matrices.

### 4.1 Vectores

Un vector es una **lista de números**. Piénsalo como:
- Coordenadas en un mapa (x, y)
- Características de una casa (metros², habitaciones, precio)
- Píxeles de una imagen

````python
# Vectores como arrays de NumPy
vector_a = np.array([3, 4])
vector_b = np.array([1, 2])

print(f"Vector A: {vector_a}")
print(f"Vector B: {vector_b}")

# Operaciones con vectores
suma = vector_a + vector_b
resta = vector_a - vector_b
multiplicacion_escalar = 2 * vector_a

print(f"\nA + B = {suma}")
print(f"A - B = {resta}")
print(f"2 * A = {multiplicacion_escalar}")

# Magnitud (longitud) de un vector
magnitud_a = np.linalg.norm(vector_a)
print(f"\nMagnitud de A: {magnitud_a:.2f}")

# Producto punto (dot product)
dot_product = np.dot(vector_a, vector_b)
print(f"A · B = {dot_product}")
````

````python
# Visualización de vectores
plt.figure(figsize=(10, 8))

# Origen
origen = np.array([0, 0])

# Dibujar vectores
plt.quiver(*origen, *vector_a, angles='xy', scale_units='xy', scale=1, color='blue', width=0.01, label='Vector A')
plt.quiver(*origen, *vector_b, angles='xy', scale_units='xy', scale=1, color='red', width=0.01, label='Vector B')
plt.quiver(*origen, *suma, angles='xy', scale_units='xy', scale=1, color='green', width=0.01, label='A + B')

# Configuración
plt.xlim(-1, 5)
plt.ylim(-1, 7)
plt.xlabel('X', fontsize=12)
plt.ylabel('Y', fontsize=12)
plt.title('Vectores en 2D', fontsize=14, fontweight='bold')
plt.grid(True, alpha=0.3)
plt.legend()
plt.axhline(0, color='black', linewidth=0.5)
plt.axvline(0, color='black', linewidth=0.5)
plt.show()
````

### 4.2 Matrices

Una matriz es una **tabla de números**. En ML:
- Cada fila = una observación (ej: una persona)
- Cada columna = una característica (ej: edad, altura, peso)

````python
# Matriz: datos de 3 personas
# Columnas: [Edad, Altura (cm), Peso (kg)]
personas = np.array([
    [25, 170, 70],  # Persona 1
    [30, 165, 65],  # Persona 2
    [22, 180, 85]   # Persona 3
])

print("Matriz de datos:")
print(personas)
print(f"\nForma de la matriz: {personas.shape} (3 personas, 3 características)")

# Acceder a elementos
print(f"\nPersona 1: {personas[0]}")
print(f"Todas las alturas: {personas[:, 1]}")

# Operaciones con matrices
matriz_A = np.array([[1, 2], [3, 4]])
matriz_B = np.array([[5, 6], [7, 8]])

print(f"\nMatriz A:")
print(matriz_A)
print(f"\nMatriz B:")
print(matriz_B)

# Multiplicación de matrices
producto = np.dot(matriz_A, matriz_B)
print(f"\nA × B:")
print(producto)
````

## 📈 5. Cálculo Básico <a id='5-calculo'></a>

El cálculo nos ayuda a entender **cómo cambian las cosas**. En ML, usamos derivadas para **optimizar** nuestros modelos.

### 5.1 Derivadas: ¿Qué son?

La derivada mide **la tasa de cambio**:
- ¿Qué tan rápido sube una función?
- ¿Dónde está el máximo o mínimo?

**Ejemplo práctico**: Si conduces un coche:
- **Posición** = dónde estás
- **Velocidad** = derivada de la posición (qué tan rápido cambias de lugar)
- **Aceleración** = derivada de la velocidad (qué tan rápido cambias de velocidad)

````python
# Función cuadrática: f(x) = x²
x = np.linspace(-3, 3, 100)
y = x**2

# Su derivada: f'(x) = 2x
y_derivada = 2 * x

# Visualización
plt.figure(figsize=(12, 5))

# Función original
plt.subplot(1, 2, 1)
plt.plot(x, y, 'b-', linewidth=2)
plt.xlabel('x', fontsize=12)
plt.ylabel('f(x) = x²', fontsize=12)
plt.title('Función Original', fontsize=14, fontweight='bold')
plt.grid(True, alpha=0.3)
plt.axhline(0, color='black', linewidth=0.5)
plt.axvline(0, color='black', linewidth=0.5)

# Derivada
plt.subplot(1, 2, 2)
plt.plot(x, y_derivada, 'r-', linewidth=2)
plt.xlabel('x', fontsize=12)
plt.ylabel("f'(x) = 2x", fontsize=12)
plt.title('Derivada (Pendiente)', fontsize=14, fontweight='bold')
plt.grid(True, alpha=0.3)
plt.axhline(0, color='black', linewidth=0.5)
plt.axvline(0, color='black', linewidth=0.5)

plt.tight_layout()
plt.show()

print("Interpretación:")
print("- Cuando x < 0: pendiente negativa (función decrece)")
print("- Cuando x = 0: pendiente = 0 (mínimo de la función)")
print("- Cuando x > 0: pendiente positiva (función crece)")
````

### 5.2 Descenso de Gradiente (Gradient Descent)

Es el **algoritmo de optimización más importante en ML**. Imagina que estás en una montaña con niebla:
- Quieres bajar al valle (mínimo)
- No ves nada, solo sientes la pendiente bajo tus pies
- Das pasos en la dirección de mayor bajada

**Eso es el descenso de gradiente**.

````python
# Descenso de gradiente para encontrar el mínimo de f(x) = x²

def f(x):
    """Función objetivo: f(x) = x²"""
    return x**2

def derivada_f(x):
    """Derivada: f'(x) = 2x"""
    return 2 * x

# Parámetros
x_inicial = 3.0  # Punto de partida
learning_rate = 0.1  # Tamaño del paso
iteraciones = 20

# Descenso de gradiente
x_actual = x_inicial
historia_x = [x_actual]
historia_y = [f(x_actual)]

print("Iteración | x | f(x) | Gradiente")
print("-" * 50)

for i in range(iteraciones):
    gradiente = derivada_f(x_actual)
    x_nuevo = x_actual - learning_rate * gradiente
    
    print(f"{i:9d} | {x_actual:6.3f} | {f(x_actual):6.3f} | {gradiente:9.3f}")
    
    x_actual = x_nuevo
    historia_x.append(x_actual)
    historia_y.append(f(x_actual))

print(f"\n✅ Mínimo encontrado en x = {x_actual:.6f}, f(x) = {f(x_actual):.6f}")

# Visualización del proceso
x_plot = np.linspace(-3.5, 3.5, 100)
y_plot = f(x_plot)

plt.figure(figsize=(12, 6))
plt.plot(x_plot, y_plot, 'b-', linewidth=2, label='f(x) = x²')
plt.scatter(historia_x, historia_y, c='red', s=100, zorder=5, label='Pasos del algoritmo')
plt.plot(historia_x, historia_y, 'r--', alpha=0.5)

# Anotar inicio y fin
plt.scatter(historia_x[0], historia_y[0], c='green', s=200, marker='*', 
            edgecolors='black', linewidths=2, zorder=10, label='Inicio')
plt.scatter(historia_x[-1], historia_y[-1], c='gold', s=200, marker='*', 
            edgecolors='black', linewidths=2, zorder=10, label='Mínimo')

plt.xlabel('x', fontsize=12)
plt.ylabel('f(x)', fontsize=12)
plt.title('Descenso de Gradiente: Buscando el mínimo', fontsize=14, fontweight='bold')
plt.legend(fontsize=10)
plt.grid(True, alpha=0.3)
plt.show()
````

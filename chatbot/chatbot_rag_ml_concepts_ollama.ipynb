# 📚 ¿Qué es RAG (Retrieval-Augmented Generation)?

## Introducción a RAG

**RAG** es una técnica avanzada que combina dos capacidades complementarias:

1. **Recuperación (Retrieval)**: Búsqueda de información relevante en una base de conocimiento
2. **Generación (Generation)**: Creación de respuestas naturales usando un modelo de lenguaje (LLM)

## El Problema que Resuelve RAG

Los LLMs tradicionales (como GPT, Llama, etc.) tienen limitaciones importantes:

### ❌ Limitaciones de LLMs sin RAG
- **Conocimiento estático**: solo saben lo que aprendieron durante su entrenamiento (datos hasta cierta fecha)
- **Alucinaciones**: inventan información que suena correcta pero es falsa
- **Sin contexto específico**: no conocen información privada/empresarial/especializada
- **Costoso actualizar**: re-entrenar un modelo es extremadamente caro

### ✅ Ventajas de RAG
- **Conocimiento actualizable**: añade nueva información sin re-entrenar el modelo
- **Respuestas fundamentadas**: el LLM responde basándose en documentos reales (reduce alucinaciones)
- **Especialización**: puedes crear chatbots expertos en dominios específicos
- **Transparencia**: puedes ver qué documentos usó el LLM para responder
- **Privacidad**: la información sensible permanece en tu base de datos local

## ¿Cómo Funciona RAG?

```
┌─────────────────────────────────────────────────────────────┐
│  USUARIO: "¿Qué es el overfitting?"                         │
└────────────────────┬────────────────────────────────────────┘
                     │
                     ▼
┌─────────────────────────────────────────────────────────────┐
│  PASO 1: EMBEDDING DE LA PREGUNTA                           │
│  Convierte la pregunta en un vector numérico                │
└────────────────────┬────────────────────────────────────────┘
                     │
                     ▼
┌─────────────────────────────────────────────────────────────┐
│  PASO 2: BÚSQUEDA EN BASE VECTORIAL (FAISS)                 │
│  Encuentra los fragmentos más similares semánticamente      │
│  Resultado: 2-3 fragmentos relevantes sobre overfitting     │
└────────────────────┬────────────────────────────────────────┘
                     │
                     ▼
┌─────────────────────────────────────────────────────────────┐
│  PASO 3: CONSTRUCCIÓN DEL PROMPT                            │
│  Contexto: [fragmentos recuperados]                         │
│  Pregunta: ¿Qué es el overfitting?                          │
└────────────────────┬────────────────────────────────────────┘
                     │
                     ▼
┌─────────────────────────────────────────────────────────────┐
│  PASO 4: LLM GENERA RESPUESTA (Llama 3)                     │
│  El modelo lee el contexto y genera una respuesta precisa   │
└────────────────────┬────────────────────────────────────────┘
                     │
                     ▼
┌─────────────────────────────────────────────────────────────┐
│  RESPUESTA: "El overfitting ocurre cuando un modelo..."     │
│  + Fuentes: [documentos consultados]                        │
└─────────────────────────────────────────────────────────────┘
```

## ¿Por Qué Usar Ollama en Lugar de APIs Remotas?

Existen dos formas de ejecutar LLMs para RAG:

### 🌐 Opción 1: APIs Remotas (OpenAI, Anthropic, etc.)
**Ventajas:**
- ✅ Fácil de usar (solo necesitas una API key)
- ✅ Modelos muy potentes (GPT-4, Claude 3.5)
- ✅ No requiere hardware potente

**Desventajas:**
- ❌ **Costo por token** (puede volverse muy caro en producción)
- ❌ **Privacidad**: tus datos se envían a servidores externos
- ❌ **Dependencia de internet**: no funciona offline
- ❌ **Latencia**: cada llamada requiere round-trip a servidores remotos
- ❌ **Límites de rate**: restricciones de uso

### 🖥️ Opción 2: Ollama (LLMs Locales)
**Ventajas:**
- ✅ **100% Privado**: tus datos nunca salen de tu máquina
- ✅ **Sin costos recurrentes**: no pagas por tokens/llamadas
- ✅ **Funciona offline**: no requiere conexión a internet
- ✅ **Baja latencia**: ejecución local instantánea
- ✅ **Sin límites de uso**: usa el modelo cuanto quieras
- ✅ **Fácil de usar**: Ollama simplifica enormemente la ejecución local

**Desventajas:**
- ❌ Requiere hardware decente (8-16GB RAM recomendado)
- ❌ Modelos locales son menos potentes que GPT-4 (pero suficientes para muchos casos)
- ❌ Instalación inicial requerida

## ¿Cuándo Usar Ollama?

**Usa Ollama si:**
- Trabajas con información sensible/confidencial (datos médicos, legales, empresariales)
- Necesitas control total sobre costos
- Quieres experimentar libremente sin preocuparte por facturas
- Trabajas en entornos sin internet o con restricciones de red
- Estás aprendiendo/desarrollando (experimentación sin costo)

**Usa APIs remotas si:**
- Necesitas la máxima calidad de respuestas (GPT-4 nivel)
- No tienes hardware potente disponible
- El costo no es una limitación importante
- Los datos no son sensibles

# 🤖 Chatbot RAG con Ollama + Llama 3.2

## Descripción General

Este notebook implementa un **sistema RAG (Retrieval-Augmented Generation)** para responder preguntas sobre conceptos de Machine Learning, utilizando:

- **Ollama** como servidor local de LLM (ejecuta Llama 3.2 3B cuantizado)
- **LangChain** para orquestar el pipeline RAG completo
- **FAISS** como base de datos vectorial para búsqueda semántica
- **all-MiniLM-L6-v2** para generar embeddings de texto

## Arquitectura RAG

El flujo de trabajo es el siguiente:

1. **Ingestión de documentos**: Cargamos un corpus de conceptos de ML
2. **Chunking**: Dividimos el texto en fragmentos manejables
3. **Embedding**: Convertimos cada fragmento en un vector numérico
4. **Indexación**: Almacenamos los vectores en FAISS para búsqueda eficiente
5. **Recuperación**: Cuando el usuario hace una pregunta, buscamos los fragmentos más relevantes
6. **Generación**: Pasamos los fragmentos recuperados + la pregunta al LLM (Llama 3.2) para generar una respuesta contextualizada

Un corpus es simplemente un conjunto de textos que contiene la información que queremos que nuestro chatbot conozca. En un sistema RAG, el LLM no "memoriza" esta información, sino que la recupera cuando es relevante para responder preguntas.

### 🔹 Opción 3: Cargar desde Múltiples Archivos de Texto

**Ventaja**: Organiza el conocimiento por temas/categorías.

```python

)
documents = loader.load()
```

## Paso 3: Carga y División de Documentos (Chunking)

### ¿Por qué dividir los documentos?

Los modelos de embeddings y LLMs tienen **límites de contexto** (cantidad máxima de texto que pueden procesar). Además, fragmentos más pequeños permiten:
- **Recuperación más precisa**: encontrar exactamente la información relevante
- **Mejor rendimiento**: procesar fragmentos pequeños es más eficiente
- **Evitar dilución de información**: cada fragmento se enfoca en un concepto específico

### Proceso de chunking

1. **TextLoader**: Carga el archivo de texto plano
2. **RecursiveCharacterTextSplitter**: Divide el texto inteligentemente:
   - `chunk_size=300`: cada fragmento tendrá aprox. 300 caracteres
   - `chunk_overlap=50`: solapamiento entre fragmentos para no perder contexto
   - `separators`: divide primero por párrafos (`\n\n`), luego por líneas (`\n`), luego por frases (`. `), etc.

Esta estrategia asegura que los conceptos no se partan a mitad de frase.

````python
# =============================================================================
# CARGA Y DIVISIÓN DE DOCUMENTOS (CHUNKING)
# =============================================================================
# En esta celda cargamos el corpus y lo dividimos en fragmentos (chunks) más
# pequeños para facilitar la búsqueda semántica y el procesamiento por el LLM.
# =============================================================================

# IMPORTS ACTUALIZADOS PARA LANGCHAIN 1.0+
# - TextLoader ahora está en langchain_community.document_loaders
# - RecursiveCharacterTextSplitter ahora está en langchain_text_splitters
from langchain_community.document_loaders import TextLoader
from langchain_text_splitters import RecursiveCharacterTextSplitter

# 1. CARGA DEL DOCUMENTO
# TextLoader lee archivos de texto plano
# encoding="utf-8" asegura que se lean correctamente caracteres especiales
loader = TextLoader("ml_concepts_corpus_extended.txt", encoding="utf-8")
documents = loader.load()  # Retorna una lista de objetos Document

# 2. DIVISIÓN EN FRAGMENTOS (CHUNKING)
# RecursiveCharacterTextSplitter divide el texto de forma inteligente:
# - chunk_size: tamaño objetivo de cada fragmento (en caracteres)
# - chunk_overlap: solapamiento entre fragmentos para mantener contexto
# - separators: secuencia de separadores (prioriza dividir por párrafos, luego líneas, etc.)
text_splitter = RecursiveCharacterTextSplitter(
    chunk_size=300,        # Cada fragmento tendrá ~300 caracteres
    chunk_overlap=50,      # 50 caracteres de solapamiento entre fragmentos
    separators=["\n\n", "\n", ". ", " ", ""]  # Divide primero por párrafos, luego líneas, etc.
)

# split_documents() toma la lista de documentos y retorna una lista de fragmentos
chunks = text_splitter.split_documents(documents)

# 3. VISUALIZACIÓN DE RESULTADOS
print(f"📚 Corpus dividido en {len(chunks)} fragmentos.")
print("\n🔍 Muestra de los primeros 2 fragmentos:")
for i, chunk in enumerate(chunks[:2]):
    # Mostramos solo los primeros 200 caracteres de cada fragmento
    print(f"\n--- Fragmento {i+1} ---")
    print(chunk.page_content[:200] + "...")
````

## Paso 4: Creación de la Base Vectorial (Vector Store)

### ¿Qué son los embeddings?

Los **embeddings** son representaciones numéricas (vectores) del texto que capturan su significado semántico. Textos con significado similar tienen vectores cercanos en el espacio vectorial.

### Proceso de vectorización

1. **HuggingFaceEmbeddings**: Usa el modelo `all-MiniLM-L6-v2` de Sentence Transformers
   - Modelo ligero (~90MB) pero efectivo
   - Genera vectores de 384 dimensiones
   - Optimizado para búsqueda semántica
   - `device='cpu'`: ejecuta en CPU (no requiere GPU)

2. **FAISS (Facebook AI Similarity Search)**: Base de datos vectorial altamente eficiente
   - `from_documents()`: crea el índice directamente desde los fragmentos
   - Almacena los vectores y permite búsqueda rápida por similitud de coseno
   - Versión CPU: funciona sin GPU

3. **Retriever**: Interfaz para buscar documentos relevantes
   - `search_kwargs={"k": 2}`: recupera los 2 fragmentos más similares a cada pregunta

### ¿Por qué FAISS?
- **Velocidad**: búsqueda extremadamente rápida incluso con millones de vectores
- **Eficiencia de memoria**: algoritmos optimizados
- **Escalabilidad**: funciona desde pocos documentos hasta datasets masivos

````python
# =============================================================================
# CREACIÓN DE LA BASE VECTORIAL (VECTOR STORE)
# =============================================================================
# En esta celda convertimos los fragmentos de texto en vectores numéricos
# (embeddings) y los almacenamos en una base de datos vectorial (FAISS)
# para realizar búsquedas semánticas eficientes.
# =============================================================================

# IMPORTS ACTUALIZADOS PARA LANGCHAIN 1.0+
# - HuggingFaceEmbeddings ahora está en langchain_community.embeddings
# - FAISS ahora está en langchain_community.vectorstores
from langchain_community.embeddings import HuggingFaceEmbeddings
from langchain_community.vectorstores import FAISS

# 1. CONFIGURACIÓN DEL MODELO DE EMBEDDINGS
# HuggingFaceEmbeddings usa modelos de Sentence Transformers
# all-MiniLM-L6-v2 es un modelo ligero (~90MB) pero muy efectivo:
# - Genera vectores de 384 dimensiones
# - Entrenado para tareas de búsqueda semántica
# - Balance ideal entre tamaño y rendimiento
embeddings = HuggingFaceEmbeddings(
    model_name="sentence-transformers/all-MiniLM-L6-v2",  # Modelo pre-entrenado
    model_kwargs={'device': 'cpu'}  # Ejecuta en CPU (no requiere GPU)
)

# 2. CREACIÓN DE LA BASE VECTORIAL CON FAISS
# FAISS (Facebook AI Similarity Search) es una biblioteca optimizada para
# búsqueda de vectores similares. from_documents() realiza dos acciones:
# a) Convierte cada fragmento en un vector usando el modelo de embeddings
# b) Crea un índice FAISS para búsqueda rápida por similitud
print("🔄 Creando base vectorial (esto puede tardar un minuto la primera vez)...")
vectorstore = FAISS.from_documents(
    chunks,      # Lista de fragmentos de texto
    embeddings   # Modelo para convertir texto -> vectores
)

# 3. CONFIGURACIÓN DEL RETRIEVER
# El retriever es una interfaz para buscar documentos relevantes
# search_kwargs={"k": 2} significa que recuperará los 2 fragmentos más similares
retriever = vectorstore.as_retriever(
    search_kwargs={"k": 2}  # Número de fragmentos a recuperar por pregunta
)

print("✅ Base vectorial creada correctamente.")
print(f"📊 Índice FAISS contiene {len(chunks)} fragmentos vectorizados.")
````

## Paso 6: Construcción de la Cadena RAG

### Arquitectura RAG Moderna (LangChain 1.0+)

En LangChain 1.0+, el patrón recomendado para RAG usa **LCEL (LangChain Expression Language)** con el operador pipe (`|`). Esta es una arquitectura más flexible y moderna que `RetrievalQA` (deprecado).

### Componentes de la cadena

1. **RunnablePassthrough**: Pasa la pregunta del usuario a través de la cadena
2. **Retriever**: Busca los fragmentos más relevantes en FAISS
3. **format_docs**: Función helper que concatena los fragmentos recuperados
4. **ChatPromptTemplate**: Define el prompt que se envía al LLM
   - `{context}`: fragmentos recuperados de la base vectorial
   - `{question}`: pregunta original del usuario
5. **LLM (Llama 3.2)**: Genera la respuesta basándose en el contexto
6. **StrOutputParser**: Convierte la respuesta del LLM a string

### Flujo de ejecución

```
Usuario pregunta "¿Qué es overfitting?"
    ↓
Retriever busca en FAISS → recupera 2 fragmentos relevantes
    ↓
format_docs concatena los fragmentos
    ↓
Prompt combina contexto + pregunta
    ↓
Llama 3.2 genera respuesta
    ↓
StrOutputParser extrae el texto
    ↓
Respuesta final al usuario
```

### Ventajas de este patrón

- **Transparencia**: ves claramente cada paso
- **Flexibilidad**: puedes modificar cualquier componente
- **Debugging**: más fácil de depurar que cadenas opacas
- **Futuro-proof**: compatible con LangChain 1.0+

````python
# =============================================================================
# CONSTRUCCIÓN DE LA CADENA RAG (LANGCHAIN 1.0+ - PATRÓN MODERNO)
# =============================================================================
# En esta celda construimos el pipeline RAG completo usando LCEL (LangChain
# Expression Language), el patrón recomendado en LangChain 1.0+.
# Este patrón reemplaza a RetrievalQA que está deprecado.
# =============================================================================

# IMPORTS PARA RAG MODERNO
from langchain_core.runnables import RunnablePassthrough  # Pasa datos a través de la cadena
from langchain_core.output_parsers import StrOutputParser  # Convierte salida a string
from langchain_core.prompts import ChatPromptTemplate       # Define el template del prompt

# 1. FUNCIÓN HELPER PARA FORMATEAR DOCUMENTOS
# Esta función toma la lista de documentos recuperados y los concatena
# en un único string separado por saltos de línea dobles
def format_docs(docs):
    """
    Formatea una lista de documentos recuperados en un string único.
    
    Args:
        docs: Lista de objetos Document recuperados por el retriever
    
    Returns:
        String con el contenido de todos los documentos concatenado
    """
    return "\n\n".join(doc.page_content for doc in docs)

# 2. DEFINICIÓN DEL PROMPT
# ChatPromptTemplate.from_template() crea un prompt con variables
# {context}: será reemplazado por los fragmentos recuperados
# {question}: será reemplazado por la pregunta del usuario
prompt_template = """Eres un asistente experto en Machine Learning. Usa el siguiente contexto para responder la pregunta del usuario de forma clara y precisa.

Contexto recuperado de la base de conocimiento:
{context}

Pregunta del usuario: {question}

Respuesta (basa tu respuesta SOLO en el contexto proporcionado):"""

prompt = ChatPromptTemplate.from_template(prompt_template)

# 3. CONSTRUCCIÓN DE LA CADENA RAG CON LCEL
# El operador | (pipe) encadena componentes de izquierda a derecha
# RunnablePassthrough() pasa la pregunta sin modificarla
rag_chain = (
    {
        # El contexto se obtiene así: pregunta → retriever → format_docs
        "context": retriever | format_docs,
        # La pregunta se pasa directamente sin modificación
        "question": RunnablePassthrough()
    }
    | prompt              # Combina context + question en el template
    | llm                 # Envía el prompt a Llama 3.2
    | StrOutputParser()   # Convierte la respuesta a string
)

# 4. FUNCIÓN PARA HACER PREGUNTAS
def ask_ml_question(question):
    """
    Realiza una pregunta al sistema RAG y muestra la respuesta con fuentes.
    
    Args:
        question: String con la pregunta sobre conceptos de ML
    """
    print(f"\n{'='*70}")
    print(f"❓ PREGUNTA: {question}")
    print(f"{'='*70}\n")
    
    # Invocar la cadena RAG
    print("🔄 Buscando en la base de conocimiento...")
    answer = rag_chain.invoke(question)
    
    print(f"\n💡 RESPUESTA:\n{answer}\n")
    
    # Recuperar también los documentos fuente para mostrarlos
    print("📄 FUENTES CONSULTADAS:")
    source_docs = retriever.invoke(question)
    for i, doc in enumerate(source_docs, 1):
        print(f"\n  [{i}] {doc.page_content[:200]}...")
    
    print(f"\n{'='*70}\n")

# 5. EJEMPLO DE USO
print("✅ Cadena RAG configurada correctamente.")
print("🧪 Probando con una pregunta de ejemplo...\n")

# Pregunta de prueba
ask_ml_question("¿Qué es el overfitting?")
````

## Paso 7: Interfaz Interactiva (Opcional)

### Widget de IPython

Esta celda crea una **interfaz interactiva** dentro del notebook usando `ipywidgets`. Permite:

- **Campo de texto**: para escribir preguntas
- **Submit on Enter**: presiona Enter para enviar la pregunta
- **Output dinámico**: las respuestas aparecen debajo del campo de texto

### Componentes

1. **widgets.Text**: cuadro de entrada de texto
   - `placeholder`: texto de ejemplo que aparece cuando está vacío
   - `description`: etiqueta del campo
   
2. **widgets.Output**: área donde se mostrarán las respuestas
   - `clear_output()`: limpia respuestas anteriores
   
3. **on_submit**: callback que se ejecuta al presionar Enter

### Uso

1. Ejecuta esta celda
2. Aparecerá un campo de texto
3. Escribe una pregunta (ej: "¿Qué es Random Forest?")
4. Presiona **Enter**
5. La respuesta aparecerá debajo

**Nota**: Esta celda requiere que estés ejecutando el notebook en un entorno con soporte para widgets (Jupyter Notebook, JupyterLab, VS Code con extensión de Jupyter).

````python
# =============================================================================
# INTERFAZ INTERACTIVA CON IPYWIDGETS
# =============================================================================
# Esta celda crea una interfaz gráfica simple dentro del notebook para
# interactuar con el chatbot RAG de forma más amigable.
# =============================================================================

# IMPORTS PARA WIDGETS
from IPython.display import display  # Función para mostrar widgets
import ipywidgets as widgets          # Biblioteca de widgets interactivos

# 1. CREACIÓN DEL CAMPO DE ENTRADA DE TEXTO
# widgets.Text crea un cuadro de texto interactivo
text_input = widgets.Text(
    value='',                                    # Valor inicial vacío
    placeholder='Ej: ¿Qué es Random Forest?',    # Texto de ayuda
    description='Pregunta:',                     # Etiqueta del campo
    layout=widgets.Layout(width='80%')           # Ancho del campo (80% del contenedor)
)

# 2. CREACIÓN DEL ÁREA DE SALIDA
# widgets.Output es un contenedor donde se mostrarán las respuestas
output = widgets.Output(
    layout=widgets.Layout(
        width='100%',            # Ancho completo
        border='1px solid #ddd', # Borde gris claro
        padding='10px'           # Espaciado interno
    )
)

# 3. FUNCIÓN CALLBACK PARA MANEJAR EL ENVÍO
def on_submit(sender):
    """
    Se ejecuta cuando el usuario presiona Enter en el campo de texto.
    
    Args:
        sender: El widget que disparó el evento (text_input)
    """
    # with output: redirige todo el print() al widget de salida
    with output:
        # Limpiar respuestas anteriores
        output.clear_output()
        
        # Obtener el texto ingresado
        user_question = sender.value.strip()
        
        # Validar que no esté vacío
        if user_question:
            # Llamar a la función que procesa la pregunta
            ask_ml_question(user_question)
        else:
            # Mensaje de error si no hay texto
            print("⚠️  Por favor, escribe una pregunta antes de presionar Enter.")

# 4. VINCULAR EL EVENTO on_submit AL CAMPO DE TEXTO
# Cuando el usuario presione Enter, se ejecutará on_submit()
text_input.on_submit(on_submit)

# 5. MOSTRAR LA INTERFAZ
print("🎨 Interfaz interactiva del chatbot:")
print("   • Escribe tu pregunta en el campo de texto")
print("   • Presiona Enter para obtener la respuesta")
print("   • El chatbot buscará en la base de conocimiento y generará una respuesta\n")

# display() muestra los widgets en el notebook
display(text_input, output)
````

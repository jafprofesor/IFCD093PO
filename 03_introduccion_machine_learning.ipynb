## 🧠 1. ¿Qué es el Machine Learning? <a id='1-que-es-ml'></a>

**Machine Learning (Aprendizaje Automático)** es una rama de la Inteligencia Artificial que se enfoca en dar a las computadoras la habilidad de **aprender a partir de datos, sin ser explícitamente programadas para ello**.

![IA,ML, Aprendizaje profundo](imagenes/ai-ml-ds.png)

### 1.1 Programación Tradicional vs. Machine Learning <a id='1.1-comparativa'></a>

Imagina que quieres crear un programa para detectar si un email es spam.

#### Enfoque de Programación Tradicional:
Tú, como programador, tendrías que escribir un conjunto de **reglas explícitas**:
- `SI` el email contiene la palabra "oferta", `ENTONCES` es spam.
- `SI` el email contiene "viagra", `ENTONCES` es spam.
- `SI` el email tiene muchas mayúsculas, `ENTONCES` es spam.

**Problema**: Este sistema es frágil, difícil de mantener y no se adapta a nuevos tipos de spam.

![Programación Tradicional](imagenes/programacion_tradicional.png)

#### Enfoque de Machine Learning:
En lugar de escribir reglas, le das al sistema:
1. **Datos**: Miles de emails de ejemplo.
2. **Respuestas**: Cada email etiquetado como "spam" o "no spam".

El algoritmo de Machine Learning **aprende por sí mismo los patrones** que distinguen el spam del no spam. El resultado es un **modelo** que puede predecir si un *nuevo* email es spam.

**Ventaja**: El sistema es más robusto, se adapta y puede descubrir patrones que un humano no vería.

![Machine Learning](imagenes/machine_learning.png)

## 🏷️ 2. Tipos de Machine Learning <a id='2-tipos-ml'></a>

Hay tres categorías principales de aprendizaje automático.

### 2.1 Aprendizaje Supervisado <a id='2.1-supervisado'></a>

Es el tipo más común. Se caracteriza porque le damos al algoritmo **datos etiquetados** (como en el ejemplo del spam).

**Analogía**: Aprender con un profesor. Tienes las preguntas (datos) y las respuestas correctas (etiquetas).

Se divide en dos subtipos:

1. **Clasificación**: La etiqueta es una categoría discreta.

    - *Ejemplos*: ¿Es este email spam o no spam? ¿Es este tumor benigno o maligno? ¿Qué tipo de animal es el de la foto (perro, gato, pájaro)?

3. **Regresión**: La etiqueta es un valor numérico continuo.

    - *Ejemplos*: ¿Cuál será el precio de esta casa? ¿Qué temperatura hará mañana? ¿Cuántas ventas tendrá este producto el próximo mes?

![Supervisado](imagenes/supervisado.png)

### 2.2 Aprendizaje No Supervisado <a id='2.2-no-supervisado'></a>

En este caso, le damos al algoritmo **datos sin etiquetar** y le pedimos que encuentre estructura o patrones por sí mismo.

**Analogía**: Aprender sin profesor. Te dan un montón de datos y tienes que encontrar las relaciones tú solo.

Principales tareas:

1. **Clustering (Agrupamiento)**: Agrupar datos similares.

    - *Ejemplos*: Segmentar clientes en grupos de compra similares. Agrupar noticias por tema.
      

3. **Reducción de Dimensionalidad**: Simplificar los datos eliminando características redundantes.

    - *Ejemplos*: Comprimir imágenes. Visualizar datos complejos en 2D o 3D.

5. **Detección de Anomalías**: Encontrar datos que no encajan con el resto.

    - *Ejemplos*: Detectar transacciones fraudulentas con tarjetas de crédito. Identificar fallos en motores de avión.

![No Supervisado](imagenes/no_supervisado.png)

### 2.3 Aprendizaje por Refuerzo <a id='2.3-refuerzo'></a>

Un **agente** aprende a tomar decisiones interactuando con un **entorno**. El agente recibe **recompensas** o **castigos** por sus acciones.

**Analogía**: Enseñar a un perro un truco. Si hace lo correcto, le das un premio (recompensa). Si no, no le das nada (o un castigo leve).

- *Ejemplos*: Un programa que aprende a jugar al ajedrez. Un coche autónomo que aprende a conducir. Un robot que aprende a caminar.

![Refuerzo](imagenes/refuerzo.png)

**En este curso, nos centraremos principalmente en el Aprendizaje Supervisado y No Supervisado.**

## 🗺️ 3. El Flujo de Trabajo de un Proyecto de ML <a id='3-workflow'></a>

Un proyecto de Machine Learning no es solo entrenar un modelo. Es un proceso iterativo con varios pasos clave.

![Workflow](imagenes/workflow.png)

### 3.1 Paso a Paso: De los Datos al Modelo <a id='3.1-pasos'></a>

1. **Definición del Problema**: ¿Qué queremos predecir? ¿Qué métrica de éxito usaremos? (Ej: Predecir el precio de casas, con un error medio de menos de 10.000€).

2. **Obtención de Datos**: Recolectar los datos necesarios. Pueden venir de bases de datos, archivos CSV, APIs, etc.

3. **Análisis Exploratorio de Datos (EDA)**: ¡Un paso crucial! Entender los datos a fondo. Buscar patrones, anomalías, correlaciones. Aquí es donde la visualización (Matplotlib, Seaborn) es fundamental.

4. **Preparación de Datos (Preprocesamiento)**: Limpiar y transformar los datos para que el modelo pueda usarlos. Esto incluye:
   - Manejar valores nulos.
   - Codificar variables categóricas (convertir texto a números).
   - Escalar características numéricas (ej: normalización Z-score).

5. **División de Datos (Train/Test Split)**: Dividir los datos en dos conjuntos:
   - **Conjunto de Entrenamiento (Train)**: Para que el modelo aprenda (típicamente 70-80% de los datos).
   - **Conjunto de Prueba (Test)**: Para evaluar el rendimiento del modelo en datos que **nunca ha visto** (20-30%). Esto es vital para saber si el modelo generaliza bien.

6. **Selección y Entrenamiento del Modelo**: Elegir un algoritmo (o varios) y entrenarlo con el conjunto de entrenamiento.

7. **Evaluación del Modelo**: Usar el conjunto de prueba para medir el rendimiento del modelo. ¿Cumple con la métrica de éxito que definimos al principio?

8. **Ajuste de Hiperparámetros (Tuning)**: Modificar la configuración del modelo (hiperparámetros) para mejorar su rendimiento. Es un proceso iterativo.

9. **Despliegue (Deployment)**: Si el modelo es bueno, ponerlo en producción para que pueda ser utilizado en una aplicación real.

### Regresión Lineal: La Función Hipótesis

En Machine Learning, necesitamos crear una **función matemática** que haga predicciones basadas en los datos. Esta función se llama **Función Hipótesis**.

**¿Cómo funciona?**
1. Recibe un dato de entrada (ej: tamaño de una casa).
2. Calcula una predicción usando una fórmula matemática.
3. La fórmula tiene **parámetros ajustables** (θ) que el algoritmo aprende de los datos.

Para **regresión lineal** (predicción de números), usamos la **ecuación de una recta**:

$$
h_θ(x) = θ_0 + θ_1x
$$

Donde:
- **h(x)** = predicción (ej: precio estimado)
- **θ₀** (theta cero) = número inicial (dónde comienza la recta)
- **θ₁** (theta uno) = pendiente (cómo sube o baja)
- **x** = entrada (ej: tamaño en m²)

**En lenguaje simple**: El algoritmo intenta encontrar dos números (θ₀ y θ₁) que, cuando los usamos en la fórmula, predigan correctamente el precio de las casas.

### Regresión lineal: Construcción del modelo

Ahora la pregunta importante: **¿Cómo encuentra el algoritmo los valores correctos de θ₀ y θ₁?**

Veremos dos cosas:
1. **Cómo medir si el modelo está haciendo un buen trabajo** (Función de Coste)
2. **Cómo mejorar automáticamente hasta encontrar los mejores valores** (Gradient Descent)

#### Diferentes Modelos, Diferentes Resultados

Primero, veamos visualmente por qué es importante encontrar los parámetros correctos:

````python
# Visualizar la diferencia entre una línea bien ajustada y otra mal ajustada
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# Usaremos los datos de las casas
X_plot = casas_df['Tamaño'].values
y_plot = casas_df['Precio'].values

# --- Gráfico 1: Línea BIEN AJUSTADA ---
ax1 = axes[0]
ax1.scatter(X_plot, y_plot, color='blue', s=80, alpha=0.6, label='Datos Reales')

# Línea bien ajustada (aproximamos la tendencia)
theta0_good = 5000
theta1_good = 1500
y_pred_good = theta0_good + theta1_good * X_plot
ax1.plot(X_plot, y_pred_good, color='green', linewidth=2.5, label=f'Hipótesis: h(x) = {theta0_good} + {theta1_good}x')

# Mostrar algunos residuos (errores)
for i in [0, 10, 20, 30, 40]:
    ax1.plot([X_plot[i], X_plot[i]], [y_plot[i], y_pred_good[i]], color='red', linewidth=1, alpha=0.5)

ax1.set_xlabel('Tamaño (m²)', fontsize=12)
ax1.set_ylabel('Precio (€)', fontsize=12)
ax1.set_title('✅ Línea BIEN AJUSTADA\n(Errores pequeños)', fontsize=13, fontweight='bold')
ax1.legend()
ax1.grid(True, alpha=0.3)

# --- Gráfico 2: Línea MAL AJUSTADA ---
ax2 = axes[1]
ax2.scatter(X_plot, y_plot, color='blue', s=80, alpha=0.6, label='Datos Reales')

# Línea mal ajustada (pendiente incorrecta)
theta0_bad = 100000
theta1_bad = 500
y_pred_bad = theta0_bad + theta1_bad * X_plot
ax2.plot(X_plot, y_pred_bad, color='red', linewidth=2.5, label=f'Hipótesis: h(x) = {theta0_bad} + {theta1_bad}x')

# Mostrar algunos residuos (errores)
for i in [0, 10, 20, 30, 40]:
    ax2.plot([X_plot[i], X_plot[i]], [y_plot[i], y_pred_bad[i]], color='orange', linewidth=1, alpha=0.5)

ax2.set_xlabel('Tamaño (m²)', fontsize=12)
ax2.set_ylabel('Precio (€)', fontsize=12)
ax2.set_title('❌ Línea MAL AJUSTADA\n(Errores grandes)', fontsize=13, fontweight='bold')
ax2.legend()
ax2.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()

print("En el gráfico izquierdo, los residuos (líneas verdes) son PEQUEÑOS.")
print("En el gráfico derecho, los residuos (líneas rojas) son GRANDES.")
````

#### La Función de Coste (Cost Function)

Para saber qué línea es "mejor", necesitamos una **métrica matemática** que mida cuán mal se ajusta el modelo. Esta métrica se llama **Función de Coste** o **Función de Error**.

La función de coste más común en regresión lineal es el **Error Cuadrático Medio (MSE)**:

**MSE = Mean Squared Error** (Error Cuadrático Medio en español)
- **Mean**: Promedio
- **Squared**: Elevado al cuadrado
- **Error**: Diferencia entre predicción y valor real

En otras palabras, MSE calcula el **promedio de los errores al cuadrado**.

$$
J(θ_0, θ_1) = \frac{1}{2m} \sum_{i=1}^{m} (h_θ(x^{(i)}) - y^{(i)})^2
$$

Donde:
- **m**: número de ejemplos de entrenamiento.
- **h_θ(x⁽ⁱ⁾)**: predicción del modelo para el i-ésimo ejemplo.
- **y⁽ⁱ⁾**: valor real (etiqueta) del i-ésimo ejemplo.
- **(h_θ(x⁽ⁱ⁾) - y⁽ⁱ⁾)**: el error (residuo) para ese ejemplo.

**¿Qué significa?** Calcula el error promedio entre nuestras predicciones y los valores reales. **Cuanto menor sea J, mejor se ajusta el modelo.**

#### Visualizando los Residuos

Los residuos son las distancias verticales entre los puntos reales y la línea de predicción. En la imagen anterior, las líneas rojas/naranjas representan estos residuos.

- **Línea bien ajustada**: residuos pequeños → J(θ) es pequeño ✅
- **Línea mal ajustada**: residuos grandes → J(θ) es grande ❌

#### El Proceso de Optimización

El algoritmo de regresión lineal funciona así:

1. **Inicializa** θ₀ y θ₁ (generalmente en 0).
2. **Calcula** la función de coste J(θ₀, θ₁) con esos parámetros.
3. **Modifica** θ₀ y θ₁ para **minimizar** J (hacer que sea más pequeño).
4. **Repite** hasta que J no pueda disminuir más.

Este proceso de "descenso gradual" hacia el mínimo se llama **Gradient Descent** (Descenso por Gradiente), y es uno de los algoritmos de optimización más importantes en Machine Learning.

**El objetivo final**: Encontrar los valores de θ₀ y θ₁ que **minimicen la función de coste**, resultando en la línea que **mejor se ajusta** a nuestros datos.

### Regresión lineal: Función de Coste

Ahora que ya comprendemos por qué necesitamos una función de coste, vamos a entender **cómo funciona exactamente** la que se utiliza en regresión lineal.

#### La Ecuación de la Función de Coste

La función de coste más común para regresión lineal es el **Error Cuadrático Medio (MSE)**:

**MSE = Mean Squared Error** → Promedio de los Errores al Cuadrado

$$
J(θ_0, θ_1) = \frac{1}{2m} \sum_{i=1}^{m} (h_θ(x^{(i)}) - y^{(i)})^2
$$

**Desglosando la ecuación:**

- **J(θ₀, θ₁)**: La función de coste. Devuelve un número que mide el error total.
- **m**: El número de ejemplos de entrenamiento (en nuestro caso, 50 casas).
- **Σ** (sumatorio): Suma todos los errores individuales.
- **h_θ(x⁽ⁱ⁾)**: La predicción del modelo para la i-ésima casa.
- **y⁽ⁱ⁾**: El precio real de la i-ésima casa (la etiqueta).
- **(h_θ(x⁽ⁱ⁾) - y⁽ⁱ⁾)**: El error (residuo) para esa casa.
- **Al cuadrado**: Elevamos al cuadrado para que los errores negativos no se cancelen.

#### Paso a Paso: Calculando la Función de Coste

Imaginemos que nuestro modelo está inicializado con θ₀ = 5.000€ y θ₁ = 1.500€/m².

Para cada casa en nuestro conjunto de datos:

1. **Casa 1**: Tamaño = 80 m², Precio Real = 120.000€
   - Predicción: h(80) = 5.000 + 1.500×80 = **125.000€**
   - Error: 125.000 - 120.000 = **5.000€**
   - Error²: (5.000)² = **25.000.000**

2. **Casa 2**: Tamaño = 120 m², Precio Real = 185.000€
   - Predicción: h(120) = 5.000 + 1.500×120 = **185.000€**
   - Error: 185.000 - 185.000 = **0€**
   - Error²: (0)² = **0**

3. **Casa 3**: Tamaño = 150 m², Precio Real = 225.000€
   - Predicción: h(150) = 5.000 + 1.500×150 = **230.000€**
   - Error: 230.000 - 225.000 = **5.000€**
   - Error²: (5.000)² = **25.000.000**

Después de calcular esto para **todas las 50 casas**, sumamos todos los errores al cuadrado y dividimos entre 2m (100):

**J(θ₀, θ₁) = (Suma de todos los errores²) / 100**

#### ¿Qué Significa el Resultado?

- **J = 0**: El modelo predice perfectamente todas las casas (casi imposible en datos reales).
- **J = pequeño**: El modelo está bien ajustado (errores pequeños).
- **J = grande**: El modelo está mal ajustado (errores grandes).

#### Visualizando la Función de Coste

````python
# Calcular la función de coste para diferentes valores de parámetros
import numpy as np
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D

# Usamos los datos de las casas
X = casas_df['Tamaño'].values
y = casas_df['Precio'].values
m = len(X)

# Definir la función de coste
def cost_function(X, y, theta0, theta1):
    """Calcula el Error Cuadrático Medio"""
    predictions = theta0 + theta1 * X
    errors = predictions - y
    return (1 / (2 * m)) * np.sum(errors ** 2)

# Crear una malla de valores para theta0 y theta1
theta0_range = np.linspace(0, 50000, 50)
theta1_range = np.linspace(500, 2500, 50)
J_values = np.zeros((len(theta0_range), len(theta1_range)))

for i, t0 in enumerate(theta0_range):
    for j, t1 in enumerate(theta1_range):
        J_values[i, j] = cost_function(X, y, t0, t1)

# Crear gráfico de contorno (vista 2D)
fig, axes = plt.subplots(1, 2, figsize=(15, 5))

# Gráfico 1: Contorno (curvas de nivel)
ax1 = axes[0]
contour = ax1.contourf(theta1_range, theta0_range, J_values, levels=20, cmap='viridis')
ax1.set_xlabel('θ₁ (pendiente en €/m²)', fontsize=12)
ax1.set_ylabel('θ₀ (intercepto en €)', fontsize=12)
ax1.set_title('Función de Coste: Vista de Contorno\n(Colores más oscuros = menor error)', fontsize=13, fontweight='bold')
plt.colorbar(contour, ax=ax1, label='J(θ₀, θ₁)')

# Marcar el punto óptimo aproximado
optimal_theta0 = 5000
optimal_theta1 = 1500
ax1.plot(optimal_theta1, optimal_theta0, 'r*', markersize=20, label='Mínimo óptimo')
ax1.legend()

# Gráfico 2: Gráfico 3D
ax2 = fig.add_subplot(122, projection='3d')
Theta0, Theta1 = np.meshgrid(theta0_range, theta1_range)
ax2.plot_surface(Theta1, Theta0, J_values.T, cmap='viridis', alpha=0.8, edgecolor='none')
ax2.set_xlabel('θ₁ (€/m²)', fontsize=10)
ax2.set_ylabel('θ₀ (€)', fontsize=10)
ax2.set_zlabel('J(θ₀, θ₁)', fontsize=10)
ax2.set_title('Función de Coste: Vista 3D\n(El valle = mínimo error)', fontsize=13, fontweight='bold')

plt.tight_layout()
plt.show()

print("📊 Interpretación del gráfico:")
print("- El valle (color más oscuro) representa los parámetros que minimizan el error.")
print("- Las laderas representan combinaciones de parámetros con mayor error.")
print("- El algoritmo busca 'descender' hacia el valle (Gradient Descent).")
````

#### El Significado Intuitivo

**La función de coste es como un "mapa de errores":**
- Cada combinación de θ₀ y θ₁ tiene un valor de error asociado (J).
- El algoritmo necesita encontrar la combinación que produce el **menor error**.
- Es como buscar el punto más bajo en un paisaje montañoso: ese es el mejor lugar para estar (nuestro modelo óptimo).

#### ¿Por Qué Elevar al Cuadrado?

Elevamos el error al cuadrado por dos razones:

1. **Penalizar errores grandes**: Un error de 10.000€ se convierte en 100.000.000, lo que penaliza mucho más que un error pequeño.
2. **Eliminar signos negativos**: Así (-5.000)² = (5.000)², y los errores no se cancelan entre sí.

#### Resumen: ¿Qué Hace la Función de Coste?

✅ **Mide** cuán mal predice el modelo para unos parámetros dados.  
✅ **Guía** al algoritmo de optimización hacia mejores parámetros.  
✅ **Detiene** el entrenamiento cuando el error es mínimo.  

Sin la función de coste, el algoritmo no tendría forma de saber si está mejorando o empeorando.

### Regresión lineal: Función de Optimización (Gradient Descent)

Ahora que comprendemos cómo mide el error la función de coste, surge la pregunta crucial: **¿Cómo encuentra el algoritmo los valores óptimos de θ₀ y θ₁?**

La respuesta es el **Gradient Descent** (Descenso por Gradiente), uno de los algoritmos de optimización más importantes en Machine Learning.

#### La Intuición: Bajar una Montaña

Imagina que estás en una montaña y quieres llegar al valle más bajo (donde la función de coste es mínima). No puedes ver el valle, pero tienes una estrategia:

1. Mira hacia abajo (calcula la pendiente del terreno).
2. Da un paso en la dirección que baja más pronunciadamente.
3. Repite hasta que ya no puedas bajar más.

**Gradient Descent funciona exactamente así**, pero con ecuaciones matemáticas.

#### La Matemática del Descenso por Gradiente

El algoritmo actualiza los parámetros así:

$$
θ_0 := θ_0 - α \frac{\partial J}{\partial θ_0}
$$

$$
θ_1 := θ_1 - α \frac{\partial J}{\partial θ_1}
$$

Donde:
- **α** (alpha): el **learning rate** (tasa de aprendizaje), típicamente 0.01 o 0.1. Controla qué tan grandes son los pasos.
- **∂J/∂θ₀** y **∂J/∂θ₁**: las derivadas parciales (gradientes). Indican la dirección y magnitud del cambio.
- **:=**: significa "actualizar el valor de".

#### Paso a Paso: Ejemplo Práctico Simplificado

Para entender mejor, usemos una función de coste más simple:

$$J(θ) = (θ - 3)^2$$

El mínimo está claramente en θ = 3 (donde J = 0).

**Inicializamos**: θ = 10 (comenzamos lejos del mínimo)

**Iteración 1**:
- Derivada: dJ/dθ = 2(θ - 3) = 2(10 - 3) = **14**
- Con α = 0.5: θ_nuevo = 10 - 0.5 × 14 = **3** ✅
- ¡Hemos alcanzado el mínimo en una iteración!

**Iteración 2** (verificación):
- Derivada: dJ/dθ = 2(3 - 3) = **0**
- Con α = 0.5: θ_nuevo = 3 - 0.5 × 0 = **3**
- El valor no cambia → **Convergencia alcanzada**

#### Visualizando Gradient Descent

````python
# Visualización de cómo funciona Gradient Descent

import numpy as np
import matplotlib.pyplot as plt

# Definir la función de coste simple: J(theta) = (theta - 3)^2
def cost_simple(theta):
    return (theta - 3) ** 2

# Definir la derivada: dJ/dtheta = 2(theta - 3)
def derivative_simple(theta):
    return 2 * (theta - 3)

# Implementar Gradient Descent
def gradient_descent(theta_init, learning_rate, iterations):
    theta = theta_init
    history = [theta]
    cost_history = [cost_simple(theta)]
    
    for i in range(iterations):
        gradient = derivative_simple(theta)
        theta = theta - learning_rate * gradient
        history.append(theta)
        cost_history.append(cost_simple(theta))
        
        if abs(gradient) < 1e-6:  # Convergencia
            break
    
    return np.array(history), np.array(cost_history)

# Ejecutar Gradient Descent
theta_init = 10
learning_rate = 0.5
iterations = 10
theta_history, cost_history = gradient_descent(theta_init, learning_rate, iterations)

# Crear visualización
fig, axes = plt.subplots(1, 2, figsize=(15, 5))

# Gráfico 1: La función de coste y el camino del Gradient Descent
ax1 = axes[0]
theta_range = np.linspace(-2, 12, 100)
cost_range = cost_simple(theta_range)

ax1.plot(theta_range, cost_range, 'b-', linewidth=2.5, label='J(θ) = (θ - 3)²')
ax1.plot(theta_history, cost_history, 'ro-', linewidth=2, markersize=8, label='Pasos de Gradient Descent')
ax1.plot(3, 0, 'g*', markersize=20, label='Mínimo óptimo')

# Anotaciones de pasos
for i, (theta, cost) in enumerate(zip(theta_history[:5], cost_history[:5])):
    ax1.annotate(f'Paso {i}', xy=(theta, cost), xytext=(5, 5), 
                textcoords='offset points', fontsize=9, 
                bbox=dict(boxstyle='round,pad=0.3', facecolor='yellow', alpha=0.5))

ax1.set_xlabel('θ (parámetro)', fontsize=12)
ax1.set_ylabel('J(θ) (función de coste)', fontsize=12)
ax1.set_title('Gradient Descent: Bajando Hacia el Mínimo', fontsize=13, fontweight='bold')
ax1.legend()
ax1.grid(True, alpha=0.3)

# Gráfico 2: Evolución del parámetro y el coste
ax2 = axes[1]
iterations_range = range(len(theta_history))

ax2_twin = ax2.twinx()

line1 = ax2.plot(iterations_range, theta_history, 'b-o', linewidth=2, markersize=6, label='θ (parámetro)')
line2 = ax2_twin.plot(iterations_range, cost_history, 'r-s', linewidth=2, markersize=6, label='J(θ) (coste)')

ax2.axhline(y=3, color='g', linestyle='--', alpha=0.5, label='θ óptimo = 3')
ax2.set_xlabel('Iteración', fontsize=12)
ax2.set_ylabel('Valor de θ', fontsize=12, color='b')
ax2_twin.set_ylabel('Coste J(θ)', fontsize=12, color='r')
ax2.set_title('Convergencia: θ y Coste en el Tiempo', fontsize=13, fontweight='bold')
ax2.tick_params(axis='y', labelcolor='b')
ax2_twin.tick_params(axis='y', labelcolor='r')
ax2.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()

# Mostrar los valores
print("📊 Resumen del Gradient Descent:")
print(f"Valor inicial de θ: {theta_init}")
print(f"Valor final de θ: {theta_history[-1]:.6f}")
print(f"Valor óptimo (teórico): 3")
print(f"Coste inicial: {cost_history[0]:.6f}")
print(f"Coste final: {cost_history[-1]:.6f}")
print(f"Iteraciones necesarias: {len(theta_history) - 1}")
````

#### ¿Qué Hace el Learning Rate?

El learning rate (α) es crucial:

- **α muy pequeño** (ej: 0.001): Pasos muy pequeños → Convergencia lenta, pero más precisa.
- **α muy grande** (ej: 1.0): Pasos muy grandes → Puede saltarse el mínimo, diverge.
- **α óptimo** (ej: 0.01-0.1): Balance entre velocidad y precisión.

#### Gradient Descent en Regresión Lineal

En nuestro problema de precios de viviendas:

1. **Inicializamos** θ₀ y θ₁ aleatoriamente.
2. **Calculamos** las derivadas de J respecto a θ₀ y θ₁.
3. **Actualizamos** ambos parámetros simultáneamente.
4. **Repetimos** hasta que los cambios sean insignificantes.

El resultado: encontramos automáticamente la línea que mejor predice los precios.

#### ¿Por Qué Funciona?

✅ **Es automático**: No tenemos que escribir código para "buscar" el mínimo.  
✅ **Es escalable**: Funciona igual con 1 variable que con 1.000 variables.  
✅ **Es garantizado**: Para funciones convexas (como MSE - Error Cuadrático Medio), siempre encuentra el mínimo global.  

#### El Flujo Completo del Algoritmo

1. **Función Hipótesis**: Define cómo predecimos.
2. **Función de Coste**: Mide cuán mal predecimos.
3. **Gradient Descent**: Ajusta los parámetros para minimizar el coste.
4. **Resultado**: Modelo entrenado listo para hacer predicciones.

## 🎭 5. Overfitting y Underfitting <a id='5-over-under'></a>

Estos son dos de los problemas más comunes y fundamentales en Machine Learning.

- **Underfitting (Subajuste)**: El modelo es **demasiado simple** y no captura la estructura de los datos. Funciona mal tanto en los datos de entrenamiento como en los de prueba. Es como un estudiante que no ha estudiado lo suficiente.

- **Overfitting (Sobreajuste)**: El modelo es **demasiado complejo** y se aprende los datos de entrenamiento **de memoria**, incluyendo el ruido. Funciona muy bien en los datos de entrenamiento, pero **muy mal** en los datos de prueba (no generaliza). Es como un estudiante que memoriza las respuestas del examen pero no entiende los conceptos.

**El objetivo es encontrar un modelo equilibrado (Good Fit) que generalice bien a datos nuevos.**

![Overfitting](imagenes/overfitting.png)

![Overfitting](imagenes/overfitting1.png)

````python
# Ejemplo visual de Overfitting
from sklearn.preprocessing import PolynomialFeatures
from sklearn.pipeline import make_pipeline

# Generar datos con una relación no lineal
np.random.seed(0) # Fijar semilla para reproducibilidad
X_poly = np.sort(5 * np.random.rand(80, 1), axis=0) # 80 puntos entre 0 y 5
# y = sin(x) con algo de ruido
y_poly = np.sin(X_poly).ravel() + np.random.randn(80) * 0.1 # Añadir ruido

plt.figure(figsize=(14, 6)) # Crear figura para los subplots

# Modelos con diferentes grados de complejidad
degrees = [1, 4, 15] # Grados del polinomio para comparar
for i, degree in enumerate(degrees):
    ax = plt.subplot(1, len(degrees), i + 1) # Crear subplot
    
    # Crear y entrenar el pipeline del modelo
    model = make_pipeline(PolynomialFeatures(degree), LinearRegression()) # Pipeline de características polinomiales y regresión lineal
    model.fit(X_poly, y_poly) # Entrenar el modelo
    
    # Predecir y graficar
    y_pred = model.predict(X_poly) # Hacer predicciones
    plt.scatter(X_poly, y_poly, edgecolor='b', s=20, label='Datos Reales') # Datos reales
    plt.plot(X_poly, y_pred, color='red', linewidth=2, label='Modelo') # Línea del modelo
    plt.xlabel('x')
    plt.ylabel('y')
    plt.legend()
    
    title = f"Grado {degree}"
    if degree == 1: title += ' (Underfitting)'
    if degree == 4: title += ' (Good Fit)'
    if degree == 15: title += ' (Overfitting)'
    plt.title(title)

plt.tight_layout()
plt.show()
````

## 🔄 6. Validación Cruzada (Cross-Validation) <a id='6-cross-val'></a>

El `train_test_split` es bueno, pero el resultado puede depender de *cómo* se hizo la división. Si por casualidad en el conjunto de prueba caen los datos más "fáciles", nuestro modelo parecerá mejor de lo que es.

La **Validación Cruzada (K-Fold Cross-Validation)** es una técnica más robusta para evaluar un modelo.

**¿Cómo funciona?**
1. Divide los datos en **K** "pliegues" (folds) iguales (ej: K=5).
2. En cada iteración, usa **1 pliegue** para **prueba** y los **K-1 restantes** para **entrenamiento**.
3. Repite el proceso K veces, usando un pliegue diferente para la prueba en cada ocasión.
4. El rendimiento final del modelo es la **media** de los rendimientos obtenidos en los K pliegues.

![Cross-Validation](imagenes/cross-validation.png)

**Ventajas**: Usa todos los datos tanto para entrenar como para probar, dando una estimación mucho más fiable del rendimiento del modelo en datos nuevos.

````python
from sklearn.model_selection import cross_val_score # Importar función para validación cruzada

# Usaremos los datos de las casas de nuevo
X = casas_df[['Tamaño']]
y = casas_df['Precio']

# Crear el modelo
modelo_cv = LinearRegression() # Instancia del modelo de regresión lineal

# Realizar validación cruzada con 5 folds (cv=5)
# 'scoring' especifica la métrica a usar. 'neg_mean_absolute_error' es el error absoluto medio (en negativo porque sklearn busca maximizar)
scores = cross_val_score(modelo_cv, X, y, cv=5, scoring='neg_mean_absolute_error')

print("Errores de cada uno de los 5 folds:", -scores)
print(f"\nError Absoluto Medio (promedio de CV): {-scores.mean():.2f}€")
print(f"Desviación estándar de los errores: {-scores.std():.2f}€")
````

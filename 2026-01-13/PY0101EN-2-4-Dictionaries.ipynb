## ¿Qué son los Diccionarios?


Un diccionario consta de claves y valores. Es útil comparar un diccionario con una lista. En lugar de estar indexados numéricamente como una lista, los diccionarios tienen claves. Estas claves son las que se utilizan para acceder a los valores dentro de un diccionario.   


El mejor ejemplo de un diccionario puede ser acceder a los detalles de una persona utilizando el **número de seguridad social**.   
Aquí, el número de seguridad social, que es un número único, será la **clave** y los detalles de las personas serán los **valores** asociados con él.

### Crear un Diccionario y acceder a los elementos


Un ejemplo de un Diccionario <code>Dict</code>:
Aquí estamos creando un diccionario llamado **Dict** con los siguientes detalles

* Las claves son **key1, key2, key3, key4, key5, (0,1)**.
* Los valores son {1,2,[3,3,3],(4,4,4),5,6} correspondientes a las claves


````python
# Crear el diccionario

Dict = {"key1": 1, "key2": "2", "key3": [3, 3, 3], "key4": (4, 4, 4), ('key5'): 5, (0, 1): 6}
Dict
````

Las claves pueden ser cadenas:


````python
# Acceso al valor por la clave

Dict["key1"]
````

Las claves también pueden ser cualquier objeto inmutable como una tupla:


````python
# Acceso al valor por la clave

Dict[(0, 1)]
````

Cada clave está separada de su valor por dos puntos "<code>:</code>". Las comas separan los elementos, y todo el diccionario está encerrado entre llaves. Un diccionario vacío sin elementos se escribe con solo dos llaves, como esto "<code>{}</code>".


````python
# Crear un diccionario de ejemplo

release_year_dict = {"Thriller": "1982", "Back in Black": "1980", \
                    "The Dark Side of the Moon": "1973", "The Bodyguard": "1992", \
                    "Bat Out of Hell": "1977", "Their Greatest Hits (1971-1975)": "1976", \
                    "Saturday Night Fever": "1977", "Rumours": "1977"}
release_year_dict
````

En resumen, como una lista, un diccionario contiene una secuencia de elementos. Cada elemento está representado por una clave y su valor correspondiente. Los diccionarios se crean con dos llaves que contienen claves y valores separados por dos puntos. Para cada clave, solo puede haber un valor único, sin embargo, múltiples claves pueden contener el mismo valor. Las claves solo pueden ser cadenas, números o tuplas, pero los valores pueden ser cualquier tipo de dato.


Es útil visualizar el diccionario como una tabla, como en la siguiente imagen. La primera columna representa las claves, la segunda columna representa los valores.

## Claves (Keys)


Puedes recuperar los valores basándote en los nombres:


````python
# Conocer el valor por la clave

release_year_dict['Thriller'] 
````

Esto corresponde a:


<img src="https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/IBMDeveloperSkillsNetwork-PY0101EN-SkillsNetwork/labs/Module%202/images/DictsKeyOne.png" width="500">


De manera similar para <b>The Bodyguard</b>


````python
# Conocer el valor por la clave

release_year_dict['The Bodyguard'] 
````

<img src="https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/IBMDeveloperSkillsNetwork-PY0101EN-SkillsNetwork/labs/Module%202/images/DictsKeyTwo.png" width="500">


Ahora recuperemos las claves del diccionario usando el método <code>keys()</code>:


````python
# Conocer todas las claves en el diccionario

release_year_dict.keys() 
````

Puedes recuperar los valores usando el método <code>values()</code>:


````python
# Conocer todos los valores en el diccionario

release_year_dict.values() 
````

Podemos agregar una entrada:


````python
# Añadir un valor con clave al diccionario

release_year_dict['Graduation'] = '2007'
release_year_dict
````

Podemos eliminar una entrada:


````python
# Borrar entradas por clave

del(release_year_dict['Thriller'])
del(release_year_dict['Graduation'])
release_year_dict
````

Podemos verificar si un elemento está en el diccionario:


````python
# Verificar si la clave está en el diccionario

'The Bodyguard' in release_year_dict
````

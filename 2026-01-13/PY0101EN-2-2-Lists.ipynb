## Indexación


Vamos a echar un vistazo a las listas en Python. Una lista es una colección secuenciada de diferentes objetos como enteros, cadenas e incluso otras listas. La dirección de cada elemento dentro de una lista se llama <b>índice</b>. Un índice se usa para acceder y referirse a elementos dentro de una lista.


<img src="https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/IBMDeveloperSkillsNetwork-PY0101EN-SkillsNetwork/labs/Module%202/images/ListsIndex.png" width="1000">


Para crear una lista, escribe la lista dentro de corchetes <b>\[ ]</b>, con tu contenido dentro del paréntesis y separado por comas. ¡Intentémoslo!


````python
# Crear una lista

L = ["The Bodyguard", 7.0, 1992]
L
````

Podemos usar indexación negativa y regular con una lista:


<img src="https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/kkMiHZTtmXzY4BWInK5Azg/Slide1.PNG" width="1000">


````python
# Imprimir los elementos en cada índiceex

print('el mismo elemento usando indexación negativa y positiva:\n Positivo:',L[0],
'\n Negativo:' , L[-3]  )
print('el mismo elemento usando indexación negativa y positiva:\n Positivo:',L[1],
'\n Negativo:' , L[-2]  )
print('el mismo elemento usando indexación negativa y positiva:\n Positivo:',L[2],
'\n Negativo:' , L[-1]  )
````

<a id="content"></a>
## Contenido de la Lista


Las listas pueden contener cadenas, flotantes y enteros. Podemos anidar otras listas, y también podemos anidar tuplas y otras estructuras de datos. Las mismas convenciones de indexación se aplican para el anidamiento:


````python
# Ejemplo de lista con diferentes tipos de datos

["The Bodyguard", 7.0, 1992, [1, 2], ("A", 1)]
````

<a id="op"></a>
## Operaciones de Lista


También podemos realizar segmentación en listas. Por ejemplo, si queremos los últimos dos elementos, usamos el comando siguiente:


````python
# Ejemplo de lista con diferentes tipos de datos

L = ["The Bodyguard", 7.0,1992,"BG",1]
L
````

<img src="https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/KyRT0sxnEHkVDzA5XS6h0w/Slide2.PNG" width="1000">


````python
# Segmento de lista (List slicing)

L[3:5]
````

Podemos usar el método <code>extend</code> para agregar nuevos elementos a la lista:


````python
# Usar extend para agregar elementos a la lista

L = [ "The Bodyguard", 7.0]
L.extend(['pop', 10])
L
````

Otro método similar es <code>append</code>. Si aplicamos <code>append</code> en lugar de <code>extend</code>, agregamos un elemento a la lista:


````python
# Usar append para agregar elementos a la lista

L = [ "The Bodyguard", 7.0]
L.append(['pop', 10])
L
````

Cada vez que aplicamos un método, la lista cambia. Si aplicamos <code>extend</code> agregamos dos nuevos elementos a la lista. La lista <code>L</code> se modifica agregando dos nuevos elementos:


````python
# Usar extend para agregar elementos a la lista

L = [ "The Bodyguard", 7.0]
L.extend(['pop', 10])
L
````

Si adjuntamos la lista <code>\['a','b']</code> tenemos un nuevo elemento que consiste en una lista anidada:


````python
# Usar append para agregar elementos a la lista

L.append(['a','b'])
L
````

Como las listas son mutables, podemos cambiarlas. Por ejemplo, podemos cambiar el primer elemento de la siguiente manera:


````python
# Cambiar el elemento basado en el índice

A = ["disco", 10, 1.2]
print('Antes del cambio:', A)
A[0] = 'hard rock'
print('Después del cambio:', A)
````

También podemos eliminar un elemento de una lista usando el comando <code>del</code>:


````python
# Borrar el elemento basado en el índice

print('Antes del cambio:', A)
del(A[0])
print('Después del cambio:', A)
````

Podemos convertir una cadena a una lista usando <code>split</code>. Por ejemplo, el método <code>split</code> traduce cada grupo de caracteres separados por un espacio en un elemento en una lista:


````python
# Separa la cadena, el valor predeterminado es por espacio

'hard rock'.split()
````

Podemos usar la función split para separar cadenas en un carácter específico que llamamos **delimitador**. Pasamos el carácter en el que nos gustaría dividir en el argumento, que en este caso es una coma. El resultado es una lista, y cada elemento corresponde a un conjunto de caracteres que han sido separados por una coma:


````python
# Separa la cadena por coma

'A,B,C,D'.split(',')
````

<a id="co"></a>
## Copiar y Clonar Listas


Cuando establecemos una variable <b>B</b> igual a <b>A</b>, tanto <b>A</b> como <b>B</b> hacen referencia a la misma lista en la memoria:


````python
# Copiar (copiar por referencia) la lista A

A = ["hard rock", 10, 1.2]
B = A
print('A:', A)
print('B:', B)
````

<img src="https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/IBMDeveloperSkillsNetwork-PY0101EN-SkillsNetwork/labs/Module%202/images/ListsRef.png" width="1000" align="center">


Inicialmente, el valor del primer elemento en <b>B</b> se establece como "hard rock". Si cambiamos el primer elemento en <b>A</b> a <b>"banana"</b>, obtenemos un efecto secundario inesperado. Como <b>A</b> y <b>B</b> hacen referencia a la misma lista, si cambiamos la lista <b>A</b>, la lista <b>B</b> también cambia. Si verificamos el primer elemento de <b>B</b> obtenemos "banana" en lugar de "hard rock":


````python
# Examinar la copia por referencia

print('B[0]:', B[0])
A[0] = "banana"
print('B[0]:', B[0])
````

Esto se demuestra en la siguiente figura:


<img src="https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/IBMDeveloperSkillsNetwork-PY0101EN-SkillsNetwork/labs/Module%202/images/ListsRefGif.gif" width="1000">


Puedes clonar la lista **A** usando la siguiente sintaxis:


````python
# Clonar (clonar por valor) la lista A

B = A[:]
B
````

Variable **B** references a new copy or clone of the original list. Esto se demuestra en la siguiente figura:


<img src="https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/IBMDeveloperSkillsNetwork-PY0101EN-SkillsNetwork/labs/Module%202/images/ListsVal.gif" width="1000">


Ahora, si cambias <b>A</b>, <b>B</b> no cambiará:


````python
print('B[0]:', B[0])
A[0] = "hard rock"
print('B[0]:', B[0])
````

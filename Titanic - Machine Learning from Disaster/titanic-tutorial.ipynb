# Parte 3: Tu primera presentación

Recuerda nuestro objetivo: queremos encontrar patrones en **train.csv** que nos ayuden a predecir si los pasajeros en **test.csv** sobrevivieron.

Inicialmente puede parecer abrumador buscar patrones, cuando hay tantos datos para ordenar. Así que comenzaremos de manera simple.

## Explorar un patrón

Recuerda que el archivo de presentación de muestra en **gender_submission.csv** asume que todas las pasajeras mujeres sobrevivieron (y todos los pasajeros hombres murieron).  

¿Es esta una primera suposición razonable? Verificaremos si este patrón se mantiene verdadero en los datos (en **train.csv**).

Copia el código a continuación en una nueva celda de código. Luego, ejecuta la celda.

```python
women = train_data.loc[train_data.Sex == 'female']["Survived"]
rate_women = sum(women)/len(women)

print("% de mujeres que sobrevivieron:", rate_women)
```

Antes de continuar, asegúrate de que tu código devuelva la salida anterior. El código anterior calcula el porcentaje de pasajeras mujeres (en **train.csv**) que sobrevivieron.

Luego, ejecuta el código a continuación en otra celda de código:

```python
men = train_data.loc[train_data.Sex == 'male']["Survived"]
rate_men = sum(men)/len(men)

print("% de hombres que sobrevivieron:", rate_men)
```

El código anterior calcula el porcentaje de pasajeros hombres (en **train.csv**) que sobrevivieron.

A partir de esto puedes ver que casi el 75% de las mujeres a bordo sobrevivieron, mientras que solo el 19% de los hombres vivieron para contarlo. Dado que el género parece ser un indicador tan fuerte de supervivencia, ¡el archivo de presentación en **gender_submission.csv** no es una mala primera suposición!

Pero al final del día, esta presentación basada en género basa sus predicciones en solo una única columna. Como puedes imaginar, al considerar múltiples columnas, podemos descubrir patrones más complejos que potencialmente pueden generar predicciones mejor informadas. Dado que es bastante difícil considerar varias columnas a la vez (o, tomaría mucho tiempo considerar todos los posibles patrones en muchas columnas diferentes simultáneamente), usaremos el aprendizaje automático para automatizar esto por nosotros.

## Tu primer modelo de aprendizaje automático

Construiremos lo que se conoce como un modelo de **bosque aleatorio** (random forest). Este modelo está construido de varios "árboles" (hay tres árboles en la imagen a continuación, ¡pero construiremos 100!) que considerarán individualmente los datos de cada pasajero y votarán sobre si el individuo sobrevivió. Luego, el modelo de bosque aleatorio toma una decisión democrática: ¡el resultado con más votos gana!

![](https://i.imgur.com/AC9Bq63.png)

La celda de código a continuación busca patrones en cuatro columnas diferentes (**"Pclass"**, **"Sex"**, **"SibSp"**, y **"Parch"**) de los datos. Construye los árboles en el modelo de bosque aleatorio basándose en patrones en el archivo **train.csv**, antes de generar predicciones para los pasajeros en **test.csv**. El código también guarda estas nuevas predicciones en un archivo CSV **submission.csv**.

Copia este código en tu notebook, y ejecútalo en una nueva celda de código.

```python
from sklearn.ensemble import RandomForestClassifier # Importamos el modelo

y = train_data["Survived"] # Etiqueta que queremos predecir

features = ["Pclass", "Sex", "SibSp", "Parch"] # Características que usaremos para predecir si alguien sobrevivió
X = pd.get_dummies(train_data[features]) # Convertimos las variables categóricas en variables dummy/indicadoras
X_test = pd.get_dummies(test_data[features]) # Convertimos las variables categóricas en variables dummy/indicadoras

model = RandomForestClassifier(n_estimators=100, max_depth=5, random_state=1) # Definimos el modelo
# n_estimators es el número de árboles en el bosque, si cambiamos este valor, el modelo puede mejorar o empeorar
# max_depth es la profundidad máxima de cada árbol, si cambiamos este valor, el modelo puede mejorar o empeorar
# random_state es la semilla utilizada por el generador de números aleatorios, si cambiamos este valor, el modelo puede mejorar o empeorar
# La elección de estos valores es un tema complejo que no cubriremos en este tutorial
model.fit(X, y) # Entrenamos el modelo
predictions = model.predict(X_test) # Hacemos predicciones sobre el conjunto de prueba

output = pd.DataFrame({'PassengerId': test_data.PassengerId, 'Survived': predictions}) # Preparamos el archivo de salida
output.to_csv('submission.csv', index=False) # Guardamos el archivo de salida
print("¡Tu envío fue guardado correctamente!") 
```

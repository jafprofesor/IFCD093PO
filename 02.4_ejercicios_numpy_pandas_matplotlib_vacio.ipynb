### Ejercicio 3.5 — Subplots

Crea una figura con 4 subplots (2x2):

2. Gráfico de barras
3. Histograma
4. Scatter plot

````python
# Tu código aquí
plt.subplots(2, 2, figsize=(10, 8))
# Subplot 1: Línea (seno)
x = np.linspace(0, 2 * np.pi, 100)
y = np.sin(x)
plt.subplot(2, 2, 1)
plt.plot(x, y, label='y = sin(x)', color='blue')
plt.title('Gráfico de Seno')
plt.xlabel('x')
plt.ylabel('y')
plt.legend()
plt.grid()

# Subplot 2: Barras
x_barras = ['A', 'B', 'C', 'D']
y_barras = [10, 15, 7, 12]
plt.subplot(2, 2, 2)
plt.bar(x_barras, y_barras, color='orange')
plt.title('Gráfico de Barras')
plt.xlabel('Categorías')
plt.ylabel('Valores')
plt.grid()  

# Subplot 3: Histograma
datos_histograma = np.random.randn(1000)
plt.subplot(2, 2, 3)
plt.hist(datos_histograma, bins=50, color='green', alpha=0.7)
plt.title('Histograma de Datos Aleatorios')
plt.xlabel('Valor')
plt.ylabel('Frecuencia')
plt.grid()

# Subplot 4: Scatter
puntos_x = np.random.rand(50)
puntos_y = np.random.rand(50)
plt.subplot(2, 2, 4)
plt.scatter(puntos_x, puntos_y, color='purple')
plt.title('Gráfico de Dispersión')
plt.xlabel('Eje X')
plt.ylabel('Eje Y')
plt.grid()
plt.tight_layout()

plt.show()
````

### Análisis de ventas mensual

Combina NumPy, Pandas y Matplotlib para realizar un análisis completo de ventas.

````python
# Tu código aquí

# 1. Crear datos de ventas mensuales
datos_ventas_mensuales = {
    'Mes': ['Enero', 'Febrero', 'Marzo', 'Abril', 'Mayo', 'Junio'],
    'Producto_A': [1500, 1800, 1700, 1600, 2000, 2200],
    'Producto_B': [1200, 1300, 1250, 1400, 1500, 1600]
}

# 2. Crear DataFrame con los datos
df_ventas = pd.DataFrame(datos_ventas_mensuales)

# 3. Agregar columna de ventas totales
df_ventas['Total'] = df_ventas['Producto_A'] + df_ventas['Producto_B']

# 4. Calcular estadísticas
suma_total_ventas = df_ventas['Total'].sum()
promedio_mensual_ventas = df_ventas['Total'].mean()
desviacion_estandar_ventas = df_ventas['Total'].std()
print(f"Suma total de ventas: {suma_total_ventas}")
print(f"Promedio mensual de ventas: {promedio_mensual_ventas}")
print(f"Desviación estándar de ventas: {desviacion_estandar_ventas}") 
# mes_max_ventas = df_ventas['Total'].max() no nos proporciona el mes
mes_max_ventas = df_ventas.loc[df_ventas['Total'].idxmax(), 'Mes']
print(f"Mes con mayores ventas: {mes_max_ventas}")
# Crecimiento porcentual del Producto A
crecimiento_producto_A = df_ventas['Producto_A'].pct_change() * 100
print("Crecimiento porcentual del Producto A:")
print(crecimiento_producto_A)
# Crecimiento porcentual del Producto B
crecimiento_producto_B = df_ventas['Producto_B'].pct_change() * 100
print("Crecimiento porcentual del Producto B:")
print(crecimiento_producto_B)
# Imprimir resultados
````

````python
# 5. Crear visualizaciones

# a) Gráfico de líneas comparando ambos productos
plt.figure(figsize=(10, 6))
plt.plot(df_ventas['Mes'], df_ventas['Producto_A'], marker='X', label='Producto A', color='blue')
plt.plot(df_ventas['Mes'], df_ventas['Producto_B'], marker='o', label='Producto B', color='green')
plt.title('Ventas Mensuales de Productos A y B')
plt.xlabel('Mes')
plt.ylabel('Ventas')
plt.legend()
plt.grid(True)
plt.show()
````

````python
# b) Gráfico de barras con ventas totales por mes
plt.figure(figsize=(10, 6))
plt.bar(df_ventas['Mes'], df_ventas['Total'], color='purple', alpha=0.7)
plt.title('Ventas Totales por Mes')
plt.xlabel('Mes')
plt.ylabel('Ventas Totales')
plt.show()
````

````python
# c) Gráfico de barras agrupadas (productos lado a lado)
x = np.arange(len(df_ventas['Mes']))  # Posiciones en el eje x
width = 0.40  # Ancho de las barras 
plt.figure(figsize=(10, 6))
plt.bar(x - width/2, df_ventas['Producto_A'], width, label='Producto A', color='blue')
# x - width/2 para Producto A y x + width/2 para Producto B
plt.bar(x + width/2, df_ventas['Producto_B'], width, label='Producto B', color='green')
plt.xticks(x, df_ventas['Mes']) # Etiquetas de los meses
plt.title('Ventas Mensuales de Productos A y B (Agrupadas)')
plt.xlabel('Mes')
plt.ylabel('Ventas')
plt.legend()
plt.show()  
````

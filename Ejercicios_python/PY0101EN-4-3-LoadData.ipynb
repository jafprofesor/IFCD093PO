Después del comando de importación, ahora tenemos acceso a una gran cantidad de clases y funciones preconstruidas. Esto asume que la librería está instalada; en nuestro entorno de laboratorio, todas las librerías necesarias están instaladas. Una forma en que pandas te permite trabajar con datos es un un dataframe. Repasemos el proceso para pasar de un archivo de valores separados por comas (<b>.csv</b>) a un dataframe. Esta variable <code>csv_path</code> almacena la ruta del <b>.csv</b>, que se utiliza como argumento para la función <code>read_csv</code>. El resultado se almacena en el objeto <code>df</code>, esta es una forma abreviada común utilizada para una variable que se refiere a un dataframe de Pandas.


````python
# Leer datos desde archivo CSV

# csv_path = 'https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/LXjSAttmoxJfEG6il1Bqfw/Product-sales.csv'
# df = pd.read_csv(csv_path)

from pyodide.http import pyfetch
import pandas as pd

filename = "https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/LXjSAttmoxJfEG6il1Bqfw/Product-sales.csv"

async def download(url, filename):
    response = await pyfetch(url)
    if response.status == 200:
        with open(filename, "wb") as f:
            f.write(await response.bytes())


await download(filename, "Product-sales.csv")
df = pd.read_csv("Product-sales.csv")
````

## Viendo Datos y Accediendo a Datos


También puedes obtener una columna como una serie. Puedes pensar en una serie de Pandas como un dataframe 1-D. Solo usa un corchete:


````python
#  Obtener la columna como una serie

x = df['Product']
x
````

También puedes obtener una columna como un dataframe. Por ejemplo, podemos asignar la columna <b>Quantity</b>:


````python
#  Obtener la columna como un dataframe

x = df[['Quantity']]
type(x)
````

Puedes hacer lo mismo para múltiples columnas; simplemente colocamos el nombre del dataframe, en este caso, <code>df</code>, y el nombre de los encabezados de múltiples columnas encerrados entre corchetes dobles. El resultado es un nuevo dataframe compuesto por las columnas especificadas:


````python
#  Acceso a múltiples columnas

y = df[['Product','Category', 'Quantity']]
y
````

El proceso se muestra en la figura:

Una forma de acceder a elementos únicos es el método <code>iloc</code>, donde puedes acceder a la 1ª fila y la 1ª columna de la siguiente manera:


````python
#  Acceder al valor en la primera fila y la primera columna

df.iloc[0, 0]
````

Puedes acceder a la 2ª fila y la 1ª columna de la siguiente manera:


````python
#  Acceder al valor en la segunda fila y la primera columna

df.iloc[1,0]
````

Puedes acceder a la 1ª fila y la 3ª columna de la siguiente manera:


````python
#  Acceder al valor en la primera fila y la tercera columna

df.iloc[0,2]
````

````python
#  Acceder al valor en la segunda fila y la tercera columna
df.iloc[1,2]
````

## Secuencias de Escape


Las barras invertidas representan el comienzo de secuencias de escape. Las secuencias de escape representan cadenas que pueden ser difíciles de ingresar. Por ejemplo, la barra invertida "n" representa una nueva línea. La salida se da por una nueva línea después de que se encuentra la barra invertida "n":


````python
# Nueva secuencia de escape de nueva línea

print(" El Guardaespaldas\n es el mejor álbum" )
````

De manera similar, la barra invertida "t" representa una tabulación:


````python
# Secuencia de escape de tabulación

print(" El\ Guardaespaldas \t es el mejor álbum" )
````

Si deseas colocar una barra invertida en tu cadena, usa una doble barra invertida:


````python
# Incluir barra invertida en la cadena

print(" El Guardaespaldas \\ es el mejor álbum" )
````

También podemos colocar una "r" antes de la cadena para mostrar la barra invertida:


````python
# r puede decirle a python que la cadena se mostrará como una cadena sin procesar

print(r" El Guardaespaldas \ es el mejor álbum" )
````

## Operaciones de Manipulación de Cadenas


Hay muchos métodos de operación de cadenas en Python que se pueden usar para manipular los datos. Vamos a usar algunas operaciones básicas de cadenas en los datos.


Probemos con el método <code>upper</code>; este método convierte caracteres en minúsculas a caracteres en mayúsculas:


````python
# Convierta todos los caracteres de la cadena a mayúsculas

a = "Thriller es el sexto álbum de estudio"
print("antes de mayúsculas:", a)
b = a.upper()
print("Después de mayúsculas:", b)
````

El método <code>replace</code> reemplaza un segmento de la cadena, es decir, una subcadena con una nueva cadena. Ingresamos la parte de la cadena que nos gustaría cambiar. El segundo argumento es con qué nos gustaría intercambiar el segmento, y el resultado es una nueva cadena con el segmento cambiado:

## RegEx (Expresiones Regulares)


En Python, RegEx (abreviatura de Expresión Regular) es una herramienta para coincidir y manejar cadenas. 


Este módulo RegEx proporciona varias funciones para trabajar con expresiones regulares, incluyendo <code>search, split, findall,</code> y <code>sub</code>. 


Python proporciona un módulo integrado llamado <code>re</code>, que te permite trabajar con expresiones regulares. 
Primero, importa el módulo <code>re</code>


````python
import re
````

La función search() busca patrones especificados dentro de una cadena. Aquí hay un ejemplo que explica cómo usar la función search() para buscar la palabra "Body" en la cadena "The BodyGuard is the best".


````python
frase_completa = "El Guardaespaldas es el mejor álbum"

# Definir el patrón a buscar
pattern = "Guardaespaldas"
# result =  s1.find(pattern)
# Usar la función search() para buscar el patrón en la cadena
result = re.search(pattern, frase_completa)
# Comprobar si se encontró una coincidencia
if result:
    print("¡Coincidencia encontrada!")
else:
    print("No se encontró ninguna coincidencia.")
````

Las expresiones regulares (RegEx) son patrones utilizados para coincidir y manipular cadenas de texto. Hay varias secuencias especiales en RegEx que se pueden usar para coincidir con caracteres o patrones específicos.

| Secuencia Especial | Significado             | 	Ejemplo             |
| -----------  | ----------------------- | ----------------------|
| \d|Coincide con cualquier carácter numérico (0-9)|"123" coincide con "\d\d\d"|
|\D|Coincide con cualquier carácter no numérico|"hello" coincide con "\D\D\D\D\D"|
|\w|Coincide con cualquier carácter de palabra (a-z, A-Z, 0-9 y _)|"hello_world" coincide con "\w\w\w\w\w\w\w\w\w\w\w"|
|\W|Coincide con cualquier carácter que no sea de palabra|	"@#$%" coincide con "\W\W\W\W"|
|\s|Coincide con cualquier carácter de espacio en blanco (espacio, tabulación, nueva línea, etc.)|"hello world" coincide con "\w\w\w\w\w\s\w\w\w\w\w"|
|\S|Coincide con cualquier carácter que no sea espacio en blanco|"hello_world" coincide con "\S\S\S\S\S\S\S\S\S\S\S"|
|\b|Coincide con el límite entre un carácter de palabra y un carácter que no es de palabra|"cat" coincide con "\bcat\b" en "The cat sat on the mat"|
|\B|Coincide con cualquier posición que no sea un límite de palabra|"cat" coincide con "\Bcat\B" en "category" pero no en "The cat sat on the mat"|


Ejemplos de Secuencias Especiales:

Un ejemplo simple de uso de la secuencia especial <code>\d</code> en un patrón de expresión regular con código Python:


````python
pattern = r"\d\d\d\d\d\d\d\d\d"  #  Coincide con cualquier diez dígitos consecutivos
text = "Mi número de teléfono es 123467890"
match = re.search(pattern, text)

if match:
    print("Número de teléfono encontrado:", match.group())
else:
    print("No se encontró ninguna coincidencia")
````

El método match.group() se usa en el módulo re de Python para recuperar la parte de la cadena donde coincidió el patrón de expresión regular. Aquí hay una explicación detallada:

**Propósito**
- Extraer Texto Coincidente: match.group() devuelve la subcadena exacta que coincidió con el patrón.
 
**Uso**
- Cuando usas funciones como re.search() o re.match(), devuelven un objeto match si se encuentra el patrón. Luego puedes usar match.group() para obtener el texto coincidente.

Aquí `match.group()` recupera la subcadena 1234567890 del texto, que es la parte que coincidió con el patrón.


El patrón de expresión regular se define como r"\d\d\d\d\d\d\d\d\d\d", que usa la secuencia especial \d para coincidir con cualquier carácter numérico (0-9), y la secuencia \d se repite diez veces para coincidir con diez dígitos consecutivos


Un ejemplo simple de uso de la secuencia especial <code>\W</code> en un patrón de expresión regular con código Python:


````python
pattern = r"\W"  #  Coincide con cualquier carácter que no sea una palabra
text = "Hola, mundo!"
matches = re.findall(pattern, text)

print("Coincidencias:", matches)
````

El patrón de expresión regular se define como r"\W", que usa la secuencia especial \W para coincidir con cualquier carácter que no sea un carácter de palabra (a-z, A-Z, 0-9, o _). La cadena en la que buscamos coincidencias es "Hello, world!".


La función <code>findall()</code> encuentra todas las ocurrencias de un patrón especificado dentro de una cadena.


````python
s2 = "El 'The BodyGuard' es el mejor álbum de 'Whitney Houston'."


# Usar la función findall() para encontrar todas las ocurrencias de "st" en la cadena
result = re.findall("es", s2)

# Imprimir la lista de palabras coincidentes
print(result)
````

La función <code>split()</code> de una expresión regular divide una cadena en una matriz de subcadenas según un patrón especificado.


````python
# Usar la función split para dividir la cadena por el "\s"
split_array = re.split(r"\s", s2)

# El split_array contiene todas las subcadenas, divididas por caracteres de espacio en blanco
print(split_array)
````

Aquí hay una explicación detallada: 

<code>re.split("\s", s2)</code>:

**re.split**: Esta función divide una cadena por las ocurrencias de un patrón.
- **r"\s"**: Este es un patrón de expresión regular que coincide con cualquier carácter de espacio en blanco (espacios, tabulaciones, nuevas líneas, etc.).
- **s2**: Esta es la cadena que deseas dividir.


La función <code>sub</code> de una expresión regular en Python se usa para reemplazar todas las ocurrencias de un patrón en una cadena con un reemplazo especificado.


````python
s2 = "El 'The BodyGuard' es el mejor álbum de 'Whitney Houston' que es WHITNEY HOUSTON."
# Definir el patrón de expresión regular a buscar
pattern = r"Whitney Houston"

# Definir la cadena de reemplazo
replacement = "leyenda"

# Usar la función sub para reemplazar el patrón con la cadena de reemplazo
new_string = re.sub(pattern, replacement, s2, flags=re.IGNORECASE) # re.IGNORECASE hace que la búsqueda no distinga entre mayúsculas y minúsculas, por lo que coincide con "Whitney Houston" en cualquier caso de letra

# La new_string contiene la cadena original con el patrón reemplazado por la cadena de reemplazo
print(new_string) 
````

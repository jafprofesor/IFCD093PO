# Ingeniería de Datos Process


Hay varios pasos en el proceso de Ingeniería de Datos.

1.  **Extraer (Extract)** - La extracción de datos es la obtención de datos de múltiples fuentes. Ej. Extracción de datos de un sitio web mediante Web scraping o recopilación de información de los datos almacenados en diferentes formatos (JSON, CSV, XLSX, etc.).

# Formato de archivo de valores separados por comas (CSV)


El formato de archivo **Valores separados por comas** cae bajo un formato de archivo de hoja de cálculo.

En un formato de archivo de hoja de cálculo, los datos se almacenan en celdas. Cada celda se organiza en filas y columnas. Una columna en el archivo de hoja de cálculo puede tener diferentes tipos. Por ejemplo, una columna puede ser de tipo cadena, tipo fecha o tipo entero.

Cada línea en el archivo CSV representa una observación, o comúnmente llamada registro. Cada registro puede contener uno o más campos que están separados por una coma.

# Formato de archivo JSON


**JSON (Notación de Objetos JavaScript)** es un formato ligero de intercambio de datos. Es fácil de leer y escribir para los humanos.

JSON se basa en dos estructuras:

1.  Una colección de pares nombre/valor. En varios lenguajes, esto se realiza como un objeto, registro, estructura, diccionario, tabla hash, lista con clave o matriz asociativa.

2.  Una lista ordenada de valores. En la mayoría de los lenguajes, esto se realiza como una matriz, vector, lista o secuencia.

JSON es un formato de datos independiente del lenguaje. Se derivó de JavaScript, pero muchos lenguajes de programación modernos incluyen código para generar y analizar datos en formato JSON. Es un formato de datos muy común con una amplia gama de aplicaciones.


El texto en JSON se realiza a través de cadenas entre comillas que contienen los valores en asignaciones clave-valor dentro de { }. Es similar al diccionario en Python.


Python soporta JSON a través de un paquete integrado llamado **json**. Para usar esta característica, importamos el paquete json en el script de Python.


```python
import json
```

# Escribiendo JSON en un Archivo

Esto generalmente se llama **serialización**. Es el proceso de convertir un objeto en un formato especial que sea adecuado para transmitir a través de la red o almacenar en un archivo o base de datos.

Para manejar el flujo de datos en un archivo, la biblioteca JSON en Python usa la función **dump()** o **dumps()** para convertir los objetos de Python en su respectivo objeto JSON. Esto facilita la escritura de datos en archivos.


```python
import json
person = {
    'first_name' : 'Mark',
    'last_name' : 'abc',
    'age' : 27,
    'address': {
        "streetAddress": "21 2nd Street",
        "city": "New York",
        "state": "NY",
        "postalCode": "10021-3100"
    }
}
```

#### serialización usando la función dump()

El método **json.dump()** se puede usar para escribir en un archivo JSON.

Sintaxis: json.dump(dict, puntero_archivo)

Parámetros:

1.  **dictionary** – nombre del diccionario que debe convertirse a objeto JSON.
2.  **file pointer** – puntero del archivo abierto en modo de escritura o anexado.


```python
with open('person.json', 'w') as f:  # Escribiendo objeto JSON en un archivo
    json.dump(person, f)
```

#### serialización usando la función dumps()

**json.dumps()** ayuda a convertir un diccionario en un objeto JSON.

Toma dos parámetros:

1.  **dictionary** – nombre del diccionario que debe convertirse a objeto JSON.
2.  **indent** – define el número de unidades para la sangría


```python
#  Serializando json  
json_object = json.dumps(person, indent = 4) 
  
# Escribiendo en sample.json 
with open("sample.json", "w") as outfile: 
    outfile.write(json_object) 
```

```python
print(json_object)
```

Nuestros objetos Python ahora están serializados en el archivo. Para deserializarlo de nuevo al objeto Python, usamos la función load().


# Leyendo JSON de un Archivo

# Formato de archivo XML


**XML también se conoce como Lenguaje de Marcado Extensible**. Como su nombre indica, es un lenguaje de marcado. Tiene ciertas reglas para codificar datos. El formato de archivo XML es un formato de archivo legible por humanos y legible por máquinas.

Veremos cómo podemos usar otros módulos para leer datos de un archivo XML y cargarlos en un DataFrame de Pandas.


### Escribiendo con xml.etree.ElementTree


El módulo **xml.etree.ElementTree** viene integrado con Python. Proporciona funcionalidad para analizar y crear documentos XML. **ElementTree** representa el documento XML como un árbol. Podemos movernos a través del documento usando nodos que son elementos y subelementos del archivo XML.

Para obtener más información, lee la documentación de [xml.etree.ElementTree](https://docs.python.org/3/library/xml.etree.elementtree.html?utm_medium=Exinfluencer&utm_source=Exinfluencer&utm_content=000026UJ&utm_term=10006555&utm_id=NA-SkillsNetwork-Channel-SkillsNetworkCoursesIBMDeveloperSkillsNetworkPY0101ENSkillsNetwork19487395-2021-01-01).


```python
import xml.etree.ElementTree as ET

# Crear la estructura del archivo
employee = ET.Element('employee')
details = ET.SubElement(employee, 'details')
first = ET.SubElement(details, 'firstname')
second = ET.SubElement(details, 'lastname')
third = ET.SubElement(details, 'age')
first.text = 'Shiv'
second.text = 'Mishra'
third.text = '23'

# Crear un nuevo archivo XML con los resultados
mydata1 = ET.ElementTree(employee)
# myfile = open("items2.xml", "wb")
# myfile.write(mydata)
with open("new_sample.xml", "wb") as files:
    mydata1.write(files)
```

### Lectura con xml.etree.ElementTree


Echemos un vistazo a una forma de leer datos XML y ponerlos en un DataFrame de Pandas. Puedes ver el archivo XML en el Bloc de notas de tu máquina local.


```python
%wget https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/IBMDeveloperSkillsNetwork-PY0101EN-SkillsNetwork/labs/Module%205/data/Sample-employee-XML-file.xml

import xml.etree.ElementTree as etree

filename = "https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/IBMDeveloperSkillsNetwork-PY0101EN-SkillsNetwork/labs/Module%205/data/Sample-employee-XML-file.xml"

async def download(url, filename):
    response = await pyfetch(url)
    if response.status == 200:
        with open(filename, "wb") as f:
            f.write(await response.bytes())

await download(filename, "Sample-employee-XML-file.xml")
```

Primero necesitarías analizar un archivo XML y crear una lista de columnas para el dataframe, luego extraer información útil del archivo XML y agregarla a un dataframe de pandas.

Aquí hay un código de muestra que puedes usar.:


```python
# Parsear el archivo XML
tree = etree.parse("Sample-employee-XML-file.xml")

# Obtener la raíz del árbol XML
root = tree.getroot()

# Definir las columnas para el DataFrame
columns = ["firstname", "lastname", "title", "division", "building", "room"]

# Inicializar un DataFrame vacío
datatframe = pd.DataFrame(columns=columns)

# Iterar a través de cada nodo en la raíz del XML
for node in root:
    # Extraer texto de cada elemento
    firstname = node.find("firstname").text
    lastname = node.find("lastname").text
    title = node.find("title").text
    division = node.find("division").text
    building = node.find("building").text
    room = node.find("room").text
    
    # Crear un DataFrame para la fila actual
    row_df = pd.DataFrame([[firstname, lastname, title, division, building, room]], columns=columns)
    
    # Concatenar con el DataFrame existente
    datatframe = pd.concat([datatframe, row_df], ignore_index=True)

```

# Formato de Archivo Binario

Los archivos "binarios" son archivos donde el formato no está compuesto por caracteres legibles. Contienen información de formato que solo ciertas aplicaciones o procesadores pueden entender. Mientras que los humanos pueden leer archivos de texto, los archivos binarios deben ejecutarse en el software o procesador adecuado antes de que los humanos puedan leerlos.

Los archivos binarios pueden variar desde archivos de imagen como JPEGs o GIFs, archivos de audio como MP3s o formatos de documentos binarios como Word o PDF.

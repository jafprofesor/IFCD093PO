## Usando la Biblioteca yfinance para Extraer Datos de Acciones


Usando el módulo `Ticker` podemos crear un objeto que nos permitirá acceder a funciones para extraer datos. Para hacer esto necesitamos proporcionar el símbolo ticker para la acción, aquí la empresa es Apple y el símbolo ticker es `AAPL`.


```python
apple = yf.Ticker("AAPL")
```

Ahora podemos acceder a funciones y variables para extraer el tipo de datos que necesitamos. Puedes verlos y lo que representan aquí https://aroussi.com/post/python-yahoo-finance.

### Información de Acciones


Usando el atributo `info` podemos extraer información sobre la acción como un diccionario de Python.


```python
import json
with open('apple.json') as json_file:
    apple_info = json.load(json_file)
    # Print the type of data variable    
    #print("Type:", type(apple_info))
apple_info
```

Podemos obtener el <code>'country'</code> (país) usando la clave country


```python
apple_info['country']
```

### Extrayendo Precio de Acciones


Una acción es la parte única más pequeña de las acciones de una empresa que puedes comprar, los precios de estas acciones fluctúan con el tiempo. Usando el método `history()` podemos obtener el precio de la acción durante un cierto período de tiempo. Usando el parámetro `period` podemos establecer qué tan atrás desde el presente obtener datos. Las opciones para `period` son 1 día (1d), 5d, 1 mes (1mo), 3mo, 6mo, 1 año (1y), 2y, 5y, 10y, ytd, y max.


```python
apple_share_price_data = apple.history(period="max")
```

El formato en que se devuelven los datos es un DataFrame de Pandas. Con la `Date` (fecha) como índice, se dan `Open` (apertura), `High` (alto), `Low` (bajo), `Close` (cierre), `Volume` (volumen) y `Stock Splits` (divisiones de acciones) para cada día.


```python
apple_share_price_data.head()
```

Podemos restablecer el índice del DataFrame con la función `reset_index`. También establecemos el parámetro `inplace` en `True` para que el cambio tenga lugar en el mismo DataFrame.


```python
apple_share_price_data.reset_index(inplace=True)
```

Podemos graficar el precio `Open` contra la `Date`:


```python
apple_share_price_data.plot(x="Date", y="Open")
```

### Extrayendo Dividendos


Los dividendos son la distribución de las ganancias de una empresa a los accionistas. En este caso se definen como una cantidad de dinero devuelta por acción que posee un inversor. Usando la variable `dividends` podemos obtener un dataframe de los datos. El período de los datos viene dado por el período definido en la función `history`.


```python
apple.dividends
```

Podemos graficar los dividendos a lo largo del tiempo:


```python
apple.dividends.plot()
```

Las ventajas de usar APIs:

La desventaja de usar APIs:

Para generar una tabla con información sobre los usuarios, podemos escribir una función que contenga todos los parámetros deseables. Por ejemplo, nombre, género, ciudad, etc. Los parámetros dependerán de los requisitos de la prueba a realizar. Llamamos a los Métodos Get, enumerados al comienzo de este cuaderno. Luego, devolvemos un dataframe de pandas con los usuarios.


````python
def get_users():
    users =[]
     
    for user in RandomUser.generate_users(10):
        users.append({"Name":user.get_full_name(),"Gender":user.get_gender(),"City":user.get_city(),"State":user.get_state(),"Email":user.get_email(), "DOB":user.get_dob(),"Picture":user.get_picture()})
      
    return pd.DataFrame(users)     
````

````python
get_users()
````

````python
df1 = pd.DataFrame(get_users())  
````

Ahora tenemos un dataframe de *pandas* que se puede usar para cualquier propósito de prueba que pueda tener el probador.


## Ejemplo 2: API Fruityvice

Otra forma más común de usar APIs es a través de la biblioteca `requests`. El siguiente laboratorio, Solicitudes y HTTP, contendrá más información sobre las solicitudes.

Comenzaremos importando todas las bibliotecas requeridas.


````python
import requests
import json
````

Obtendremos los datos de la API [fruityvice](https://www.fruityvice.com) usando la función `requests.get("url")`. Los datos están en un formato json.


````python
data = requests.get("https://web.archive.org/web/20240929211114/https://fruityvice.com/api/fruit/all")
````

Recuperaremos los resultados usando la función `json.loads()`.


````python
results = json.loads(data.text)
````

Convertiremos nuestros datos json en un dataframe de *pandas*. 


````python
pd.DataFrame(results)
````

El resultado está en un formato json anidado. La columna 'nutrition' contiene múltiples subcolumnas, por lo que los datos deben 'aplanarse' o normalizarse.


````python
df2 = pd.json_normalize(results)
````

````python
df2
````

Veamos si podemos extraer alguna información de este dataframe. Quizás, necesitamos saber la familia y el género de una cereza.


````python
cherry = df2.loc[df2["name"] == 'Cherry']
(cherry.iloc[0]['family']) , (cherry.iloc[0]['genus'])
````

## Resumen de HTTP


Cuando tú, el **cliente**, accedes a una página web, tu navegador envía una solicitud **HTTP** al **servidor web** que aloja la página. Si no se menciona ningún archivo específico en la URL, el servidor generalmente busca un archivo predeterminado, como "<code>index.html</code>". Si tu solicitud es exitosa, el servidor enviará el objeto al cliente en una **respuesta HTTP**. Esto incluye información como el tipo del **recurso**, la longitud del **recurso** y otra información.

<p>
La siguiente figura representa el proceso. El círculo de la izquierda representa al cliente, el círculo de la derecha representa al servidor web. La tabla debajo del servidor web representa una lista de recursos almacenados en el servidor web. En este caso un archivo <code>HTML</code>, una imagen <code>png</code> y un archivo <code>txt</code>.
</p>
<p>
El protocolo <b>HTTP</b> te permite enviar y recibir información a través de la web, incluyendo páginas web, imágenes y otros recursos web. En este laboratorio, proporcionaremos una descripción general de la biblioteca Requests para interactuar con el protocolo <code>HTTP</code>. 
</p

## Localizador Uniforme de Recursos: URL


El localizador uniforme de recursos (URL) es la forma más popular de encontrar recursos en la web. Podemos dividir la URL en tres partes.

<ul>
    <li><b>Esquema</b>:- Este es el protocolo, para este laboratorio siempre será <code>http://</code>  </li>
    <li><b> Dirección de Internet o URL Base </b>:- Esto se usará para encontrar la ubicación, aquí hay algunos ejemplos: <code>www.ibm.com</code> y <code> www.gitlab.com </code> </li>
    <li><b>Ruta</b>:- Ubicación en el servidor web por ejemplo: <code>/images/IDSNlogo.png</code> </li>
</ul>


También puedes escuchar el término Identificador Uniforme de Recursos (URI), las URL son en realidad un subconjunto de los URI. Otro término popular es endpoint, esta es la URL de una operación proporcionada por un servidor web.

## Solicitud (Request)s in Python


Requests es una biblioteca de Python que te permite enviar solicitudes <code>HTTP/1.1</code> fácilmente. Podemos importar la biblioteca de la siguiente manera:


````python
import requests
````

También usaremos las siguientes bibliotecas:


````python
import os 
from PIL import Image
from IPython.display import IFrame
````

Puedes hacer una solicitud <code>GET</code> a través del método <code>get</code> a [www.ibm.com](http://www.ibm.com/?utm_source=Exinfluencer&utm_content=000026UJ&utm_id=NA-SkillsNetwork-Channel-SkillsNetworkCoursesIBMDeveloperSkillsNetworkPY0101ENSkillsNetwork19487395-2021-01-01&utm_medium=Exinfluencer&utm_term=10006555):


````python

url='https://www.ibm.com/'
r=requests.get(url)
````

Tenemos el objeto de respuesta <code>r</code>, este tiene información sobre la solicitud, como el estado de la solicitud. Podemos ver el código de estado usando el atributo <code>status_code</code>.


````python
r.status_code
````

Puedes ver los encabezados de la solicitud:


````python
print(r.request.headers)
````

Puedes ver el cuerpo de la solicitud, en la siguiente línea, como no hay cuerpo para una solicitud get obtenemos un <code>None</code>:


````python
print("request body:", r.request.body)
````

Puedes ver el encabezado de respuesta <code>HTTP</code> usando el atributo <code>headers</code>. Esto devuelve un diccionario de python de encabezados de respuesta <code>HTTP</code>.


````python
header=r.headers
print(r.headers)
````

Podemos obtener la fecha en que se envió la solicitud usando la clave <code>Date</code>.


````python
header['Date']
````

<code>Content-Type</code> indica el tipo de datos:


````python
header['Content-Type']
````

También puedes verificar la <code>encoding</code> (codificación):


````python
r.encoding
````

Como el <code>Content-Type</code> es <code>text/html</code> podemos usar el atributo <code>text</code> para mostrar el <code>HTML</code> en el cuerpo. Podemos revisar los primeros 100 caracteres:


````python
r.text[0:100]
````

## Solicitud Get con Parámetros de URL


Puedes usar el método <b>GET</b> para modificar los resultados de tu consulta, por ejemplo recuperando datos de una API. Enviamos una solicitud <b>GET</b> al servidor. Al igual que antes tenemos la <b>URL Base</b>, en la <b>Ruta</b> adjuntamos <code>/get</code>, esto indica que nos gustaría realizar una solicitud <code>GET</code>. 


La URL Base es para <code>[http://httpbin.org/](http://httpbin.org/?utm_medium=Exinfluencer&utm_source=Exinfluencer&utm_content=000026UJ&utm_term=10006555&utm_id=NA-SkillsNetwork-Channel-SkillsNetworkCoursesIBMDeveloperSkillsNetworkPY0101ENSkillsNetwork19487395-2021-01-01)</code> es un servicio simple de Solicitud y Respuesta HTTP. La <code>URL</code> en Python está dada por:


````python
url_get='http://httpbin.org/get'
````

Una <a href="https://en.wikipedia.org/wiki/Query_string?utm_medium=Exinfluencer&utm_source=Exinfluencer&utm_content=000026UJ&utm_term=10006555&utm_id=NA-SkillsNetwork-Channel-SkillsNetworkCoursesIBMDeveloperSkillsNetworkPY0101ENSkillsNetwork19487395-2021-01-01">cadena de consulta</a> es parte de un localizador uniforme de recursos (URL), esto envía otra información al servidor web. El inicio de la consulta es un <code>?</code>, seguido de una serie de pares de parámetro y valor, como se muestra en la tabla a continuación. El primer nombre de parámetro es <code>name</code> y el valor es <code>Joseph</code>. El segundo nombre de parámetro es <code>ID</code> y el valor es <code>123</code>. Cada par, parámetro y valor está separado por un signo igual, <code>=</code>.
La serie de pares está separada por el ampersand <code>&</code>.


<div class="alert alert-block alert-info" style="margin-top: 20px">
         <img src="https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/IBMDeveloperSkillsNetwork-PY0101EN-SkillsNetwork/labs/Module%205/images/query_string.png" width="500" align="center">
</div>


Para crear una cadena de Consulta, agrega un diccionario. Las claves son los nombres de los parámetros y los valores son el valor de la cadena de Consulta.


````python
payload={"name":"Joseph","ID":"123"}
````

Luego pasando el diccionario <code>payload</code> al parámetro <code>params</code> de la función <code> get()</code>:


````python
r=requests.get(url_get,params=payload)
````

Podemos imprimir la <code>URL</code> y ver el nombre y los valores.


````python
r.url
````

No hay cuerpo de solicitud.


````python
print("request body:", r.request.body)
````

Podemos imprimir el código de estado.


````python
print(r.status_code)
````

Podemos ver la respuesta como texto:


````python
print(r.text)
````

Podemos ver el <code>'Content-Type'</code>.


````python
r.headers['Content-Type']
````

Como el contenido <code>'Content-Type'</code> está en el formato <code>JSON</code> podemos usar el método <code>json()</code>, devuelve un <code>dict</code> de Python:


````python
r.json()
````

La clave <code>args</code> tiene el nombre y los valores:


````python
r.json()['args']
````

## Solicitudes Post


Al igual que una solicitud <code>GET</code>, un <code>POST</code> se usa para enviar datos a un servidor, pero la solicitud <code>POST</code> envía los datos en un cuerpo de solicitud. Para enviar la Solicitud Post en Python, en la <code>URL</code> cambiamos la ruta a <code>POST</code>:


````python
url_post='http://httpbin.org/post'
````

Este endpoint esperará datos como un archivo o como un formulario. Un formulario es una forma conveniente de configurar una solicitud HTTP para enviar datos a un servidor.


Para hacer una solicitud <code>POST</code> usamos la función <code>post()</code>, la variable <code>payload</code> se pasa al parámetro <code> data </code>:


````python
r_post=requests.post(url_post,data=payload)
````

Comparando la URL del objeto de respuesta de la solicitud <code>GET</code> y <code>POST</code> vemos que la solicitud <code>POST</code> no tiene pares de nombre o valor.


````python
#print("POST request URL:",response.url )
#print("GET request URL:",r.url)
print("POST request URL:", r_post.url)  # Usa r_post en lugar de response
````

Podemos comparar el cuerpo de la solicitud <code>POST</code> y <code>GET</code>, vemos que solo la solicitud <code>POST</code> tiene un cuerpo:


````python
print("POST request body:",r_post.request.body)
print("GET request body:",r.request.body)
````

Podemos ver el formulario también:


````python
r_post.json()['form']
````

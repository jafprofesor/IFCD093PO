## ¿Qué es Numpy?


NumPy es una biblioteca de Python utilizada para trabajar con matrices, álgebra lineal, transformada de Fourier y matrices. NumPy significa Numerical Python y es un proyecto de código abierto. El objeto array en NumPy se llama **ndarray**, proporciona muchas funciones de soporte que hacen que trabajar con ndarray sea muy fácil.

Los arrays se utilizan con mucha frecuencia en la ciencia de datos, donde la velocidad y los recursos son muy importantes.

NumPy generalmente se importa bajo el alias np.

Por lo general, tiene un tamaño fijo y cada elemento es del mismo tipo. Podemos convertir una lista en un array de numpy importando primero `numpy`:


````python
# importar la biblioteca numpy

import numpy as np 
````

Luego convertimos la lista de la siguiente manera:


````python
# Crear un arreglo numpy unidimensional

a = np.array([0, 1, 2, 3, 4])
a
````

### Slicing (Segmentación)


Al igual que las listas, podemos segmentar el array de numpy. Slicing en python significa tomar los elementos de un índice dado a otro índice dado.

Pasamos el slice así: [inicio:fin]. El elemento en el índice final no se incluye en la salida.

Podemos seleccionar los elementos del 1 al 3 y asignarlo a un nuevo array de numpy <code>d</code> de la siguiente manera:


````python
# Dividiendo el arreglo numpy

d = c[1:4]
d
````

Podemos asignar los índices correspondientes a nuevos valores de la siguiente manera:


````python
#  Asignar el cuarto elemento y el quinto elemento a 300 y 400

c[3:5] = 300, 400
c
````

También podemos definir los pasos en el slicing, así: [inicio:fin:paso].


````python
arr = np.array([1, 2, 3, 4, 5, 6, 7])

print(arr[1:5:2])
````

Si no pasamos el inicio se considera 0


````python
print(arr[:4])
````

Si no pasamos el final se considera hasta la longitud del array.


````python
print(arr[4:])
````

Si no pasamos el paso se considera 1


````python
print(arr[1:5:])
````

### Suma de Arrays



Considera el array de numpy <code>u</code>:


````python
u = np.array([1, 0])
u
````

Considera el array de numpy <code>v</code>:


````python
v = np.array([0, 1])
v
````

Podemos sumar los dos arrays y asignarlo a z:


````python
# Suma de Arrays de Numpy

z = np.add(u, v)
z
````

La operación es equivalente a la suma de vectores:


````python
# Funciones para grafica


import time 
import sys
import numpy as np 

import matplotlib.pyplot as plt


def Plotvec1(u, z, v):
    
    ax = plt.axes() # para generar los ejes de la ventana completa
    ax.arrow(0, 0, *u, head_width=0.05, color='r', head_length=0.1)# Agrega una flecha a los ejes U con ancho de cabeza de flecha 0.05, color rojo y longitud de cabeza de flecha 0.1
    plt.text(*(u + 0.1), 'u')#Agrega el texto u a los ejes 
    
    ax.arrow(0, 0, *v, head_width=0.05, color='b', head_length=0.1)# Agrega una flecha a los ejes v con ancho de cabeza de flecha 0.05, color azul y longitud de cabeza de flecha 0.1
    plt.text(*(v + 0.1), 'v') # Agrega el texto v a los ejes 
    
    ax.arrow(0, 0, *z, head_width=0.05, head_length=0.1)
    plt.text(*(z + 0.1), 'z') # Agrega el texto z a los ejes 
    plt.ylim(-2, 2) # establece el límite inferior y superior del eje y a -2 y 2 respectivamente
    plt.xlim(-2, 2) # establece el límite izquierdo y derecho del eje x a -2 y 2 respectivamente

  
````

````python
# Graficar arreglos numpy

Plotvec1(u, z, v)
````

## Linspace


Una función útil para trazar funciones matemáticas es <code>linspace</code>. Linspace devuelve números espaciados uniformemente sobre un intervalo específico. 

 **numpy.linspace(start, stop, num = int value)**
 
start  :  start of interval range

stop   :  end of interval range

num    :  Number of samples to generate.


````python
#  Crea un arreglo numpy dentro de [-2, 2] con 5 elementos

np.linspace(-2, 2, num=5)
````

Si cambiamos el parámetro <code>num</code> a 9, obtenemos 9 números espaciados uniformemente sobre el intervalo de -2 a 2:


````python
#  Crea un arreglo numpy dentro de [-2, 2] con 9 elementos

np.linspace(-2, 2, num=9)
````

Podemos usar la función <code>linspace</code> para generar 100 muestras espaciadas uniformemente del intervalo 0 a 2π:


````python
#  Crea un arreglo numpy dentro de [0, 2π] con 100 elementos 

x = np.linspace(0, 2*np.pi, num=100)
````

Podemos aplicar la función seno a cada elemento en el array <code>x</code> y asignarlo al array <code>y</code>:


````python
#  Calcula el seno de la lista x

y = np.sin(x)
````

````python
#  Grafica del resultado

plt.plot(x, y)
````

### Ejercicio 1: Pandas: DataFrame y Series 

**Pandas** es una biblioteca popular para análisis de datos construida sobre el lenguaje de programación Python. Pandas generalmente proporciona dos estructuras de datos para manipular datos, son: 
 
* DataFrame
* Series

Un **DataFrame** es una estructura de datos bidimensional, es decir, los datos se alinean en forma tabular en filas y columnas.

* Un DataFrame de Pandas se creará cargando los conjuntos de datos desde un almacenamiento existente. 
* El almacenamiento puede ser una base de datos SQL, archivo CSV, archivo Excel, etc. 
* También se puede crear desde listas, diccionarios y desde una lista de diccionarios.

**Series** representa una matriz unidimensional de datos indexados.
Tiene dos componentes principales :
1. Una matriz de datos reales.
2. Una matriz asociada de índices o etiquetas de datos.

El índice se usa para acceder a valores de datos individuales. También puedes obtener una columna de un dataframe como una **Series**. Puedes pensar en una serie de Pandas como un dataframe 1-D.

**En este laboratorio de práctica, aprenderemos cómo crear un DataFrame a partir de un diccionario.**


Consideremos un diccionario 'x' con claves y valores como se muestra a continuación.

Luego creamos un dataframe desde el diccionario usando la función pd.DataFrame(dict)


````python
# Definir un diccionario 'x'

x = {'Name': ['Rose','John', 'Jane', 'Mary'], 'ID': [1, 2, 3, 4], 'Department': ['Architect Group', 'Software Group', 'Design Team', 'Infrastructure'], 
      'Salary':[100000, 80000, 50000, 60000]}

# Convertir el diccionario a un DataFrame
df = pd.DataFrame(x)

# Mostrar el DataFrame resultante
df
````

Podemos ver la correspondencia directa entre la tabla. Las claves corresponden a las etiquetas de columna y los valores o listas corresponden a las filas.


#### Selección de Columna:

Para seleccionar una columna en un DataFrame de Pandas, podemos acceder a las columnas llamándolas por su nombre de columna. 

Recuperemos los datos presentes en la columna <code>ID</code>.


````python
#  Recuperar la columna "ID" y asignarla a una variable x
x = df[['ID']]
x
````

Usemos la función <code>type()</code> y verifiquemos el tipo de la variable.


````python
#  comprobar el tipo de x
type(x)
````

La salida nos muestra que el tipo de la variable es un objeto DataFrame.


#### Acceso a múltiples columnas

Recuperemos los datos para las columnas <code>Department</code>, <code>Salary</code> e <code>ID</code>


````python
# Recuperar las columnas Department, Salary y ID y asignarlas a una variable z

z = df[['Department','Salary','ID']]
z
````

### Ejercicio 2: Funciones <code>loc()</code> e <code>iloc()</code>

<code>loc()</code> es un método de selección de datos basado en etiquetas, lo que significa que tenemos que pasar el nombre de la fila o columna que queremos seleccionar. Este método incluye el último elemento del rango pasado en él.

Sintaxis simple para tu entendimiento: 

 - loc[etiqueta_fila, etiqueta_columna]

<code>iloc()</code> es un método de selección basado en índices, lo que significa que tenemos que pasar un índice entero en el método para seleccionar una fila/columna específica. Este método no incluye el último elemento del rango pasado en él.

Sintaxis simple para tu entendimiento: 
   
 - iloc[índice_fila, índice_columna]

<h4 id="data">Veamos algunos ejemplos sobre lo mismo.</h4>


````python
#  Acceder al valor en la primera fila y la primera columna

df.iloc[0, 0]
````

````python
#  Acceder al valor en la primera fila y la tercera columna

df.iloc[0,2]
````

````python
#  Acceder a la columna usando el nombre

df.loc[0, 'Salary']
````

Creemos un nuevo dataframe llamado 'df2' y asignémosle 'df'. Ahora, establezcamos la columna "Name" como una columna de índice usando el método set_index().


````python
df2=df
df2=df2.set_index("Name")
````

````python
#  Mostrar las primeras 5 filas del nuevo DataFrame
df2.head()
````

````python
#  Ahora, accedamos a la columna usando el nombre
df2.loc['Jane', 'Salary']
````

### Ejercicio 3: Slicing (Rebanado)

Slicing usa el operador [] para seleccionar un conjunto de filas y/o columnas de un DataFrame.

Para rebanar un conjunto de filas, usa esta sintaxis: data[inicio:parada], 

aquí el inicio representa el índice desde donde considerar, y la parada representa el índice un paso MÁS ALLÁ de la fila que quieres seleccionar. Puedes realizar slicing usando tanto el índice como el nombre de la columna.

> NOTA: Al hacer slicing en pandas, el límite de inicio se incluye en la salida.

Así que si quieres seleccionar las filas 0, 1 y 2, tu código se vería así: df.iloc[0:3].

Significa que le estás diciendo a Python que comience en el índice 0 y seleccione las filas 0, 1, 2 hasta pero no incluyendo 3.

> NOTA: Las etiquetas deben encontrarse en el DataFrame o obtendrás un KeyError.

Indexar por etiquetas (es decir, usando <code>loc()</code>) difiere de indexar por enteros (es decir, usando <code>iloc()</code>). Con <code>loc()</code>, tanto el límite de inicio como el límite de parada son inclusivos. Al usar <code>loc()</code>, se pueden usar enteros, pero los enteros se refieren a la etiqueta del índice y no a la posición. 

Por ejemplo, usar <code>loc()</code> y seleccionar 1:4 obtendrá un resultado diferente que usar <code>iloc()</code> para seleccionar filas 1:4.

<h4 id="data">También podemos seleccionar un valor de datos específico usando una ubicación de fila y columna dentro del DataFrame y la indexación iloc.


````python
#  hagamos el slicing usando el antiguo dataframe df

df.iloc[0:2, 0:3]
````

````python
# Haremos el slicing usando la función loc() en el antiguo dataframe df donde la columna de índice tiene etiquetas como 0,1,2
df.loc[0:2,'ID':'Department']
````

````python
# Haremos el slicing usando la función loc() en el nuevo dataframe df2 donde la columna de índice es Name con etiquetas: Rose, John y Jane
df2.loc['Rose':'Jane', 'ID':'Department']
````

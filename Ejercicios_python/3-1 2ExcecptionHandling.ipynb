## ¿Qué es una Excepción?


En esta sección aprenderás qué es una excepción y verás ejemplos de ellas.


### Definición


Una excepción es un error que ocurre durante la ejecución del código. Este error causa que el código lance una excepción y, si no está preparado para manejarla, detendrá la ejecución del código.


### Ejemplos


Ejecuta cada fragmento de código y observa la excepción lanzada


````python
1/0
````

<code>ZeroDivisionError</code> ocurre cuando intentas dividir por cero.


````python
y = a + 5
````

<code>NameError</code> -- en este caso, significa que intentaste usar la variable a cuando no estaba definida.


````python
a = [1, 2, 3]
a[10]
````

<code>IndexError</code> -- en este caso, ocurrió porque intentaste acceder a datos de una lista usando un índice que no existe para esta lista.


Hay muchas más excepciones integradas en Python, aquí hay una lista de ellas https://docs.python.org/3/library/exceptions.html


## Manejo de Excepciones


En esta sección aprenderás cómo manejar excepciones. Entenderás cómo hacer que tu programa realice tareas especificadas en lugar de detener la ejecución del código cuando se encuentra una excepción.


### Try Except


Un <code>try except</code> te permitirá ejecutar código que podría lanzar una excepción y, en el caso de cualquier excepción o una específica, podemos manejar o capturar la excepción y ejecutar código específico. Esto nos permitirá continuar la ejecución de nuestro programa incluso si hay una excepción.

Python intenta ejecutar el código en el bloque <code>try</code>. En este caso, si hay alguna excepción lanzada por el código en el bloque <code>try</code>, será capturada y se ejecutará el bloque de código en el bloque <code>except</code>. Después de eso, el código que viene <em>después</em> del try except será ejecutado.


````python
# potencial código antes del try catch

try:
    # código a intentar ejecutar
except:
    # código a ejecutar si hay una excepción
    
# código que se ejecutará incluso si hay una excepción
````

### Ejemplo de Try Except


En este ejemplo estamos tratando de dividir un número dado por el usuario, guardar el resultado en la variable <code>a</code>, y luego nos gustaría imprimir el resultado de la operación. Al tomar la entrada del usuario y dividir un número por ella, hay un par de excepciones que pueden lanzarse. Por ejemplo, si dividimos por cero. Intenta ejecutar el siguiente bloque de código con <code>b</code> como un número. Solo se lanzará una excepción si <code>b</code> es cero.


````python
a = 1

try:
    b = int(input(" Por favor ingrese un número para dividir a"))
    a = a/b
    print("Éxito a=",a)
except:
    print("Hubo un error")
        
````

### Try Except Específico


Un <code>try except</code> específico te permite capturar ciertas excepciones y también ejecutar cierto código dependiendo de la excepción. Esto es útil si no quieres lidiar con algunas excepciones y la ejecución debería detenerse. También puede ayudarte a encontrar errores en tu código de los que podrías no ser consciente. Además, puede ayudarte a diferenciar las respuestas a diferentes excepciones. En este caso, el código después del try except podría no ejecutarse dependiendo del error.


<b>No ejecutar, solo para ilustrar:</b>


````python
# potencial código antes del try catch

try:
    # código a intentar ejecutar
except (ZeroDivisionError, NameError):
    # código a ejecutar si hay una excepción de los tipos dados
    
# código que se ejecutará si no hay una excepción o si es una que estamos manejando
````

````python
# potencial código antes del try catch

try:
    # código a intentar ejecutar
except ZeroDivisionError:
    # código a ejecutar si hay una ZeroDivisionError
except NameError:
    # código a ejecutar si hay una NameError
    
# código que se ejecutará si no hay una excepción o si es una que estamos manejando
````

También puedes tener un <code>except</code> vacío al final para capturar una excepción inesperada:


<b>No ejecutar, solo para ilustrar:</b>


````python
# potencial código antes del try catch

try:
    # código a intentar ejecutar
except ZeroDivisionError:
    # código a ejecutar si hay una ZeroDivisionError
except NameError:
    # código a ejecutar si hay una NameError
except:
    # código a ejecutar si hay cualquier otra excepción
    
# código que se ejecutará si no hay una excepción o si es una que estamos manejando
````

### Ejemplo de Try Except Específico


Este es el mismo ejemplo de arriba, pero ahora agregaremos mensajes diferenciados dependiendo de la excepción, informando al usuario qué está mal con la entrada.


````python
a = 1

try:
    b = int(input("Por favor, introduzca un número para dividir por a"))
    a = a/b
    print("Exito a=",a)
except ZeroDivisionError:
    print("El número que has proporcionado no es correcto por que es un 0")
except ValueError:
    print("No has escrito ningún número")
except:
    print("Algo ha salido mal")
        
````

### Try Except Else y Finally


<code>else</code> permite verificar si no hubo excepción al ejecutar el bloque try. Esto es útil cuando queremos ejecutar algo solo si no hubo errores.


<b>no ejecutar, solo para ilustrar</b>


````python
# potencial código antes del try catch

try:
    # código a intentar ejecutar
except ZeroDivisionError:
    # código a ejecutar si hay una ZeroDivisionError
except NameError:
    # código a ejecutar si hay una NameError
except:
    # código a ejecutar si hay cualquier otra excepción
else:
    # código a ejecutar si no hay una excepción
    
# código que se ejecutará si no hay una excepción o si es una que estamos manejando
````

<code>finally</code> nos permite ejecutar siempre algo, ya sea que haya una excepción o no. Esto se usa generalmente para indicar el final del try except.


````python
# potencial código antes del try catch

try:
    # código a intentar ejecutar
except ZeroDivisionError:
    # código a ejecutar si hay una ZeroDivisionError
except NameError:
    # código a ejecutar si hay una NameError
except:
    # código a ejecutar si hay cualquier otra excepción
else:
    # código a ejecutar si no hay una excepción
finally:
    # código a ejecutar al final del try except sin importar qué
    
# código que se ejecutará si no hay una excepción o si es una que estamos manejando
````

### Ejemplo de Try Except Else y Finally


Puede que hayas notado que incluso si hay un error, el valor de <code>a</code> siempre se imprime. Usemos el <code>else</code> e imprimamos el valor de <code>a</code> solo si no hay error.


````python
a = 1

try:
    b = int(input(" Por favor ingrese un número para dividir a"))
    a = a/b
except ZeroDivisionError:
    print("El número que has proporcionado no es correcto por que es un 0")
except ValueError:
    print("No has escrito ningún número")
except:
    print("Algo ha salido mal")
else:
    print("Éxito a=",a)
````

Ahora dejemos saber al usuario que hemos terminado de procesar su respuesta. Usando el <code>finally</code>, agreguemos un print.


````python
a = 1

try:
    b = int(input(" Por favor ingrese un número para dividir a"))
    a = a/b
except ZeroDivisionError:
    print("El número que has proporcionado no es correcto por que es un 0")
except ValueError:
    print("No has escrito ningún número")
except:
    print("Algo ha salido mal")
else:
    print("Éxito a=",a)
finally:
    print("Proceso completado")
````

## Ejercicio 1: Manejando ZeroDivisionError


Imagina que tienes dos números y quieres determinar qué sucede cuando divides un número por el otro. Para hacer esto, necesitas crear una función de Python llamada `safe_divide`. Le das a esta función dos números, un `'numerador'` y un `'denominador'`. El 'numerador' es el número que quieres dividir, y el `'denominador'` es el número por el cual quieres dividir. Usa el método de entrada del usuario de Python para tomar los valores.

La función debería ser capaz de hacer la división por ti y darte el resultado. Pero aquí está el detalle: si intentas dividir por cero (lo cual no está permitido en matemáticas), la función debe ser lo suficientemente inteligente como para detectar eso y decirte que no es posible dividir por cero. En lugar de mostrar un error, debería devolver None, lo que significa 'nada' o 'sin valor', e imprimir `"Error: Cannot divide by Zero.`


````python
# Escribe tu código aquí
def safe_divide(numerator,denominator):
    try:
        result = numerator / denominator
        return result
    except ZeroDivisionError:
        print("Error: no se puede dividir por cero.")
        return None
# Test case
numerator=int(input("Ingrese el valor del numerador:-"))
denominator=int(input("Ingrese el valor del denominador:-"))
print(safe_divide(numerator,denominator))
````

## Ejercicio 3: Manejando Excepciones Genéricas


Imagina que tienes un número y quieres realizar una tarea matemática compleja. El cálculo requiere dividir el valor del argumento de entrada `"num"` por la diferencia entre `"num"` y 5, y el resultado tiene que ser almacenado en una variable llamada `"result"`.

Tienes que definir una función para que pueda realizar esa tarea matemática compleja. La función debe manejar cualquier error potencial que ocurra durante el cálculo. Para hacer esto, puedes usar un bloque try-except. Si surge alguna excepción durante el cálculo, debe capturar el error usando la clase de excepción genérica `"Exception" as "e"`. Cuando ocurre una excepción, la función debe mostrar `"An error occurred during calculation.`


````python
#Type your code here
def complex_calculation(num):
    try:
        result = num / (num - 5)
        print (f" Resultado: {result}")
    except Exception as e:
        print("Ocurrió un error durante el cálculo.")
# Test case
user_input = float(input("Ingrese un número: "))
complex_calculation(user_input)
````

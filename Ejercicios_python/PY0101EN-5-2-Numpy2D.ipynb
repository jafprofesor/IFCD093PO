## Crear un Array de Numpy 2D


````python
# Importar las bibliotecas

import numpy as np
````

Considera la lista <code>a</code>, que contiene tres listas anidadas **cada una de igual tamaño**. 


````python
# Crear una lista 2D

a = [[11, 12, 13], [21, 22, 23], [31, 32, 33]]
a
````

Podemos convertir la lista a un Array de Numpy de la siguiente manera:


````python
# Convertir lista a arreglo Numpy
# Cada elemento es del mismo tipo

A = np.array(a)
A
````

Podemos usar el atributo <code>ndim</code> para obtener el número de ejes o dimensiones, referido como el rango. 


````python
#  Mostrar las dimensiones del arreglo numpy

A.ndim
````

El atributo <code>shape</code> devuelve una tupla correspondiente al tamaño o número de cada dimensión.


````python
#  Mostrar la forma del arreglo numpy

A.shape
````

El número total de elementos en el array está dado por el atributo <code>size</code>.


````python
#  Mostrar el tamaño del arreglo numpy

A.size
````

## Operaciones Básicas


También podemos sumar arrays. El proceso es idéntico a la suma de matrices. La suma de matrices de <code>X</code> e <code>Y</code> se muestra en la siguiente figura:


<img src="https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/IBMDeveloperSkillsNetwork-PY0101EN-SkillsNetwork/labs/Module%205/images/NumTwoAdd.png" width="500">


El array de numpy está dado por <code>X</code> e <code>Y</code>


````python
#  Crear un arreglo numpy X

X = np.array([[1, 0], [0, 1]]) 
X
````

````python
#  Crear un arreglo numpy Y

Y = np.array([[2, 1], [1, 2]]) 
Y
````

 Podemos sumar los arrays de numpy de la siguiente manera.


````python
#  Sumar X y Y

Z = X + Y
Z
````

Multiplicar un array de numpy por un escalar es idéntico a multiplicar una matriz por un escalar. Si multiplicamos la matriz <code>Y</code> por el escalar 2, simplemente multiplicamos cada elemento en la matriz por 2, como se muestra en la figura.


<img src="https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/IBMDeveloperSkillsNetwork-PY0101EN-SkillsNetwork/labs/Module%205/images/NumTwoDb.png" width="500">


Podemos realizar la misma operación en numpy de la siguiente manera 


````python
#  Crear un arreglo numpy Y

Y = np.array([[2, 1], [1, 2]]) 
Y
````

````python
#  Multiplicar Y por 2

Z = 2 * Y
Z
````

La multiplicación de dos arrays corresponde a un producto elemento a elemento o <em>producto de Hadamard</em>. Considera la matriz <code>X</code> e <code>Y</code>. El producto de Hadamard corresponde a multiplicar cada uno de los elementos en la misma posición, es decir, multiplicar elementos contenidos en las mismas cajas de color juntos. El resultado es una nueva matriz que es del mismo tamaño que la matriz <code>Y</code> o <code>X</code>, como se muestra en la siguiente figura.


<img src="https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/IBMDeveloperSkillsNetwork-PY0101EN-SkillsNetwork/labs/Module%205/images/NumTwoMul.png" width="500">


Podemos realizar el producto elemento a elemento del array <code>X</code> e <code>Y</code> de la siguiente manera:


````python
#  Crear un arreglo numpy Y

Y = np.array([[2, 1], [1, 2]]) 
Y
````

````python
#  Crear un arreglo numpy X

X = np.array([[1, 0], [0, 1]]) 
X
````

````python
#  Multiplicar X por Y

Z = X * Y
Z
````

También podemos realizar la multiplicación de matrices con los arrays de numpy <code>A</code> y <code>B</code> de la siguiente manera:


Primero, definimos la matriz <code>A</code> y <code>B</code>:


````python
#  Crear una matriz A

A = np.array([[0, 1, 1], [1, 0, 1]])
A
````

````python
#  Crear una matriz B

B = np.array([[1, 1], [1, 1], [-1, 1]])
B
````

Usamos la función <code>dot</code> de numpy para multiplicar los arrays juntos.


````python
#  Calcular el producto punto

Z = np.dot(A,B)
Z
````

````python
#  Calcular el seno de Z

np.sin(Z)
````

Usamos el atributo <code>T</code> de numpy para calcular la matriz transpuesta


````python
#  Crear una matriz C

C = np.array([[1,1],[2,2],[3,3]])
C
````

````python
#  Obtener la transpuesta de C

C.T
````

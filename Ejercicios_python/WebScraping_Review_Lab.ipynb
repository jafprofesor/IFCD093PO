<h2 id="BSO">Objetos Beautiful Soup</h2>


Beautiful Soup es una biblioteca de Python para extraer datos de archivos HTML y XML, nos centraremos en archivos HTML. Esto se logra representando el HTML como un conjunto de objetos con métodos utilizados para analizar el HTML. Podemos navegar por el HTML como un árbol y/o filtrar lo que estamos buscando.

Considera el siguiente HTML:


````python
%%html
<!DOCTYPE html>
<html>
<head>
<title>Page Title</title>
</head>
<body>
<h3><b id='boldest'>Lebron James</b></h3>
<p> Salary: $ 92,000,000 </p>
<h3> Stephen Curry</h3>
<p> Salary: $85,000, 000 </p>
<h3> Kevin Durant </h3>
<p> Salary: $73,200, 000</p>
</body>
</html>
````

Podemos almacenarlo como una cadena en la variable HTML:


````python
html="<!DOCTYPE html><html><head><title>Page Title</title></head><body><h3><b id='boldest'>Lebron James</b></h3><p> Salary: $ 92,000,000 </p><h3> Stephen Curry</h3><p> Salary: $85,000, 000 </p><h3> Kevin Durant </h3><p> Salary: $73,200, 000</p></body></html>"
````

Para analizar un documento, pásalo al constructor <code>BeautifulSoup</code>, el objeto <code>BeautifulSoup</code>, que representa el documento como una estructura de datos anidada:


````python
soup = BeautifulSoup(html, "html.parser")
````

Primero, el documento se convierte a Unicode, (similar a ASCII), y las entidades HTML se convierten a caracteres Unicode. Beautiful Soup transforma un documento HTML complejo en un árbol complejo de objetos de Python. El objeto <code>BeautifulSoup</code> puede crear otros tipos de objetos. En este laboratorio, cubriremos objetos <code>BeautifulSoup</code> y <code>Tag</code> que para los propósitos de este laboratorio son idénticos, y objetos <code>NavigableString</code>.


Podemos usar el método <code>prettify()</code> para mostrar el HTML en la estructura anidada:


````python
print(soup.prettify())
````

## Hacer scraping de datos de tablas HTML


````python
import pandas as pd
````

````python
# La siguiente URL contiene tablas HTML con datos sobre la población mundial.
url = "https://en.wikipedia.org/wiki/World_population"
````

Antes de proceder a hacer scraping de un sitio web, necesitas examinar los contenidos y la forma en que los datos están organizados en el sitio web. Abre la url anterior en tu navegador y verifica las tablas en la página web.


````python
#  obtener el contenido de la página web en formato de texto y almacenarlo en una variable llamada data
headers = {
    "User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64) "
                  "AppleWebKit/537.36 (KHTML, like Gecko) "
                  "Chrome/91.0.4472.124 Safari/537.36"
}

data  = requests.get(url, headers=headers)
````

````python
soup = BeautifulSoup(data.text,"html.parser")
````

````python
# encontrar todas las tablas html en la página web
tables = soup.find_all('table') # en html, una tabla está representada por la etiqueta <table>
````

````python
# podemos ver cuántas tablas se encontraron verificando la longitud de la lista tables
len(tables)
````

Supongamos que estamos buscando la tabla `10 most densly populated countries` (10 países más densamente poblados), podemos buscar a través de la lista de tablas y encontrar la correcta que estamos buscando basándonos en los datos en cada tabla o podemos buscar por el nombre de la tabla si está en la tabla, pero esta opción podría no funcionar siempre.


````python
for index,table in enumerate(tables):
    if ("10 most densely populated countries" in str(table)):
        table_index = index
print(table_index)
````

Mira si puedes localizar el nombre de la tabla, `10 most densly populated countries`, a continuación.


````python
print(tables[table_index].prettify())
````

````python
population_data = pd.DataFrame(columns=["Rank", "Country", "Population", "Area", "Density"])

for row in tables[table_index].tbody.find_all("tr"):
    col = row.find_all("td")
    if col:
        rank = col[0].text.strip()
        country = col[1].text.strip()
        population = col[2].text.strip()
        area = col[3].text.strip()
        density = col[4].text.strip()

        #  Crear un DataFrame temporal para la nueva fila
        new_row = pd.DataFrame([{"Rank": rank, "Country": country, "Population": population, "Area": area, "Density": density}])

        # Usar concat 
        population_data = pd.concat([population_data, new_row], ignore_index=True)

population_data
````

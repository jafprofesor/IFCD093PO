### Escribiendo comentarios en Python


<p>
    Además de escribir código, ten en cuenta que siempre es una buena idea agregar comentarios a tu código. Ayudará a otros a entender lo que intentabas lograr (la razón por la que escribiste un fragmento de código dado). Esto no solo ayuda a <strong>otras personas</strong> a entender tu código, sino que también puede servir como recordatorio <strong>para ti</strong> cuando vuelvas a él semanas o meses después.</p>

<p>
    Para escribir comentarios en Python, usa el símbolo de número <code>#</code> antes de escribir tu comentario. Cuando ejecutes tu código, Python ignorará todo lo que esté después del <code>#</code> en una línea dada.
</p>


````python
# Prueba tu primera salida en Python

print('¡Hola, Python!') # Esta línea imprime una cadena
# print("Hola, Python!") 
````

<p>
    Después de ejecutar la celda anterior, deberías notar que <code>This line prints a string</code> no apareció en la salida, porque era un comentario (y por lo tanto ignorado por Python).
</p>
<p>
    ¡La segunda línea tampoco se ejecutó porque <code>print('¡Hola, Python!')</code> estaba precedida por el signo de número (<code>#</code>) también! Dado que este no es un comentario explicativo del programador, sino una línea de código real, podríamos decir que el programador <em>comentó</em> esa segunda línea de código.
</p>


<h3 id="errors">Errores en Python</h3>


<p>Todos cometemos errores. Para muchos tipos de errores, Python te dirá que has cometido un error dándote un mensaje de error. Es importante leer los mensajes de error cuidadosamente para entender realmente dónde cometiste un error y cómo puedes corregirlo.</p>
<p>Por ejemplo, si escribes <code>print</code> como <code>frint</code>, Python mostrará un mensaje de error. Inténtalo:</p>


````python
# Imprimir un error intencionado

frint("Hello, Python!")
````

<p>El mensaje de error te dice: 
<ol>
    <li>dónde ocurrió el error (más útil en celdas de cuaderno o scripts grandes), y</li> 
    <li>qué tipo de error fue (NameError)</li> 
</ol>
<p>Aquí, Python intentó ejecutar la función <code>frint</code>, pero no pudo determinar qué es <code>frint</code> ya que no es una función integrada y tampoco ha sido definida previamente por nosotros.</p>


<p>
    Notarás que si cometemos un tipo diferente de error, al olvidar cerrar la cadena, obtendremos un error diferente (es decir, un <code>SyntaxError</code>). Inténtalo a continuación:
</p>


````python
# Prueba con un error de sintaxis

print("Hello, Python!)
````

### ¿Sabe Python de tu error antes de ejecutar tu código?


Python es lo que se llama un <em>lenguaje interpretado</em>. Los lenguajes compilados examinan todo su programa en tiempo de compilación y pueden advertirle sobre toda una clase de errores antes de la ejecución. En contraste, Python interpreta su script línea por línea a medida que lo ejecuta. Python dejará de ejecutar todo el programa cuando encuentre un error (a menos que el error sea esperado y manejado por el programador, un tema más avanzado que cubriremos más adelante en este curso).


Intenta ejecutar el código en la celda de abajo y mira qué sucede:


````python
# Imprimir cadena y error para ver el orden de ejecución

print("Esto se imprimirá")
frint("Esto causará un error")
print("Esto NO se imprimirá")
````

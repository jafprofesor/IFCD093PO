## Escribiendo Archivos


 Podemos abrir un objeto de archivo usando el método <code>write()</code> para guardar el archivo de texto en una lista. Para escribir en un archivo, el argumento de modo debe establecerse en **w**. Escribamos un archivo **Example2.txt** con la línea: **“This is line A”**


````python
# Escribir línea en archivo
exmp2 = '/Example2.txt'
with open(exmp2, 'w') as writefile:
    writefile.write("Esta es la línea A")
````

 Podemos leer el archivo para ver si funcionó:


````python
# Leer archivo escrito

with open(exmp2, 'r') as testwritefile:
    print(testwritefile.read())
````

Podemos escribir múltiples líneas:


````python
# Escribir líneas en archivo

with open(exmp2, 'w') as writefile:
    writefile.write("Esta es la línea A\n")
    writefile.write("Esta es la línea B\n")
````

El método <code>.write()</code> funciona de manera similar al método <code>.readline()</code>, excepto que en lugar de leer una nueva línea, escribe una nueva línea. El proceso se ilustra en la figura. La codificación de colores diferente de la cuadrícula representa una nueva línea agregada al archivo después de cada llamada al método.

````python
# Comprobar si se escribió en el archivo

with open(exmp2, 'r') as testwritefile:
    print(testwritefile.read())
````

 Escribimos una lista en un archivo **.txt** de la siguiente manera:


````python
# Ejemplo de lista de texto

Lines = ["Esta es la línea A\n", "Esta es la línea B\n", "Esta es la línea C\n"]
Lines
````

````python
# Escribir las cadenas en la lista en un archivo de texto

with open('/Example2.txt', 'w') as writefile:
    for line in Lines:
        print(line)
        writefile.write(line)
````

 Podemos verificar que el archivo esté escrito leyéndolo e imprimiendo los valores:  


````python
# Verificar si la escritura en el archivo se ejecutó correctamente

with open('/Example2.txt', 'r') as testwritefile:
    print(testwritefile.read())
````

Sin embargo, ten en cuenta que establecer el modo en __w__ sobrescribe todos los datos existentes en el archivo.


````python
with open('/Example2.txt', 'w') as writefile:
    writefile.write("Overwrite\n")
with open('/Example2.txt', 'r') as testwritefile:
    print(testwritefile.read())
````

## Anexando Archivos


 Podemos escribir en archivos sin perder ninguno de los datos existentes de la siguiente manera estableciendo el argumento de modo en append (anexar): **a**. puedes anexar una nueva línea de la siguiente manera:


````python
# Escribir una nueva línea en el archivo de texto

with open('/Example2.txt', 'a') as testwritefile:
    testwritefile.write("Esta es la línea C\n")
    testwritefile.write("Esta es la línea D\n")
    testwritefile.write("Esta es la línea E\n")
````

 Puedes verificar que el archivo ha cambiado ejecutando la siguiente celda:


````python
# Verificar si la nueva línea está en el archivo de texto

with open('/Example2.txt', 'r') as testwritefile:
    print(testwritefile.read())
````

## Modos adicionales


Es bastante ineficiente abrir el archivo en **a** o **w** y luego volver a abrirlo en **r** para leer líneas. Afortunadamente, podemos acceder al archivo en los siguientes modos:
- **r+** : Lectura y escritura. No se puede truncar el archivo.
- **w+** : Escritura y lectura. Trunca el archivo.
- **a+** : Anexar y Leer. Crea un nuevo archivo, si no existe ninguno.
No tienes que detenerte en los detalles de cada modo para este laboratorio. 


Probemos el modo __a+__:


````python
with open('/Example2.txt', 'a+') as testwritefile:
    testwritefile.write("Esta es la línea E\n")
    print(testwritefile.read())
````

No hubo errores, pero <code>read()</code> tampoco generó nada. Esto se debe a nuestra ubicación en el archivo.


La mayoría de los métodos de archivo que hemos visto funcionan en una ubicación determinada en el archivo. <code>.write() </code> escribe en una ubicación determinada en el archivo. <code>.read()</code> lee en una ubicación determinada en el archivo y así sucesivamente. Puedes pensar en esto como mover tu puntero en el bloc de notas para realizar cambios en una ubicación específica.


Abrir el archivo en **w** es similar a abrir el archivo .txt, mover el cursor al principio del archivo de texto, escribir nuevo texto y eliminar todo lo que sigue.
Mientras que abrir el archivo en **a** es similar a abrir el archivo .txt, mover el cursor al final y luego agregar los nuevos fragmentos de texto. <br>
A menudo es muy útil saber dónde está el 'cursor' en un archivo y poder controlarlo. Los siguientes métodos nos permiten hacer precisamente esto -
- <code>.tell()</code> - devuelve la posición actual en bytes
- <code>.seek(offset,from)</code> - cambia la posición por 'offset' bytes con respecto a 'from'. From puede tomar el valor de 0, 1, 2 correspondiendo al principio, relativo a la posición actual y al final


Ahora revisemos **a+**


````python
with open('/Example2.txt', 'a+') as testwritefile:
    print("Lugar inicial: {}".format(testwritefile.tell()))
    
    data = testwritefile.read()
    if (not data):  # cadenas vacías devuelven falso en python
            print('No se leyó nada') 
    else: 
            print(testwritefile.read())
            
    testwritefile.seek(0,0) #  mover 0 bytes desde el principio.
    
    print("\nNueva ubicación : {}".format(testwritefile.tell()))
    data = testwritefile.read()
    if (not data): 
            print('No se leyó nada') 
    else: 
            print(data)
    
    print("Ubicación después de la lectura: {}".format(testwritefile.tell()) )
````

Finalmente, una nota sobre la diferencia entre **w+** y **r+**. Ambos modos permiten el acceso a métodos de lectura y escritura, sin embargo, abrir un archivo en **w+** lo sobrescribe y elimina todos los datos preexistentes. <br>

En el siguiente bloque de código, ejecuta el código tal cual primero y luego ejecútalo sin el <code>.truncate()</code>.


````python
with open('/Example2.txt', 'r+') as testwritefile:
    testwritefile.seek(0,0) # escribir al principio del archivo
    testwritefile.write("ínea 1" + "\n")
    testwritefile.write("ínea 2" + "\n")
    testwritefile.write("ínea 3" + "\n")
    testwritefile.write("ínea 4" + "\n")
    testwritefile.write("terminado\n")
    testwritefile.seek(0,0)
    print(testwritefile.read())
````

Para trabajar con un archivo en datos existentes, usa **r+** y **a+**. Al usar **r+**, puede ser útil agregar un método <code>.truncate()</code> al final de tus datos. Esto reducirá el archivo a tus datos y eliminará todo lo que sigue. <br>


````python
with open('/Example2.txt', 'r+') as testwritefile:
    testwritefile.seek(0,0) # escribir al principio del archivo
    testwritefile.write("ínea 1" + "\n")
    testwritefile.write("ínea 2" + "\n")
    testwritefile.write("ínea 3" + "\n")
    testwritefile.write("ínea 4" + "\n")
    testwritefile.write("terminado\n")
    #Descomentar la línea de abajo para truncar el archivo
    # testwritefile.truncate()
    testwritefile.seek(0,0)
    print(testwritefile.read())
    
````

## Copiar un Archivo 


Copiemos el archivo **Example2.txt** al archivo **Example3.txt**:


````python
# Copiar archivo a otro

with open('/Example2.txt','r') as readfile:
    with open('/Example3.txt','w') as writefile:
          for line in readfile:
                writefile.write(line)
````

Podemos leer el archivo para ver si todo funciona:


````python
# VVerificar si la copia se ejecutó correctamente

with open('/Example3.txt','r') as testwritefile:
    print(testwritefile.read())
````

## Creando una Clase


Ahora vamos a crear una clase Circle, pero primero, vamos a importar una biblioteca para dibujar los objetos:


````python
# Importa la biblioteca matplotlib para gráficos

import matplotlib.pyplot as plt
%matplotlib inline  
````

El primer paso para crear tu propia clase es usar la palabra clave <code>class</code>, luego el nombre de la clase como se muestra en la Figura 4. En este curso, el padre de la clase siempre será object:


<img src="https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/IBMDeveloperSkillsNetwork-PY0101EN-SkillsNetwork/labs/Module%203/images/ClassesDefine.png" width="400">


<i>Figura 4: Creando una clase Circle.</i>


El siguiente paso es un método especial llamado constructor <code>\__init\_\_</code>, que se utiliza para inicializar el objeto. Las entradas son atributos de datos. El término <code>self</code> contiene todos los atributos del conjunto. Por ejemplo, <code>self.color</code> da el valor del atributo color y <code>self.radius</code> te dará el radio del objeto. También tenemos el método <code>add_radius()</code> con el parámetro <code>r</code>, el método suma el valor de <code>r</code> al atributo radio. Para acceder al radio usamos la sintaxis <code>self.radius</code>. La sintaxis etiquetada se resume en la Figura 5:


<img src="https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/IBMDeveloperSkillsNetwork-PY0101EN-SkillsNetwork/labs/Module%203/images/ClassesCircle.png" width="600">


<i>Figura 5: Sintaxis etiquetada del objeto circle.</i>


El objeto real se muestra a continuación. Incluimos el método <code>drawCircle</code> para mostrar la imagen de un círculo. Establecemos el radio predeterminado en 3 y el color predeterminado en azul:


````python
# Crear una clase Circle

class Circle(object):
    
    # Constructor
    def __init__(self, radius=3, color='blue'):
        self.radius = radius
        self.color = color 
    
    # Method
    def add_radius(self, r):
        self.radius = self.radius + r
        return(self.radius)
    
    # Method
    def drawCircle(self):
        plt.gca().add_patch(plt.Circle((0, 0), radius=self.radius, fc=self.color))
        plt.axis('scaled')
        plt.show()  
````

## La Clase Rectangle


Vamos a crear una clase rectangle con los atributos de alto, ancho y color. Solo añadiremos el método para dibujar el objeto rectángulo:


````python
# Crear una nueva clase Rectangle para crear un objeto rectángulo

class Rectangle(object):
    
    # Constructor 
    def __init__(self, width=2, height=3, color='r'):
        self.height = height 
        self.width = width
        self.color = color
    
    # Método para dibujar el rectángulo
    def drawRectangle(self):
        plt.gca().add_patch(plt.Rectangle((0, 0), self.width, self.height ,fc=self.color))
        plt.axis('scaled')
        plt.show()
        
````

Creemos el objeto <code>SkinnyBlueRectangle</code> de tipo Rectangle. Su ancho será 2 y su alto será 3, y el color será azul:


````python
# Crear un nuevo objeto rectángulo

SkinnyBlueRectangle = Rectangle(2, 3, 'blue')
````

Como antes podemos acceder a los atributos de la instancia de la clase usando la notación de punto:


````python
# Imprimir el atributo height del objeto

SkinnyBlueRectangle.height 
````

````python
# Imprimir el atributo width del objeto

SkinnyBlueRectangle.width
````

````python
# Imprimir el atributo color del objeto

SkinnyBlueRectangle.color
````

Podemos dibujar el objeto:

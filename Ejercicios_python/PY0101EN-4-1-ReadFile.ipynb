## Leyendo Archivos de Texto


Una forma de leer o escribir un archivo en Python es usar la función integrada <code>open</code>. La función <code>open</code> proporciona un **Objeto de Archivo (File object)** que contiene los métodos y atributos que necesitas para leer, guardar y manipular el archivo. En este cuaderno, solo cubriremos archivos **.txt**. El primer parámetro que necesitas es la ruta del archivo y el nombre del archivo. Un ejemplo se muestra a continuación:


<img src="https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/IBMDeveloperSkillsNetwork-PY0101EN-SkillsNetwork/labs/Module%204/images/ReadOpen.png" width="500">


El argumento de modo es opcional y el valor predeterminado es **r**. En este cuaderno solo cubrimos dos modos:

<ul>
    <li>**r**: Modo de lectura para leer archivos </li>
    <li>**w**: Modo de escritura para escribir archivos</li>
</ul>


Para el siguiente ejemplo, usaremos el archivo de texto **Example1.txt**. El archivo se muestra a continuación:


<img src="https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/IBMDeveloperSkillsNetwork-PY0101EN-SkillsNetwork/labs/Module%204/images/ReadFile.png" width="100">


Leemos el archivo:


```python
# Leer el archivo Example1.txt
example1 = "example1.txt"
file1 = open(example1, "r")
```

Podemos ver los atributos del archivo.


El nombre del archivo:


```python
# Imprimir la ruta del archivo

file1.name
```

El modo en el que está el objeto de archivo:


```python
# Imprimir el modo del archivo, ya sea 'r' o 'w'

file1.mode
```

Podemos leer el archivo y asignarlo a una variable:


```python
# Leer el archivo

FileContent = file1.read()
FileContent
```

El **/n** significa que hay una nueva línea.


Podemos imprimir el archivo:


```python
# Imprimir el archivo con '\n' como nueva línea

print(FileContent)
```

El archivo es de tipo cadena (string):


```python
# Tipo de contenido del archivo

type(FileContent)
```

Es muy importante que el archivo se cierre al final. Esto libera recursos y asegura la consistencia entre diferentes versiones de Python.


```python
# Cerrar el archivo después de terminar

file1.close()
```

## Una Mejor Manera de Abrir un Archivo


Usar la declaración <code>with</code> es una mejor práctica, cierra automáticamente el archivo incluso si el código encuentra una excepción. El código ejecutará todo en el bloque con sangría y luego cerrará el objeto de archivo.


```python
# Abrir archivo usando with

with open(example1, "r") as file1:
    FileContent = file1.read()
    print(FileContent)
```

El objeto de archivo está cerrado, puedes verificarlo ejecutando la siguiente celda:


```python
# Veificar si el archivo está cerrado

file1.closed
```

Podemos ver la información en el archivo:


```python
# Ver el contenido del archivo

print(FileContent)
```

La sintaxis es un poco confusa ya que el objeto de archivo está después de la declaración <code>as</code>. Tampoco cerramos explícitamente el archivo. Por lo tanto, resumimos los pasos en una figura:


<img src="https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/IBMDeveloperSkillsNetwork-PY0101EN-SkillsNetwork/labs/Module%204/images/ReadWith.png" width="500">


No tenemos que leer todo el archivo, por ejemplo, podemos leer los primeros 4 caracteres ingresando cuatro como parámetro para el método **.read()**:


```python
# Leer los primeros cuatro caracteres

with open(example1, "r") as file1:
    print(file1.read(4))
```

Una vez que se llama al método <code>.read(4)</code>, se llaman los primeros 4 caracteres. Si llamamos al método nuevamente, se llaman los siguientes 4 caracteres. La salida de la siguiente celda demostrará el proceso para diferentes entradas al método <code>read()</code>:


```python
# Leer cierta cantidad de caracteres

with open(example1, "r") as file1:
    print(file1.read(4))
    print(file1.read(4))
    print(file1.read(7))
    print(file1.read(15))
```

El proceso se ilustra en la figura a continuación, y cada color representa la parte del archivo leída después de que se llama al método <code>read()</code>:


<img src="https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/IBMDeveloperSkillsNetwork-PY0101EN-SkillsNetwork/labs/Module%204/images/read.png" width="500">


Aquí hay un ejemplo usando el mismo archivo, pero en su lugar leemos 16, 5 y luego 9 caracteres a la vez:


```python
# Leer cierta cantidad de caracteres

with open(example1, "r") as file1:
    print(file1.read(16))
    print(file1.read(5))
    print(file1.read(9))
```

También podemos leer una línea del archivo a la vez usando el método <code>readline()</code>:


```python
# Leer una línea

with open(example1, "r") as file1:
    print("first line: " + file1.readline())
```

También podemos pasar un argumento a <code> readline() </code> para especificar el número de caracteres que queremos leer. Sin embargo, a diferencia de <code> read()</code>, <code> readline()</code> solo puede leer una línea como máximo.


```python
with open(example1, "r") as file1:
    print(file1.readline(20)) # esto no lee más allá del final de la línea
    print(file1.read(20)) # Devuelve los siguientes 20 caracteres
```

Podemos usar un bucle para iterar a través de cada línea:


```python
# Iterar a través de las líneas

with open(example1,"r") as file1:
        i = 0;
        for line in file1:
            print("Iteration", str(i), ": ", line)
            i = i + 1
```

Podemos usar el método <code>readlines()</code> para guardar el archivo de texto en una lista:


```python
# Leer todas las líneas y guardar como una lista

with open(example1, "r") as file1:
    FileasList = file1.readlines()
```

Cada elemento de la lista corresponde a una línea de texto:


```python
# Imprimir la primera línea

FileasList[0]
```

```python
# Imprimir la segunda línea

FileasList[1]
```

```python
# Imprimir la tercera línea

FileasList[2]
```

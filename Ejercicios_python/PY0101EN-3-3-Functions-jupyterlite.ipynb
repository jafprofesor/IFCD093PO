<h2 id="func">Funciones</h2>


Una función es un bloque de código reutilizable que realiza operaciones especificadas en la función. Te permiten dividir tareas y reutilizar tu código en diferentes programas.

Hay dos tipos de funciones:

*   <b>Funciones predefinidas</b>
*   <b>Funciones definidas por el usuario</b>


<h3 id="content">¿Qué es una Función?</h3>


Puedes definir funciones para proporcionar la funcionalidad requerida. Aquí hay reglas simples para definir una función en Python:

*   Los bloques de funciones comienzan con <code>def</code> seguido del <code>nombre</code> de la función y paréntesis <code>()</code>.
*   Hay parámetros de entrada o argumentos que deben colocarse dentro de estos paréntesis.
*   También puedes definir parámetros dentro de estos paréntesis.
*   Hay un cuerpo dentro de cada función que comienza con dos puntos (<code>:</code>) y tiene sangría.
*   También puedes colocar documentación antes del cuerpo.
*   La declaración <code>return</code> sale de una función, opcionalmente devolviendo un valor.

An example of a function that adds on to the parameter <code>a</code> prints and returns the output as <code>b</code>:


````python
# Primera función de ejemplo: Sumar 1 a a y almacenar como b

def add(a):
    """
    sumar 1 a a
    """
    b = a + 1
    print(a, "si le sumas uno", b)
    return(b)
````

Podemos crear diferentes funciones. Por ejemplo, podemos crear una función que multiplique dos números. Los números estarán representados por las variables <code>a</code> y <code>b</code>:


````python
# Definir una función para multiplicar dos números

def Mult(a, b):
    c = a * b
    return(c)
    print('This is not printed')
    
result = Mult(12,2)
print(result)
````

La misma función se puede utilizar para diferentes tipos de datos. Por ejemplo, podemos multiplicar dos enteros:


````python
# Usar mult() para multiplicar dos enteros

Mult(2, 3)
````

Note how the function terminates at the <code> return </code> statement, while passing back a value. This value can be further assigned to a different variable as desired.

<h3 id="var">Variables</h3>


La entrada a una función se llama parámetro formal.

Una variable que se declara dentro de una función se llama variable local. El parámetro solo existe dentro de la función (es decir, el punto donde comienza y termina la función).

Una variable que se declara fuera de la definición de una función es una variable global, y su valor es accesible y modificable en todo el programa. Discutiremos más sobre variables globales al final del laboratorio.


````python
# Definición de función para calcular el cuadrado de un número más 1

def square(a):
    
    # Variable local b
    b = 1
    c = a * a + b
    print(a, "si elevas al cuadrado y sumas 1", c) 
    return(c)
````

<h4>Bloque 1:</h4>

Podemos reemplazar las líneas de código con una función. Una función combina muchas instrucciones en una sola línea de código. Una vez definida una función, se puede utilizar repetidamente. <h2 id="if">Comparación de cadenas en Funciones</h2>


Los operadores relacionales comparan los valores Unicode de los caracteres de las cadenas desde el índice cero hasta el final de la cadena. Luego devuelve un valor booleano según el operador utilizado.


````python
# Comparar dos cadenas directamente usando el operador in
# agregar cadena
string= "The BodyGuard is the best album"

# Definir una función
def check_string(text):
    
# Usar una declaración if else y el operador 'in' para comparar la cadena
    if text in string:
        return 'Cadena coincidente'
    else:
        return 'Cadena no coincidente'

check_string("The BodyGuard is the best")
````

Este programa utiliza una función definida por el usuario llamada compareStrings() para comparar dos cadenas. 

Esta función recibe ambas cadenas como argumento y devuelve 1 si ambas cadenas son iguales usando el operador ==


````python
# CoComparar dos cadenas usando el operador == y una función
def compareStrings(x, y):
# Usar una declaración if else para comparar x y y
    if x==y:
        return 1
    
# Declarar dos variables diferentes como string1 y string2 y pasar la cadena en ellas
string1 = "The BodyGuard is the best album"
string2 = "The BodyGuard is the best album"

# Declarar una variable para almacenar el resultado después de comparar ambas cadenas
check = compareStrings(string1, string2)

# Usar una declaración if else para comparar la cadena
if check==1:
    print("\nCadena coincidente")
else:
    print("\nCadena no coincidente")
````

**Contar la Frecuencia de Palabras que Aparecen en una Cadena Usando un Diccionario.** 

Encuentra el recuento de ocurrencias de cualquier palabra en nuestra cadena usando python. Esto es lo que vamos a hacer en esta sección, contar el número de palabras en una cadena dada e imprimirlo.

Supongamos que tenemos una ‘cadena’ y la ‘palabra’ y necesitamos encontrar el recuento de ocurrencias de esta palabra en nuestra cadena usando python. Esto es lo que vamos a hacer en esta sección, contar el número de palabras en una cadena dada e imprimirlo. 

Lo primero que haremos es definir una función y luego crear una lista que estará vacía inicialmente.

A continuación, agregaremos un código para convertir la cadena en una lista. La cadena de Python tiene un método split(). Toma una cadena y algún separador para devolver una lista.

Ahora declararemos un diccionario vacío.

A continuación, agregaremos código usando un bucle for para iterar las palabras y el valor contará la frecuencia de cada palabra en la cadena y las almacenará en el diccionario.

Finalmente imprimiremos el diccionario.


````python
# Programa Python para contar palabras en una cadena usando un diccionario
def freq(string):
    
    #paso1: Se declara una variable lista y se inicializa a una lista vacía.
    words = []
    
    #paso2: Dividir la cadena en una lista de palabras
    words = string.split() # o string.lower().split()
    
    #paso3: Declarar un diccionario
    Dict = {}
    
    #paso4: Usar un bucle for para iterar las palabras y agregar valores al diccionario
    for key in words:
        Dict[key] = words.count(key)
        
    #paso5: Imprimir el diccionario
    print("La frecuencia de palabras es:",Dict)
    
#paso6: Llamar a la función y pasarle una cadena
freq("María tenía un corderito Corderito, corderito María tenía un corderito. Su lana era blanca como la nieve Y a donde quiera que María iba María iba, María iba \
A donde quiera que María iba El cordero seguro iba")
````

Hasta ahora, hemos estado creando variables dentro de funciones, pero no hemos discutido variables fuera de la función. Estas se llaman variables globales. <br>
Intentemos ver qué devuelve <code>printer1</code>:


````python
# Ejemplo de variable global

album = "The BodyGuard"
def printer1(album):
    internal_var1 = "Thriller"
    print(album, "es un álbum de éxito mundial.")
    
printer1(album )
# intentar ejecutar el siguiente código
# printer1(internal_var1) 
````

<b>Obtuvimos un Error de Nombre (Name Error): <code>name 'internal_var' is not defined</code>. ¿Por qué?</b>

<h2 id="scope">Alcance de una Variable</h2>


El alcance de una variable es la parte de ese programa donde esa variable es accesible. Las variables que se declaran fuera de todas las definiciones de funciones, como la variable <code>myFavouriteBand</code> en el código que se muestra aquí, son accesibles desde cualquier lugar dentro del programa. Como resultado, se dice que tales variables tienen alcance global y se conocen como variables globales. <code>myFavouriteBand</code> es una variable global, por lo que es accesible desde dentro de la función <code>getBandRating</code>, y podemos usarla para determinar la calificación de una banda. También podemos usarla fuera de la función, como cuando la pasamos a la función de impresión para mostrarla:


````python
# Ejemplo de variable global

myFavouriteBand = "AC/DC"

def getBandRating(bandname):
    if bandname == myFavouriteBand:
        return 10.0
    else:
        return 0.0

print("La calificación de AC/DC es:", getBandRating("AC/DC"))
print("La calificación de Deep Purple es:",getBandRating("Deep Purple"))
print("Mi banda favorita es:", myFavouriteBand)
````

Echa un vistazo a esta versión modificada de nuestro código. Ahora la variable <code>myFavouriteBand</code> está definida dentro de la función <code>getBandRating</code>. Se dice que una variable que se define dentro de una función es una variable local de esa función. Eso significa que solo es accesible desde dentro de la función en la que está definida. Nuestra función <code>getBandRating</code> seguirá funcionando, porque <code>myFavouriteBand</code> todavía está definida dentro de la función. Sin embargo, ya no podemos imprimir <code>myFavouriteBand</code> fuera de nuestra función, porque es una variable local de nuestra función <code>getBandRating</code>; solo está definida dentro de la función <code>getBandRating</code>:


````python
# Borrar la variable "myFavouriteBand" del ejemplo anterior para demostrar un ejemplo de variable local 

del myFavouriteBand

# Ejemplo de variable local
def getBandRating(bandname):
    myFavouriteBand = "AC/DC"
    if bandname == myFavouriteBand:
        return 10.0
    else:
        return 0.0

print("La calificación de AC/DC es:", getBandRating("AC/DC"))
print("La calificación de Deep Purple es:", getBandRating("Deep Purple"))
print("Mi banda favorita es", myFavouriteBand)
````

Finalmente, echa un vistazo a este ejemplo. Ahora tenemos dos definiciones de variables <code>myFavouriteBand</code>. La primera de ellas tiene un alcance global, y la segunda es una variable local dentro de la función <code>getBandRating</code>. Dentro de la función <code>getBandRating</code>, la variable local tiene prioridad. **Deep Purple** recibirá una calificación de 10.0 cuando se pase a la función <code>getBandRating</code>. Sin embargo, fuera de la función <code>getBandRating</code>, la variable local de <code>getBandRating</code> no está definida, por lo que la variable <code>myFavouriteBand</code> que imprimimos es la variable global, que tiene un valor de **AC/DC**:


````python
# Ejemplo de variable global y variable local con el mismo nombre

myFavouriteBand = "AC/DC"

def getBandRating(bandname):
    myFavouriteBand = "Deep Purple"
    if bandname == myFavouriteBand:
        return 10.0
    else:
        return 0.0

print("La calificación de AC/DC es:",getBandRating("AC/DC"))
print("La calificación de Deep Purple es: ",getBandRating("Deep Purple"))
print("Mi banda favorita es:",myFavouriteBand)
````

<h2 id ="collec"> Colecciones y Funciones</h2>


Cuando se desconoce el número de argumentos para una función, todos se pueden empaquetar en una tupla como se muestra:


````python
def printAll(*args): # Todos los argumentos se 'empaquetan' en args, que puede tratarse como una tupla
    print("No of arguments:", len(args)) 
    for argument in args:
        print(argument)
# printAll con 3 argumentos
printAll('Horsefeather','Adonis','Bone')
# printAll con 4 argumentos
printAll('Sidecar','Long Island','Mudslide','Carriage')
````

De manera similar, los argumentos también se pueden empaquetar en un diccionario como se muestra:


````python
def printDictionary(**args):
    for key in args:
        print(key + " : " + args[key])

printDictionary(Country='Canada',Province='Ontario',City='Toronto')
    
````

Las funciones pueden ser increíblemente poderosas y versátiles. Pueden aceptar (y devolver) tipos de datos, objetos e incluso otras funciones como argumentos. Considera el ejemplo a continuación:


````python
def addItems(list):
    list.append("Three")
    list.append("Four")

myList = ["One","Two"]

addItems(myList)

myList
    
````

Observa cómo los cambios realizados en la lista no se limitan al alcance de las funciones. Esto ocurre ya que es la **referencia** de las listas la que se pasa a la función - Cualquier cambio realizado se realiza en la instancia original de la lista. Por lo tanto, se debe tener cuidado al pasar objetos mutables a las funciones.

```python
# Python Program to Count words in a String using Dictionary
def freq(string,passedkey):

    #step1: A list variable is declared and initialized to an empty list.
    words = []

    #step2: Break the string into list of words
    words = string.split() # or string.lower().split()

    #step3: Declare a dictionary
    Dict = {}

    #step4: Use for loop to iterate words and values to the dictionary
    for key in words:
        if(key == passedkey):
            Dict[key] = words.count(key)   
    #step5: Print the dictionary
    print("Total Count:",Dict)

#step6: Call function and pass string in it
freq("Mary had a little lamb Little lamb, little lamb Mary had a little lamb.Its fleece was white as snow And everywhere that Mary went Mary went, Mary went \
Everywhere that Mary went The lamb was sure to go","little")
```

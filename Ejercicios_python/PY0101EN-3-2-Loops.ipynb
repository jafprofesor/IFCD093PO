## Bucles


### Rango (Range)


A veces, es posible que desees repetir una operación dada muchas veces. Ejecuciones repetidas como esta se realizan mediante <b>bucles</b>. Veremos dos tipos de bucles, bucles <code>for</code> y bucles <code>while</code>.

Antes de discutir los bucles, discutamos el objeto <code>range</code>. Es útil pensar en el objeto range como una lista ordenada. Por ahora, veamos el caso más simple. Si quisiéramos generar un objeto que contenga elementos ordenados de 0 a 2, simplemente usamos el siguiente comando:


````python
# Usa el rango de 0 a 2

range(3)
````

### ¿Qué es el bucle <code>for</code>?


El bucle <code>for</code> te permite ejecutar un bloque de código varias veces. Por ejemplo, harías uso de esto si quisieras imprimir cada elemento en una lista.\
Intentemos usar un bucle <code>for</code> para imprimir todos los años presentados en la lista <code>dates</code>:


Esto se puede hacer de la siguiente manera:


````python
# Bucle For ejemplo

dates = [1982,1980,1973]
N = len(dates)

for i in range(N):
    print(dates[i])     
````

El código en la sangría se ejecuta <code>N</code> veces, cada vez el valor de <code>i</code> aumenta en 1 por cada ejecución. La declaración ejecutada es <code>print</code> (imprimir) el valor en la lista en el índice <code>i</code> como se muestra aquí:


<img src="https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/f3bJb12CNd053xJfk1wU2g/LoopsForRangee.gif" width="800">


En este ejemplo podemos imprimir una secuencia de números del 0 al 7:


````python
# Ejemplo de bucle for

for i in range(0, 8):
    print(i)
````

En Python podemos acceder directamente a los elementos en la lista de la siguiente manera:


````python
# Ejemplo de bucle for, recorrer lista

for year in dates:  
    print(year)   
````

Para cada iteración, el valor de la variable <code>year</code> se comporta como el valor de <code>dates\[i]</code> en el primer ejemplo:


<img src="https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/IBMDeveloperSkillsNetwork-PY0101EN-SkillsNetwork/labs/Module%203/images/LoopsForList.gif" width="800">


Podemos cambiar los elementos en una lista:


````python
# Usa un bucle for para cambiar los elementos en la lista

squares = ['red', 'yellow', 'green', 'purple', 'blue']

for i in range(0, 5):
    print("Antes el cuadrado ", i, 'es',  squares[i])
    squares[i] = 'white'
    print("Después el cuadrado ", i, 'es',  squares[i])
````

Podemos acceder al índice y a los elementos de una lista de la siguiente manera:


````python
# Bucle a través de la lista e itera tanto en el índice como en el valor del elemento   

squares=['red', 'yellow', 'green', 'purple', 'blue']

for i, square in enumerate(squares):
    print(i, square)
````

### ¿Qué es el bucle <code>while</code>?


Como puedes ver, el bucle <code>for</code> se utiliza para un flujo controlado de repetición. Sin embargo, ¿qué pasa si no sabemos cuándo queremos detener el bucle? ¿Qué pasa si queremos seguir ejecutando un bloque de código hasta que se cumpla una determinada condición? El bucle <code>while</code> existe como una herramienta para la ejecución repetida basada en una condición. El bloque de código se seguirá ejecutando hasta que la condición lógica dada devuelva un valor booleano **False**.


Digamos que nos gustaría iterar a través de la lista <code>dates</code> y detenernos en el año 1973, luego imprimir el número de iteraciones. Esto se puede hacer con el siguiente bloque de código:


````python
# Ejemplo de bucle While

dates = [1982, 1980, 1973, 2000]

i = 0
year = dates[0]

while(year != 1973):    
    print(year)
    i = i + 1
    year = dates[i]
    

print("Tomó ", i ,"repeticiones para salir del bucle.")
````

Un bucle while itera simplemente hasta que no se cumple la condición en el argumento, como se muestra en la siguiente figura:

## Controlando la Ejecución del Bucle con <code>break</code> y <code>continue</code>


### La declaración <code>break</code>


La declaración <code>break</code> termina inmediatamente el bucle en el que aparece, independientemente de la condición del bucle.
Se usa comúnmente cuando se cumple una condición específica y no se necesitan más iteraciones.


Ejemplo 1: Usando <code>break</code> en un bucle <code>for</code>


````python
for num in range(1, 10):
    if num == 5:
        print("Saliendo del bucle en:", num)
        break
    print(num)
````

Nota: El bucle se detiene tan pronto como num es igual a 5.


### La declaración <code>continue</code>


La declaración <code>continue</code> omite la iteración actual y pasa a la siguiente en el bucle.
Se usa cuando quieres ignorar ciertas condiciones pero continuar con el bucle.


Ejemplo 2: Usando <code>continue</code> en un bucle <code>for</code>


````python
for num in range(1, 6):
    if num == 3:
        continue
    print(num)
````

Nota: El número 3 se omite, pero el bucle continúa con las iteraciones restantes.


Ejemplo 3: Usando <code>break</code> y <code>continue</code> en un bucle <code>while</code>


````python
count = 0
while count < 10:
    count += 1
    if count == 3:
        continue  # saltarse imprimir 3
    if count == 8:
        break     # detener el bucle cuando count es 8
    print(count)
````

Nota: Aquí, el 3 se omite debido a <code>continue</code>, y el bucle se detiene cuando count llega a 8 debido a <code>break</code>.

### Operadores de Comparación


Las operaciones de comparación comparan algún valor u operando y, según una condición, producen un valor booleano. Al comparar dos valores puedes usar estos operadores:

<ul>
    <li>igual: <b>==</b></li>
    <li>no es igual: <b>!=</b></li>
    <li>mayor que: <b>></b></li>
    <li>menor que: <b>&lt;</b></li>
    <li>mayor o igual que: <b>>=</b></li>
    <li>menor o igual que: <b>&lt;=</b></li>
</ul>


Asignemos a <code> a </code> un valor de 5. Usa el operador de igualdad denotado con dos signos de igual <b>==</b> para determinar si dos valores son iguales. El caso a continuación compara la variable <code>a</code> con 6.


````python
# Condición Igualdad

a = 5
a == 6
````

El resultado es <b>False</b> (Falso), ya que 5 no es igual a 6.


Considera el siguiente operador de comparación de igualdad: <code>i > 5</code>. Si el valor del operando izquierdo, en este caso la variable <b>i</b>, es mayor que el valor del operando derecho, en este caso 5, entonces la declaración es <b>True</b> (Verdadera). De lo contrario, la declaración es <b>False</b>. Si <b>i</b> es igual a 6, porque 6 es mayor que 5, la salida es <b>True</b>.


````python
# Mayor que Signo

i = 6
i > 5
````

Establece <code>i = 2</code>. La declaración es Falsa ya que 2 no es mayor que 5:


````python
# MMayor que Signo

i = 2
i > 5
````

Mostremos algunos valores para <code>i</code> en la figura. Establece los valores mayores que 5 en verde y el resto en rojo. La región verde representa dónde la condición es **True**, la roja dónde la declaración es **False**. Si el valor de <code>i</code> es 2, obtenemos **False** ya que el 2 cae en la región roja. De manera similar, si el valor para <code>i</code> es 6 obtenemos un **True** ya que la condición cae en la región verde.


<img src="https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/IBMDeveloperSkillsNetwork-PY0101EN-SkillsNetwork/labs/Module%203/images/CondsGreater.gif" width="650">


La prueba de desigualdad utiliza un signo de exclamación que precede al signo igual, si dos operandos no son iguales entonces la condición se vuelve **True**. Por ejemplo, la siguiente condición producirá **True** siempre que el valor de <code>i</code> no sea igual a 6:


````python
# Desigualdad Signo

i = 2
i != 6
````

Cuando <code>i</code> es igual a 6 la expresión de desigualdad produce <b>False</b>.


````python
# Desigualdad Signo

i = 6
i != 6
````

Mira la recta numérica a continuación. Cuando la condición es **True**, los números correspondientes están marcados en verde y donde la condición es **False** el número correspondiente está marcado en rojo. Si establecemos <code>i</code> igual a 2, el operador es verdadero, ya que 2 está en la región verde. Si establecemos <code>i</code> igual a 6, obtenemos un **False**, ya que la condición cae en la región roja.


<img src="https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/IBMDeveloperSkillsNetwork-PY0101EN-SkillsNetwork/labs/Module%203/images/CondsIneq.gif" width="650">


Podemos aplicar los mismos métodos en cadenas. Por ejemplo, podemos usar un operador de igualdad en dos cadenas diferentes. Como las cadenas no son iguales, obtenemos un **False**.


````python
# Usa el signo de doble igualdad para comparar cadenas

"ACDC" == "The Bodyguard"
````

Si usamos el operador de desigualdad, la salida será **True** ya que las cadenas no son iguales.


````python
# Usa el signo de desigualdad para comparar cadenas

"ACDC" != "The Bodyguard"
````

### Ramificación


La ramificación nos permite ejecutar diferentes declaraciones para diferentes entradas. Es útil pensar en una **declaración if** como una habitación cerrada, si la declaración es **True** podemos entrar a la habitación y su programa ejecutará algunas tareas predefinidas, pero si la declaración es **False** el programa ignorará la tarea.


Por ejemplo, considera el rectángulo azul que representa un concierto de ACDC. Si el individuo es mayor de 18 años, puede entrar al concierto de ACDC. Si tiene 18 años o menos, no puede entrar al concierto.

Podemos usar las declaraciones de condición aprendidas antes como las condiciones que deben verificarse en la **declaración if**. La sintaxis es tan simple como <code> if <i>declaración de condición</i> :</code>, que contiene una palabra <code>if</code>, cualquier declaración de condición y dos puntos al final. Comienza tus tareas que deben ejecutarse bajo esta condición en una nueva línea con una sangría. Las líneas de código después de los dos puntos y con sangría solo se ejecutarán cuando la **declaración if** sea **True**. Las tareas terminarán cuando la línea de código no contenga la sangría.

En el caso a continuación, el código <code>print(“you can enter”)</code> se ejecuta solo si la variable <code>age</code> es mayor que 18, es un caso Verdadero porque esta línea de código tiene la sangría. Sin embargo, la ejecución de <code>print(“move on”)</code> no se verá influenciada por la declaración if.


````python
# Sentencia If ejemplo

age = 19
#age = 18

# expresión que puede ser verdadera o falsa
if age > 18:
    
    # dentro de una sangría, tenemos la expresión que se ejecuta si la condición es verdadera
    print("you can enter" )

# Las sentencias después de la sentencia if se ejecutarán independientemente de si la condición es verdadera o falsa 
print("move on")
````

<i>Intenta descomentar la variable age</i>


Es útil usar el siguiente diagrama para ilustrar el proceso. En el lado izquierdo, vemos qué sucede cuando la condición es <b>True</b>. La persona entra al concierto de ACDC representando el código en la sangría que se está ejecutando; luego siguen adelante. En el lado derecho, vemos qué sucede cuando la condición es <b>False</b>; a la persona no se le concede acceso y la persona sigue adelante. En este caso, el segmento de código en la sangría no se ejecuta, pero el resto de las declaraciones sí.


<img src="https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/IBMDeveloperSkillsNetwork-PY0101EN-SkillsNetwork/labs/Module%203/images/CondsIf.gif" width="650">


La declaración <code>else</code> ejecuta un bloque de código si ninguna de las condiciones es **True** antes de esta declaración <code>else</code>. Usemos la analogía del concierto de ACDC nuevamente. Si el usuario tiene 17 años no puede ir al concierto de ACDC, pero puede ir al concierto de Meatloaf.
La sintaxis de la declaración <code>else</code> es similar a la sintaxis de la declaración <code>if</code>, como <code>else :</code>. Ten en cuenta que no hay una declaración de condición para <code>else</code>.
Intenta cambiar los valores de <code>age</code> para ver qué sucede:


````python
# Sentencia Else ejemplo

age = 18
# age = 19

if age > 18:
    print("puedes entrar" )
else:
    print("ve a ver a Meat Loaf" )
    
print("continúa")
````

El proceso se demuestra a continuación, donde cada una de las posibilidades se ilustra en cada lado de la imagen. A la izquierda está el caso donde la edad es 17, establecemos la variable edad en 17, y esto corresponde al individuo que asiste al concierto de Meatloaf. La parte derecha muestra qué sucede cuando el individuo tiene más de 18 años, en este caso 19, y al individuo se le concede acceso al concierto.

La declaración <code>elif</code>, abreviatura de else if (sino si), nos permite verificar condiciones adicionales si las declaraciones de condición anteriores son <b>False</b>. Si la condición para la declaración <code>elif</code> es <b>True</b>, se ejecutarán las expresiones alternativas. Considera el ejemplo del concierto, donde si el individuo tiene 18 años irá al concierto de Pink Floyd en lugar de asistir al concierto de ACDC o Meat-loaf. Una persona que tiene 18 años entra al área, y como no es mayor de 18 no puede ver a ACDC, pero como tiene 18 años, asiste a Pink Floyd. Después de ver a Pink Floyd, siguen adelante. La sintaxis de la declaración <code>elif</code> es similar en que simplemente cambiamos el <code>if</code> en la declaración <code>if</code> a <code>elif</code>.


````python
# Sentencia Elif ejemplo

age = 18

if age > 18:
    print("puedes entrar" )
elif age == 18:
    print("ve a ver a Pink Floyd")
else:
    print("ve a ver a Meat Loaf" )
    
print("continúa")
````

Las tres combinaciones se muestran en la figura a continuación. La región más a la izquierda muestra qué sucede cuando el individuo tiene menos de 18 años. El componente central muestra cuando el individuo tiene exactamente 18 años. El más a la derecha muestra cuando el individuo tiene más de 18 años.

Como antes, podemos agregar un bloque <code>else</code> al bloque <code>if</code>. El código en el bloque <code>else</code> solo se ejecutará si el resultado es <b>False</b>.

### Operadores Lógicos


A veces quieres verificar más de una condición a la vez. Por ejemplo, es posible que desees verificar si una condición y otra condición son ambas **True**. Los operadores lógicos te permiten combinar o modificar condiciones.

<ul>
    <li><code>and</code></li>
    <li><code>or</code></li>
    <li><code>not</code></li>
</ul>

Estos operadores se resumen para dos variables utilizando las siguientes tablas de verdad:


<img src="https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/IBMDeveloperSkillsNetwork-PY0101EN-SkillsNetwork/labs/Module%203/images/CondsTable.png" width="650">


La declaración <code>and</code> es solo **True** cuando ambas condiciones son verdaderas. La declaración <code>or</code> es Verdadera si una condición, o ambas son **True**. La declaración <code>not</code> genera el valor de verdad opuesto.


Veamos cómo determinar si un álbum fue lanzado después de 1979 (1979 no está incluido) y antes de 1990 (1990 no está incluido). Los períodos de tiempo entre 1980 y 1989 satisfacen esta condición. Esto se demuestra en la figura a continuación. El verde en las líneas <strong>a</strong> y <strong>b</strong> representa períodos donde la declaración es **True**. El verde en la línea <strong>c</strong> representa donde ambas condiciones son **True**, esto corresponde a donde se superponen las regiones verdes.


<img src="https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/IBMDeveloperSkillsNetwork-PY0101EN-SkillsNetwork/labs/Module%203/images/CondsEgOne.png" width="650">


El bloque de código para realizar esta verificación viene dado por:


````python
# Sentencia Condicional ejemplo

album_year = 1980

if(album_year > 1979) and (album_year < 1990):
    print ("El año del álbum fue entre 1980 y 1989")
    
print("")
print("Haz algo..")
````

Para determinar si un álbum fue lanzado antes de 1980 (1979 y anteriores) o después de 1989 (1990 en adelante), se puede usar una declaración or. Los períodos antes de 1980 (1979 y anteriores) o después de 1989 (1990 y posteriores) satisfacen esta condición. Esto se demuestra en la siguiente figura, el color verde en <strong>a</strong> y <strong>b</strong> representa períodos donde la declaración es verdadera. El color verde en **c** representa donde al menos una de las condiciones es verdadera.


<img src="https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/IBMDeveloperSkillsNetwork-PY0101EN-SkillsNetwork/labs/Module%203/images/CondsEgTwo.png" width="650">


El bloque de código para realizar esta verificación viene dado por:


````python
# Sentencia Condicional ejemplo

album_year = 1990

if(album_year < 1980) or (album_year > 1989):
    print ("El álbum no fue hecho en los años 80")
else:
    print("El álbum fue hecho en los años 80")
````

La declaración <code>not</code> verifica si la declaración es falsa:


````python
# Sentencia Condicional ejemplo

album_year = 1983

if not (album_year == 1984):
    print ("El año del álbum no es 1984")
````

<details><summary>Haz clic aquí para ver la solución</summary>

```python
player_name = "Roger Federer"
sport = "Tennis"
achievements = 20

if sport == "Tennis" or achievements == 20:
    print(f"{player_name} meets the criteria! They play {sport} and have {achievements} achievements.")
else:
    print(f"{player_name} does not meet the criteria.")
```

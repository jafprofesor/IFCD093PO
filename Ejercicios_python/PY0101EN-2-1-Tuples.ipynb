<h2 id="tuple">Tuplas</h2>


En Python, hay diferentes tipos de datos: Cadena, Entero y Flotante. Todos estos tipos de datos pueden estar contenidos en una tupla de la siguiente manera:


<img src="https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/IBMDeveloperSkillsNetwork-PY0101EN-SkillsNetwork/labs/Module%202/images/TuplesType.png" width="750" align="center">


Ahora, creemos tu primera tupla con cadena, entero y flotante.


````python
# Crear tu primera tupla

tuple1 = ("disco",10,1.2 )
tuple1
````

El tipo de variable es una **tupla**.


````python
# Imprime el tipo de la tupla que creaste

type(tuple1)
````

<h3 id="index">Indexación</h3>


Cada elemento de una tupla se puede acceder a través de un índice. La siguiente tabla representa la relación entre el índice y los elementos en la tupla. Cada elemento se puede obtener mediante el nombre de la tupla seguido de un corchete con el número de índice:


<img src="https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/IBMDeveloperSkillsNetwork-PY0101EN-SkillsNetwork/labs/Module%202/images/TuplesIndex.gif" width="750" align="center">


Podemos imprimir cada valor en la tupla:


````python
# Imprime la variable en cada índice

print(tuple1[0])
print(tuple1[1])
print(tuple1[2])
````

Podemos imprimir el **tipo** de cada valor en la tupla:


````python
# Imprime el tipo de valor en cada índice

print(type(tuple1[0]))
print(type(tuple1[1]))
print(type(tuple1[2]))
````

También podemos usar indexación negativa. Usamos la misma tabla de arriba con los valores negativos correspondientes:


<img src="https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/IBMDeveloperSkillsNetwork-PY0101EN-SkillsNetwork/labs/Module%202/images/TuplesNeg.png" width="750" align="center">


Podemos obtener el último elemento de la siguiente manera (esta vez no usaremos la declaración print para mostrar los valores):


````python
# Usa índice negativo para obtener el valor del último elemento

tuple1[-1]
````

Podemos mostrar los siguientes dos elementos de la siguiente manera:


````python
# Usa índice negativo para obtener el valor del penúltimo elemento

tuple1[-2]
````

````python
# Usa índice negativo para obtener el valor del antepenúltimo elemento

tuple1[-3]
````

<h3 id="concate">Concatenar Tuplas</h3>


Podemos concatenar o combinar tuplas usando el signo **+**:


````python
# Concatenar dos tuplas

tuple2 = tuple1 + ("hard rock", 10)
tuple2
````

Podemos segmentar tuplas obteniendo múltiples valores como se demuestra en la figura a continuación:

<h3 id="slice">Segmentación (Slicing)</h3>


Podemos segmentar tuplas, obteniendo nuevas tuplas con los elementos correspondientes:


````python
# Cortar desde el índice 0 al índice 2
# tupla2 = ('disco', 10, 1.2, 'hard rock', 10)
tuple2[0:3]
````

Podemos obtener los últimos dos elementos de la tupla:


````python
# Cortar desde el índice 3 al 4

tuple2[3:5]
````

Podemos obtener la longitud de una tupla usando el comando length:


````python
# Obtener la longitud de la tupla

len(tuple2)
````

Esta figura muestra el número de elementos:

<h3 id="sort">Ordenamiento</h3>


Considera la siguiente tupla:


````python
# Una tupla de ejemplo

Ratings = (0, 9, 6, 5, 10, 8, 9, 6, 2)
Ratings
````

Podemos ordenar los valores en una tupla y guardarlos en un nuevo objeto. La salida es una nueva lista ordenada.


````python
# Ordenar la tupla

RatingsSorted = sorted(Ratings)
RatingsSorted
````

<h3 id="nest">Tupla Anidada</h3>


Una tupla puede contener otra tupla, así como otros tipos de datos más complejos. Este proceso se llama 'anidamiento'. Considera la siguiente tupla con varios elementos:


````python
# Crear una tupla anidada

NestedT =(1, 2, ("pop", "rock") ,(3,4),("disco",(1,2)))
````

Cada elemento en la tupla, incluyendo otras tuplas, se puede obtener a través de un índice como se muestra en la figura:


<img src="https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/IBMDeveloperSkillsNetwork-PY0101EN-SkillsNetwork/labs/Module%202/images/TuplesNestOne.png" width="750" align="center">


````python
# Imprimir elemento en cada índice

print("Elemento 0 de la tupla: ", NestedT[0])
print("Elemento 1 de la tupla: ", NestedT[1])
print("Elemento 2 de la tupla: ", NestedT[2])
print("Elemento 3 de la tupla: ", NestedT[3])
print("Elemento 4 de la tupla: ", NestedT[4])
````

Podemos usar el segundo índice para acceder a otras tuplas como se demuestra en la figura:


<img src="https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/IBMDeveloperSkillsNetwork-PY0101EN-SkillsNetwork/labs/Module%202/images/TuplesNestTwo.png" width="750" align="center">


Podemos acceder a las tuplas anidadas:


````python
# Imprimir elemento en cada índice, incluyendo índices anidados

print("Elemento 2, 0 de la tupla: ",   NestedT[2][0])
print("Elemento 2, 1 de la tupla: ",   NestedT[2][1])
print("Elemento 3, 0 de la tupla: ",   NestedT[3][0])
print("Elemento 3, 1 de la tupla: ",   NestedT[3][1])
print("Elemento 4, 0 de la tupla: ",   NestedT[4][0])
print("Elemento 4, 1 de la tupla: ",   NestedT[4][1])
````

Podemos acceder a las cadenas en las segundas tuplas anidadas usando un tercer índice:


````python
# Imprimir el primer elemento en la segunda tupla anidada

NestedT[4][1][1]
````

````python
# Imprimir el segundo elemento en la segunda tupla anidada

NestedT[2][1][1]
````

Podemos usar un árbol para visualizar el proceso. Cada nuevo índice corresponde a un nivel más profundo en el árbol:


<img src="https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/IBMDeveloperSkillsNetwork-PY0101EN-SkillsNetwork/labs/Module%202/images/TuplesNestThree.gif" width="750" align="center">


De manera similar, podemos acceder a elementos anidados más profundamente en el árbol con un tercer índice:


````python
# Imprimir el primer elemento en la segunda tupla anidada

NestedT[4][1][0]
````

````python
# Imprimir el segundo elemento en la segunda tupla anidada

NestedT[4][1][1]
````

La siguiente figura muestra la relación del árbol y el elemento <code>NestedT\[4]\[1]\[1]</code>:


<img src="https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/IBMDeveloperSkillsNetwork-PY0101EN-SkillsNetwork/labs/Module%202/images/TuplesNestFour.gif" width="750" align="center">

## Expresiones y Variables


### Expresiones


<p>Las expresiones en Python pueden incluir operaciones entre tipos compatibles (por ejemplo, enteros y flotantes). Por ejemplo, operaciones aritméticas básicas como sumar varios números:</p>


````python
# Suma operación expresión

43 + 60 + 16 + 41
````

<p>Podemos realizar operaciones de resta utilizando el operador menos. En este caso el resultado es un número negativo:</p>


````python
# Sustracción operación expresión

50 - 60
````

<p>Podemos hacer multiplicaciones usando un asterisco:</p>


````python
# Multiplicación operación expresión

5 * 5
````

<p>También podemos realizar divisiones con la barra inclinada:


````python
# División operación expresión

25 / 5
````

````python
# División operación expresión

25 / 6
````

<p>Como se vio en el cuestionario anterior, podemos usar la doble barra para la división entera, donde el resultado se redondea hacia abajo al entero más cercano:


````python
# División entera operación expresión

25 // 5
````

````python
# División entera operación expresión

25 // 6
````

<p>Escribamos una expresión que calcule cuántas horas hay en 160 minutos:


````python
# Escribe tu código abajo. No olvides presionar Shift+Enter para ejecutar la celda
160//60 
````

<p>Python sigue convenciones matemáticas bien aceptadas al evaluar expresiones matemáticas. En el siguiente ejemplo, Python suma 30 al resultado de la multiplicación (es decir, 120).


````python
# Expresión matemática

30 + 2 * 60
````

<p>Y al igual que en matemáticas, las expresiones encerradas en paréntesis tienen prioridad. Así que lo siguiente multiplica 32 por 60.


````python
# Expresión matemática

(30 + 2) * 60
````

### Variables


<p>Al igual que con la mayoría de los lenguajes de programación, podemos almacenar valores en <i>variables</i>, para poder usarlos más tarde. Por ejemplo:</p>


````python
# Almacenar valor en variable

x = 43 + 60 + 16 + 41
````

<p>Para ver el valor de <code>x</code> en un Notebook, simplemente podemos colocarlo en la última línea de una celda:</p>


````python
# Imprimir el valor de la variable

x
````

<p>También podemos realizar operaciones sobre <code>x</code> y guardar el resultado en una nueva variable:</p>


````python
# Usar otra variable para almacenar el resultado de la operación entre variable y valor

y = x / 60
y
````

<p>Si guardamos un valor en una variable existente, el nuevo valor sobrescribirá el valor anterior:</p>


````python
# Sobrescribir variable con nuevo valor

x = x / 60
x
````

<p>Es una buena práctica usar nombres de variables significativos, para que tú y otros puedan leer el código y entenderlo más fácilmente:</p>


````python
# Nombrar las variables de manera significativa

total_min = 43 + 42 + 57 # Total longitud de álbumes en minutos
total_min
````

````python
# Nombrar las variables de manera significativa

total_hours = total_min / 60 # Total longitud de álbumes en horas 
total_hours
````

<p>En las celdas anteriores sumamos la duración de tres álbumes en minutos y la almacenamos en <code>total_min</code>. Luego la dividimos por 60 para calcular la duración total <code>total_hours</code> en horas. También puedes hacerlo todo de una vez en una sola expresión, siempre y cuando uses paréntesis para sumar la duración de los álbumes antes de dividir, como se muestra a continuación.</p>


````python
# Expresión complicada

total_hours = (43 + 42 + 57) / 60  # Total de horas en una sola expresión
total_hours
````

<p>Si prefieres tener el total de horas como un entero, puedes reemplazar la división de punto flotante con la división entera (es decir, <code>//</code>).</p>

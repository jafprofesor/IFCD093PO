## Tipos de objetos en Python


<p>Python es un lenguaje orientado a objetos. Hay muchos tipos diferentes de objetos en Python. Comencemos con los tipos de objetos más comunes: <i>cadenas (strings)</i>, <i>enteros (integers)</i> y <i>flotantes (floats)</i>. Cada vez que escribes palabras (texto) en Python, estás usando <i>cadenas de caracteres</i> (o cadenas para abreviar). Los números más comunes, por otro lado, son <i>enteros</i> (por ejemplo, -1, 0, 100) y <i>flotantes</i>, que representan números reales (por ejemplo, 3.14, -42.0).</p>

<p>Puedes hacer que Python te diga el tipo de una expresión usando la función integrada <code>type()</code>. Notarás que Python se refiere a los enteros como <code>int</code>, a los flotantes como <code>float</code> y a las cadenas de caracteres como <code>str</code>.</p>


````python
# Type of 12, muestra el tipo de dato

type(12)
````

````python
# Type of 2.14, muestra el tipo de dato

type(2.14)
````

````python
# Type of "Hello, Python 101!", muestra el tipo de dato

type("Hello, Python 101!")
````

### Flotantes


<p>Los flotantes representan números reales; son un superconjunto de números enteros pero también incluyen "números con decimales". Existen algunas limitaciones cuando se trata de máquinas que representan números reales, pero los números de punto flotante son una buena representación en la mayoría de los casos. Puedes obtener más información sobre los detalles de los flotantes para tu entorno de ejecución comprobando el valor de <code>sys.float_info</code>. Esto también te dirá cuál es el número más grande y más pequeño que se puede representar con ellos.</p>

### Convertir de un tipo de objeto a un tipo de objeto diferente


<p>Puedes cambiar el tipo del objeto en Python; esto se llama casting o conversión de tipos. Por ejemplo, puedes convertir un <i>entero</i> en un <i>flotante</i> (por ejemplo, 2 a 2.0).</p>
<p>Vamos a intentarlo:</p>


````python
# Verifica que esto es un entero

type(2)
````

#### Convirtiendo enteros a flotantes
<p>Vamos a convertir el entero 2 a flotante:</p>


````python
# Convierta 2 a un float

float(2)
````

````python
# Convertir entero 2 a un float y verificar su tipo

type(float(2))
````

<p>Cuando convertimos un entero en un flotante, no cambiamos realmente el valor (es decir, el significando) del número. Sin embargo, si convertimos un flotante en un entero, podríamos perder algo de información potencialmente. Por ejemplo, si convertimos el flotante 1.1 a entero obtendremos 1 y perderemos la información decimal (es decir, 0.1):</p>


````python
# Convertir 1.1 a entero resultará en pérdida de información

int(1.1)
````

<h4>Convirtiendo de cadenas a enteros o flotantes</h4>


<p>A veces, podemos tener una cadena que contiene un número dentro de ella. Si este es el caso, podemos convertir esa cadena que representa un número en un entero usando <code>int()</code>:</p>


````python
# Convierta una cadena en un entero

int('1')
````

<p>Pero si intentas hacerlo con una cadena que no es una coincidencia perfecta para un número, obtendrás un error. Intenta lo siguiente:</p>


````python
# Convertir una cadena en un entero con error

int('1 or 2 people')
````

<p>También puedes convertir cadenas que contienen números de punto flotante en objetos <i>float</i>:</p>


````python
# Convertir la cadena "1.2" en un float

float('1.2')
````

<p>Si podemos convertir cadenas a números, es natural suponer que podemos convertir números a cadenas, ¿verdad?</p>

````python
#  Convertir un float a una cadena

str(1.2)
````

### Tipo de dato Booleano


<p><i>Booleano</i> es otro tipo importante en Python. Un objeto de tipo <i>Booleano</i> puede tomar uno de dos valores: <code>True</code> o <code>False</code>:</p>


````python
# Valor verdadero

True
````

<p>Observa que el valor <code>True</code> tiene una "T" mayúscula. Lo mismo ocurre con <code>False</code> (es decir, debes usar la "F" mayúscula).</p>


````python
# Valor falso

False
````

<p>Cuando le pides a Python que muestre el tipo de un objeto booleano, mostrará <code>bool</code> que significa <i>booleano</i>:</p> 


````python
# Tipo de True

type(True)
````

````python
# Tipo de False

type(False)
````

<p>Podemos convertir objetos booleanos a otros tipos de datos. Si convertimos un booleano con un valor de <code>True</code> a un entero o flotante obtendremos un uno. Si convertimos un booleano con un valor de <code>False</code> a un entero o flotante obtendremos un cero. De manera similar, si convertimos un 1 a un Booleano, obtienes un <code>True</code>. Y si convertimos un 0 a un Booleano obtendremos un <code>False</code>. Vamos a intentarlo:</p> 


````python
# Convertir True a int

int(True)
````

````python
# Convertir 1 a booleano

bool(1)
````

````python
# Convertir 0 a booleano

bool(0)
````

````python
# Convertir True a float

float(True)
````
